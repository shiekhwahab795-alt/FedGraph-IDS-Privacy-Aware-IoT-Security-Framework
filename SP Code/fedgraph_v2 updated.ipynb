```python
# ============================================================
# CELL 1 — Mount Google Drive and Inspect Dataset Structure
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

import os
from pathlib import Path
import pandas as pd

CIC_PATH = Path("/content/drive/MyDrive/CICIoT2023")
BOT_PATH = Path("/content/drive/MyDrive/IDS dataset")

print("=" * 80)
print("EFG-IDS DATASET INSPECTION")
print("=" * 80)

print(f"\nCIC-IoT-2023 path:")
print(CIC_PATH)
print("Exists:", CIC_PATH.exists())

print(f"\nBoT/IDS dataset path:")
print(BOT_PATH)
print("Exists:", BOT_PATH.exists())


def inspect_folder(folder, name):
    print("\n" + "=" * 80)
    print(f"{name}")
    print("=" * 80)

    if not folder.exists():
        print("ERROR: Folder does not exist.")
        return

    files = []

    for root, dirs, filenames in os.walk(folder):
        for filename in filenames:
            path = Path(root) / filename
            files.append(path)

    print(f"Total files found: {len(files)}")

    extensions = {}

    for f in files:
        ext = f.suffix.lower()
        extensions[ext] = extensions.get(ext, 0) + 1

    print("\nFile types:")
    for ext, count in sorted(extensions.items()):
        print(f"  {ext or '[no extension]':15s} : {count}")

    print("\nFirst 50 files:")

    for f in files[:50]:
        size_mb = f.stat().st_size / (1024 ** 2)

        print(
            f"{str(f.relative_to(folder)):<80} "
            f"{size_mb:>10.2f} MB"
        )

    return files


cic_files = inspect_folder(
    CIC_PATH,
    "CIC-IoT-2023"
)

bot_files = inspect_folder(
    BOT_PATH,
    "IDS DATASET / BoT-IoT"
)
```

```python
# ============================================================
# CELL 2 — Automatically Inspect CSV Dataset Schemas
# ============================================================

from pathlib import Path
import pandas as pd
import os

def find_tabular_files(folder):
    extensions = {
        ".csv",
        ".csv.gz",
        ".parquet",
        ".parq"
    }

    files = []

    for root, dirs, filenames in os.walk(folder):
        for filename in filenames:
            path = Path(root) / filename

            if path.suffix.lower() in extensions:
                files.append(path)

    return files


def inspect_csv_file(path, nrows=5):

    print("\n" + "=" * 100)
    print("FILE:")
    print(path)
    print("=" * 100)

    try:

        if str(path).endswith(".parquet"):
            df = pd.read_parquet(path)
        else:
            df = pd.read_csv(
                path,
                nrows=nrows,
                low_memory=False
            )

        print("Shape preview:", df.shape)

        print("\nColumns:")
        for i, col in enumerate(df.columns):
            print(f"{i:3d} | {col}")

        print("\nData types:")
        print(df.dtypes)

        print("\nFirst rows:")
        display(df.head())

        print("\nPotential label columns:")

        label_candidates = [
            c for c in df.columns
            if any(
                word in c.lower()
                for word in [
                    "label",
                    "attack",
                    "class",
                    "category",
                    "target"
                ]
            )
        ]

        for c in label_candidates:
            print(f"  {c}")

    except Exception as e:
        print("Could not inspect file:")
        print(e)


cic_tabular = find_tabular_files(CIC_PATH)
bot_tabular = find_tabular_files(BOT_PATH)

print("\nCIC-IoT-2023 tabular files:", len(cic_tabular))

for f in cic_tabular[:10]:
    print(f)

print("\nIDS/BoT tabular files:", len(bot_tabular))

for f in bot_tabular[:10]:
    print(f)


# Inspect a few files only
print("\n\n========== CIC-IoT-2023 SAMPLE FILES ==========")

for f in cic_tabular[:3]:
    inspect_csv_file(f)


print("\n\n========== IDS / BoT-IoT SAMPLE FILES ==========")

for f in bot_tabular[:3]:
    inspect_csv_file(f)
```

```python
# ============================================================
# CELL 7 — Compare CIC-IoT-2023 and IDS Feature Spaces
# ============================================================

import pandas as pd
from pathlib import Path
import os

CIC_PATH = Path("/content/drive/MyDrive/CICIoT2023")
IDS_PATH = Path("/content/drive/MyDrive/IDS dataset")

# ------------------------------------------------------------
# 1. Get CIC-IoT-2023 columns
# ------------------------------------------------------------

cic_file = sorted(CIC_PATH.glob("*.csv"))[0]

cic_sample = pd.read_csv(
    cic_file,
    nrows=5,
    low_memory=False
)

cic_columns = list(cic_sample.columns)

print("=" * 90)
print("CIC-IoT-2023 FEATURE SPACE")
print("=" * 90)

print("Number of columns:", len(cic_columns))

for i, col in enumerate(cic_columns):
    print(f"{i:02d}: {col}")


# ------------------------------------------------------------
# 2. Get IDS columns from all files
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("IDS FEATURE SPACE")
print("=" * 90)

ids_files = sorted(IDS_PATH.glob("*.csv"))

ids_schemas = {}

for file in ids_files:

    try:

        df = pd.read_csv(
            file,
            nrows=2,
            low_memory=False
        )

        ids_schemas[file.name] = list(df.columns)

    except Exception as e:

        print("ERROR:", file.name, repr(e))


# ------------------------------------------------------------
# 3. Check whether all IDS files have the same schema
# ------------------------------------------------------------

unique_schemas = {}

for filename, columns in ids_schemas.items():

    key = tuple(columns)

    if key not in unique_schemas:
        unique_schemas[key] = []

    unique_schemas[key].append(filename)


print("\nNumber of unique IDS schemas:", len(unique_schemas))

for i, (schema, filenames) in enumerate(unique_schemas.items(), 1):

    print(f"\nSchema {i}")
    print("Number of files:", len(filenames))
    print("Files:")

    for f in filenames:
        print("  ", f)

    print("Number of columns:", len(schema))


# ------------------------------------------------------------
# 4. Compare first IDS schema with CIC-IoT
# ------------------------------------------------------------

first_ids_columns = list(next(iter(ids_schemas.values())))

cic_features = set(
    c for c in cic_columns
    if c.lower() != "label"
)

ids_features = set(
    c for c in first_ids_columns
    if c.lower() != "label"
)


common = sorted(cic_features.intersection(ids_features))

only_cic = sorted(cic_features - ids_features)

only_ids = sorted(ids_features - cic_features)


print("\n" + "=" * 90)
print("FEATURE COMPATIBILITY")
print("=" * 90)

print("\nCIC-IoT features:", len(cic_features))
print("IDS features:", len(ids_features))
print("Common features:", len(common))

print("\nCommon features:")
for c in common:
    print("  ", c)

print("\nFeatures ONLY in CIC-IoT-2023:")
for c in only_cic:
    print("  ", c)

print("\nFeatures ONLY in IDS dataset:")
for c in only_ids:
    print("  ", c)
```

```python
# ============================================================
# CELL 8 — Complete IDS Label Distribution
# ============================================================

import pandas as pd
from pathlib import Path
from collections import Counter

IDS_PATH = Path("/content/drive/MyDrive/IDS dataset")

files = sorted(IDS_PATH.glob("*.csv"))

label_counter = Counter()
file_statistics = []

for i, file in enumerate(files, 1):

    print(f"Processing {i}/{len(files)}: {file.name}")

    try:

        df = pd.read_csv(
            file,
            usecols=lambda c: c.lower() == "label",
            low_memory=False
        )

        # Locate label column regardless of capitalization
        label_col = [
            c for c in df.columns
            if c.lower() == "label"
        ][0]

        counts = df[label_col].value_counts(dropna=False)

        for label, count in counts.items():
            label_counter[str(label)] += int(count)

        file_statistics.append({
            "file": file.name,
            "rows": len(df),
            "classes": df[label_col].nunique(),
            "labels": ", ".join(
                str(x) for x in counts.index
            )
        })

    except Exception as e:

        print("ERROR:", repr(e))


# ------------------------------------------------------------
# Overall distribution
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("COMPLETE IDS LABEL DISTRIBUTION")
print("=" * 90)

label_df = pd.DataFrame(
    [
        {
            "Label": label,
            "Samples": count
        }
        for label, count in label_counter.items()
    ]
)

label_df["Percentage"] = (
    label_df["Samples"]
    / label_df["Samples"].sum()
    * 100
)

label_df = label_df.sort_values(
    "Samples",
    ascending=False
).reset_index(drop=True)

display(label_df)

print(
    "\nTotal IDS samples:",
    label_df["Samples"].sum()
)

print(
    "Number of classes:",
    len(label_df)
)


# ------------------------------------------------------------
# File-level statistics
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("FILE STATISTICS")
print("=" * 90)

file_df = pd.DataFrame(file_statistics)

display(file_df)
```

```python
# ============================================================
# CELL 10 — Robust CSV Loader
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


def clean_dataframe(df, label_column="label"):

    df = df.copy()

    # Normalize column names
    df.columns = [
        str(c).strip()
        for c in df.columns
    ]

    # Locate label column case-insensitively
    label_candidates = [
        c for c in df.columns
        if c.lower() == label_column.lower()
    ]

    if not label_candidates:
        raise ValueError(
            f"Label column '{label_column}' not found."
        )

    actual_label = label_candidates[0]

    # Remove completely empty rows
    df = df.dropna(
        how="all"
    ).reset_index(drop=True)

    # Replace infinity
    df = df.replace(
        [np.inf, -np.inf],
        np.nan
    )

    # Separate labels
    y = df[actual_label].astype(str)

    X = df.drop(
        columns=[actual_label]
    )

    # Convert all features to numeric
    for col in X.columns:
        X[col] = pd.to_numeric(
            X[col],
            errors="coerce"
        )

    return X, y


def load_cic_sample(
    folder,
    n_files=5,
    rows_per_file=50000,
    seed=42
):

    files = sorted(
        Path(folder).glob("*.csv")
    )

    if len(files) == 0:
        raise FileNotFoundError(
            f"No CSV files found in {folder}"
        )

    rng = np.random.default_rng(seed)

    selected = files[:n_files]

    dataframes = []

    for file in selected:

        print(
            f"Loading: {file.name}"
        )

        df = pd.read_csv(
            file,
            nrows=rows_per_file,
            low_memory=False
        )

        X, y = clean_dataframe(
            df,
            label_column="label"
        )

        X["__label__"] = y.values

        dataframes.append(X)

    combined = pd.concat(
        dataframes,
        ignore_index=True
    )

    combined = combined.sample(
        frac=1,
        random_state=seed
    ).reset_index(drop=True)

    return combined


# ------------------------------------------------------------
# Test with a SMALL sample first
# ------------------------------------------------------------

CIC_SAMPLE = load_cic_sample(
    CIC_PATH,
    n_files=5,
    rows_per_file=50000,
    seed=SEED
)

print("\n" + "=" * 70)
print("CIC SAMPLE")
print("=" * 70)

print("Shape:", CIC_SAMPLE.shape)

print("\nLabel distribution:")

display(
    CIC_SAMPLE["__label__"]
    .value_counts()
    .to_frame("count")
)
```

```python
# ============================================================
# CELL 11 — CIC-IoT-2023 Label Analysis
# ============================================================

cic_labels = (
    CIC_SAMPLE["__label__"]
    .value_counts()
    .sort_values(
        ascending=False
    )
)

label_analysis = pd.DataFrame({
    "Class": cic_labels.index,
    "Samples": cic_labels.values
})

label_analysis["Percentage"] = (
    label_analysis["Samples"]
    / label_analysis["Samples"].sum()
    * 100
)

display(label_analysis)

print(
    "\nNumber of CIC-IoT classes:",
    len(label_analysis)
)
```

```python
# ============================================================
# CELL 12 — Binary Label Encoding
# ============================================================

def create_binary_labels(labels):

    labels = labels.astype(str).str.strip()

    benign_mask = (
        labels.str.lower() == "benign"
    )

    binary = np.where(
        benign_mask,
        0,
        1
    )

    return binary


CIC_SAMPLE["binary_label"] = create_binary_labels(
    CIC_SAMPLE["__label__"]
)

print("Binary label distribution:")

binary_counts = (
    CIC_SAMPLE["binary_label"]
    .value_counts()
    .sort_index()
)

display(
    pd.DataFrame({
        "Class": [
            "Benign",
            "Attack"
        ],
        "Label": [
            0,
            1
        ],
        "Samples": [
            binary_counts.get(0, 0),
            binary_counts.get(1, 0)
        ]
    })
)
```

```python
# ============================================================
# CELL 12 — Binary Label Encoding
# ============================================================

import numpy as np
import pandas as pd


def create_binary_labels(labels):

    labels = (
        labels
        .astype(str)
        .str.strip()
    )

    # Benign = 0
    # Everything else = Attack = 1
    benign_mask = (
        labels.str.lower() == "benign"
    )

    binary = np.where(
        benign_mask,
        0,
        1
    )

    return binary


# IMPORTANT:
# The loader created "__label__", NOT "**label**"

CIC_SAMPLE["binary_label"] = create_binary_labels(
    CIC_SAMPLE["__label__"]
)


# ------------------------------------------------------------
# Display distribution
# ------------------------------------------------------------

print("=" * 70)
print("BINARY LABEL DISTRIBUTION")
print("=" * 70)

binary_counts = (
    CIC_SAMPLE["binary_label"]
    .value_counts()
    .sort_index()
)

binary_distribution = pd.DataFrame({
    "Class": [
        "Benign",
        "Attack"
    ],
    "Label": [
        0,
        1
    ],
    "Samples": [
        binary_counts.get(0, 0),
        binary_counts.get(1, 0)
    ]
})

binary_distribution["Percentage"] = (
    binary_distribution["Samples"]
    / binary_distribution["Samples"].sum()
    * 100
)

display(binary_distribution)
```

```python
# ============================================================
# CELL 13 — Find CIC-IoT-2023 Files Containing Benign Traffic
# ============================================================

from pathlib import Path
import pandas as pd

CIC_PATH = Path("/content/drive/MyDrive/CICIoT2023")

cic_files = sorted(CIC_PATH.glob("*.csv"))

benign_files = []
attack_only_files = []
file_label_summary = []

print("=" * 80)
print("SCANNING CIC-IoT-2023 FILES FOR LABEL DISTRIBUTION")
print("=" * 80)

for i, file in enumerate(cic_files, 1):

    print(f"[{i:03d}/{len(cic_files)}] {file.name}")

    try:

        labels_df = pd.read_csv(
            file,
            usecols=["label"],
            low_memory=False
        )

        counts = (
            labels_df["label"]
            .astype(str)
            .str.strip()
            .value_counts()
        )

        has_benign = (
            labels_df["label"]
            .astype(str)
            .str.strip()
            .str.lower()
            .eq("benign")
            .any()
        )

        if has_benign:
            benign_files.append(file)
        else:
            attack_only_files.append(file)

        file_label_summary.append({
            "file": file.name,
            "rows": len(labels_df),
            "num_classes": len(counts),
            "has_benign": has_benign,
            "top_label": counts.index[0],
            "top_count": counts.iloc[0]
        })

    except Exception as e:

        print("ERROR:", repr(e))


summary_df = pd.DataFrame(file_label_summary)

print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)

print("Total CSV files:", len(cic_files))
print("Files containing Benign:", len(benign_files))
print("Attack-only files:", len(attack_only_files))

display(summary_df)
```

```python
# ============================================================
# CELL 14 — Overall CIC-IoT Label Distribution
# ============================================================

from collections import Counter

overall_counts = Counter()

for file in cic_files:

    try:

        labels = pd.read_csv(
            file,
            usecols=["label"],
            low_memory=False
        )["label"]

        labels = (
            labels
            .astype(str)
            .str.strip()
        )

        overall_counts.update(
            labels.tolist()
        )

    except Exception as e:

        print("ERROR:", file.name, repr(e))


overall_df = pd.DataFrame(
    [
        {
            "Label": label,
            "Samples": count
        }
        for label, count in overall_counts.items()
    ]
)

overall_df["Percentage"] = (
    overall_df["Samples"]
    / overall_df["Samples"].sum()
    * 100
)

overall_df = (
    overall_df
    .sort_values(
        "Samples",
        ascending=False
    )
    .reset_index(drop=True)
)

print("=" * 80)
print("COMPLETE CIC-IoT-2023 LABEL DISTRIBUTION")
print("=" * 80)

display(overall_df)

print(
    "\nTotal samples:",
    overall_df["Samples"].sum()
)

print(
    "Total classes:",
    len(overall_df)
)

print(
    "Benign samples:",
    overall_counts.get("Benign", 0)
)
```

```python
# ============================================================
# CELL 15 — ROBUST CIC-IoT-2023 SAMPLER
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
import gc

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

CIC_PATH = Path("/content/drive/MyDrive/CICIoT2023")

SEED = 42

# Start smaller for verification.
# Once everything works, we can increase this.
TARGET_TOTAL = 200_000

TARGET_BENIGN = TARGET_TOTAL // 2
TARGET_ATTACK = TARGET_TOTAL - TARGET_BENIGN

print("=" * 90)
print("ROBUST CIC-IoT-2023 SAMPLER")
print("=" * 90)

print("Target total :", TARGET_TOTAL)
print("Target benign:", TARGET_BENIGN)
print("Target attack:", TARGET_ATTACK)


# ------------------------------------------------------------
# Find files
# ------------------------------------------------------------

cic_files = sorted(CIC_PATH.glob("*.csv"))

print("\nCSV files found:", len(cic_files))


# ------------------------------------------------------------
# PASS 1
# Safely determine label counts
# ------------------------------------------------------------

file_information = []
failed_files = []

for i, file in enumerate(cic_files, 1):

    print(
        f"[{i:03d}/{len(cic_files)}] "
        f"Scanning {file.name}"
    )

    try:

        labels = pd.read_csv(
            file,
            usecols=["label"],
            low_memory=False
        )

        # Force labels to strings
        labels["label"] = (
            labels["label"]
            .astype("string")
            .str.strip()
        )

        # Remove missing labels
        labels = labels[
            labels["label"].notna()
        ]

        total = int(len(labels))

        benign_mask = (
            labels["label"]
            .str.lower()
            .eq("benign")
        )

        benign_count = int(
            benign_mask.sum()
        )

        attack_count = int(
            total - benign_count
        )

        file_information.append({
            "file": file,
            "total": total,
            "benign": benign_count,
            "attack": attack_count
        })

    except Exception as e:

        failed_files.append({
            "file": str(file),
            "error": repr(e)
        })

        print(
            "   SKIPPED:",
            repr(e)
        )


# ------------------------------------------------------------
# Create count table
# ------------------------------------------------------------

counts_df = pd.DataFrame(
    file_information
)

if counts_df.empty:

    raise RuntimeError(
        "No valid CIC-IoT CSV files could be read."
    )


print("\n" + "=" * 90)
print("FILE SCAN COMPLETE")
print("=" * 90)

print(
    "Valid files:",
    len(counts_df)
)

print(
    "Failed files:",
    len(failed_files)
)

print(
    "Total available rows:",
    int(counts_df["total"].sum())
)

print(
    "Available benign:",
    int(counts_df["benign"].sum())
)

print(
    "Available attack:",
    int(counts_df["attack"].sum())
)


# ------------------------------------------------------------
# Make sure requested sample is possible
# ------------------------------------------------------------

available_benign = int(
    counts_df["benign"].sum()
)

available_attack = int(
    counts_df["attack"].sum()
)

actual_target_benign = min(
    TARGET_BENIGN,
    available_benign
)

actual_target_attack = min(
    TARGET_ATTACK,
    available_attack
)

print("\nRequested benign:", TARGET_BENIGN)
print("Requested attack:", TARGET_ATTACK)

print(
    "Actual benign target:",
    actual_target_benign
)

print(
    "Actual attack target:",
    actual_target_attack
)


# ------------------------------------------------------------
# Calculate SAFE sampling fractions
# ------------------------------------------------------------

if available_benign > 0:

    benign_fraction = (
        actual_target_benign
        / available_benign
    )

else:

    benign_fraction = 0.0


if available_attack > 0:

    attack_fraction = (
        actual_target_attack
        / available_attack
    )

else:

    attack_fraction = 0.0


print("\nSampling fractions:")
print(
    f"Benign: {benign_fraction:.8f}"
)

print(
    f"Attack: {attack_fraction:.8f}"
)


# ------------------------------------------------------------
# PASS 2
# Load and sample files
# ------------------------------------------------------------

sampled_frames = []

successful_files = 0

for i, row in counts_df.iterrows():

    file = Path(row["file"])

    print(
        f"\n[{i+1:03d}/{len(counts_df)}] "
        f"Loading {file.name}"
    )

    try:

        # ----------------------------------------------------
        # Read file
        # ----------------------------------------------------

        df = pd.read_csv(
            file,
            low_memory=False
        )

        # ----------------------------------------------------
        # Normalize column names
        # ----------------------------------------------------

        df.columns = [
            str(c).strip()
            for c in df.columns
        ]

        if "label" not in df.columns:

            print(
                "   SKIPPED: label column missing"
            )

            continue


        # ----------------------------------------------------
        # Normalize labels
        # ----------------------------------------------------

        df["label"] = (
            df["label"]
            .astype("string")
            .str.strip()
        )

        df = df[
            df["label"].notna()
        ].copy()

        if df.empty:

            print(
                "   SKIPPED: no valid rows"
            )

            continue


        # ----------------------------------------------------
        # Separate classes
        # ----------------------------------------------------

        benign_df = df[
            df["label"]
            .str.lower()
            .eq("benign")
        ]

        attack_df = df[
            ~df["label"]
            .str.lower()
            .eq("benign")
        ]


        # ----------------------------------------------------
        # Calculate sample sizes SAFELY
        # ----------------------------------------------------

        n_benign = int(
            min(
                len(benign_df),
                round(
                    len(benign_df)
                    * benign_fraction
                )
            )
        )

        n_attack = int(
            min(
                len(attack_df),
                round(
                    len(attack_df)
                    * attack_fraction
                )
            )
        )


        # ----------------------------------------------------
        # Sample benign
        # ----------------------------------------------------

        if (
            n_benign > 0
            and len(benign_df) > 0
        ):

            sampled_benign = (
                benign_df.sample(
                    n=n_benign,
                    random_state=SEED + i
                )
            )

            sampled_frames.append(
                sampled_benign
            )


        # ----------------------------------------------------
        # Sample attacks
        # ----------------------------------------------------

        if (
            n_attack > 0
            and len(attack_df) > 0
        ):

            sampled_attack = (
                attack_df.sample(
                    n=n_attack,
                    random_state=SEED + i
                )
            )

            sampled_frames.append(
                sampled_attack
            )


        successful_files += 1

        print(
            f"   benign={n_benign:,} "
            f"attack={n_attack:,}"
        )


        # Free memory
        del df
        del benign_df
        del attack_df

        gc.collect()


    except Exception as e:

        print(
            "   ERROR:",
            repr(e)
        )

        failed_files.append({
            "file": str(file),
            "error": repr(e)
        })


# ------------------------------------------------------------
# Verify that sampling succeeded
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("SAMPLING COMPLETE")
print("=" * 90)

print(
    "Successful files:",
    successful_files
)

print(
    "Number of sampled frames:",
    len(sampled_frames)
)

if len(sampled_frames) == 0:

    raise RuntimeError(
        "No data was sampled. "
        "Check the file schemas and label column."
    )


# ------------------------------------------------------------
# Combine
# ------------------------------------------------------------

CIC_DATA = pd.concat(
    sampled_frames,
    ignore_index=True
)


# ------------------------------------------------------------
# Shuffle
# ------------------------------------------------------------

CIC_DATA = (
    CIC_DATA
    .sample(
        frac=1.0,
        random_state=SEED
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Binary labels
# ------------------------------------------------------------

CIC_DATA["binary_label"] = np.where(
    CIC_DATA["label"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("benign"),
    0,
    1
)


# ------------------------------------------------------------
# Final distribution
# ------------------------------------------------------------

distribution = (
    CIC_DATA["binary_label"]
    .value_counts()
    .sort_index()
)

final_distribution = pd.DataFrame({

    "Class": [
        "Benign",
        "Attack"
    ],

    "Label": [
        0,
        1
    ],

    "Samples": [
        int(distribution.get(0, 0)),
        int(distribution.get(1, 0))
    ]
})

final_distribution["Percentage"] = (
    final_distribution["Samples"]
    / final_distribution["Samples"].sum()
    * 100
)


print("\n" + "=" * 90)
print("FINAL DATASET")
print("=" * 90)

print(
    "Shape:",
    CIC_DATA.shape
)

display(final_distribution)


# ------------------------------------------------------------
# Save metadata
# ------------------------------------------------------------

counts_df.to_csv(
    "/content/drive/MyDrive/EFG_IDS_RESULTS/"
    "CIC_file_statistics.csv",
    index=False
)

final_distribution.to_csv(
    "/content/drive/MyDrive/EFG_IDS_RESULTS/"
    "CIC_binary_distribution.csv",
    index=False
)

print(
    "\nMetadata saved to:",
    "/content/drive/MyDrive/EFG_IDS_RESULTS/"
)
```

```python
# ============================================================
# STEP 18 — MEMORY-SAFE CIC-IoT-2023 SAMPLER
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path
import gc

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

CIC_PATH = Path("/content/drive/MyDrive/CICIoT2023")

SEED = 42

# CHANGE THIS according to Step 15A
BENIGN_LABEL = "BenignTraffic"

TARGET_TOTAL = 200_000

TARGET_BENIGN = TARGET_TOTAL // 2
TARGET_ATTACK = TARGET_TOTAL - TARGET_BENIGN

CHUNK_SIZE = 100_000

rng = np.random.default_rng(SEED)

print("=" * 90)
print("STEP 18 — CHUNK-BASED CIC-IoT-2023 SAMPLER")
print("=" * 90)

print("Benign label :", BENIGN_LABEL)
print("Target total :", TARGET_TOTAL)
print("Target benign:", TARGET_BENIGN)
print("Target attack:", TARGET_ATTACK)
print("Chunk size   :", CHUNK_SIZE)


# ------------------------------------------------------------
# Find files
# ------------------------------------------------------------

files = sorted(
    CIC_PATH.glob("*.csv")
)

print("\nCSV files:", len(files))


# ------------------------------------------------------------
# Reservoir-style collection
# ------------------------------------------------------------

benign_parts = []
attack_parts = []

benign_collected = 0
attack_collected = 0


# ------------------------------------------------------------
# Process files one by one
# ------------------------------------------------------------

for file_index, file in enumerate(files):

    if (
        benign_collected >= TARGET_BENIGN
        and attack_collected >= TARGET_ATTACK
    ):
        break

    print(
        f"\n[{file_index + 1}/{len(files)}] "
        f"{file.name}"
    )

    try:

        for chunk in pd.read_csv(
            file,
            chunksize=CHUNK_SIZE,
            low_memory=False
        ):

            # Normalize columns
            chunk.columns = [
                str(c).strip()
                for c in chunk.columns
            ]

            if "label" not in chunk.columns:
                continue

            # Normalize labels
            chunk["label"] = (
                chunk["label"]
                .astype("string")
                .str.strip()
            )

            # ------------------------------------------------
            # Benign
            # ------------------------------------------------

            benign_mask = (
                chunk["label"]
                .str.lower()
                ==
                BENIGN_LABEL.lower()
            )

            benign_chunk = chunk[
                benign_mask
            ]

            # ------------------------------------------------
            # Attack
            # ------------------------------------------------

            attack_chunk = chunk[
                ~benign_mask
                &
                chunk["label"].notna()
            ]

            # ------------------------------------------------
            # Remaining requirements
            # ------------------------------------------------

            benign_needed = (
                TARGET_BENIGN
                - benign_collected
            )

            attack_needed = (
                TARGET_ATTACK
                - attack_collected
            )

            # ------------------------------------------------
            # Sample benign
            # ------------------------------------------------

            if (
                benign_needed > 0
                and len(benign_chunk) > 0
            ):

                n = min(
                    len(benign_chunk),
                    benign_needed
                )

                sampled = benign_chunk.sample(
                    n=n,
                    random_state=SEED
                )

                benign_parts.append(
                    sampled
                )

                benign_collected += n


            # ------------------------------------------------
            # Sample attack
            # ------------------------------------------------

            if (
                attack_needed > 0
                and len(attack_chunk) > 0
            ):

                n = min(
                    len(attack_chunk),
                    attack_needed
                )

                sampled = attack_chunk.sample(
                    n=n,
                    random_state=SEED
                )

                attack_parts.append(
                    sampled
                )

                attack_collected += n


            del chunk

            gc.collect()


            # Stop if target reached
            if (
                benign_collected >= TARGET_BENIGN
                and attack_collected >= TARGET_ATTACK
            ):
                break


    except Exception as e:

        print(
            "ERROR:",
            repr(e)
        )


    print(
        "Collected benign:",
        benign_collected
    )

    print(
        "Collected attack:",
        attack_collected
    )


# ------------------------------------------------------------
# Combine
# ------------------------------------------------------------

if len(benign_parts) == 0:

    raise RuntimeError(
        "No benign samples found. "
        "Check BENIGN_LABEL."
    )

if len(attack_parts) == 0:

    raise RuntimeError(
        "No attack samples found."
    )


benign_data = pd.concat(
    benign_parts,
    ignore_index=True
)

attack_data = pd.concat(
    attack_parts,
    ignore_index=True
)


CIC_DATA = pd.concat(
    [
        benign_data,
        attack_data
    ],
    ignore_index=True
)


# ------------------------------------------------------------
# Shuffle
# ------------------------------------------------------------

CIC_DATA = (
    CIC_DATA
    .sample(
        frac=1,
        random_state=SEED
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Binary labels
# ------------------------------------------------------------

CIC_DATA["binary_label"] = np.where(
    CIC_DATA["label"]
    .astype(str)
    .str.strip()
    .str.lower()
    ==
    BENIGN_LABEL.lower(),
    0,
    1
)


# ------------------------------------------------------------
# Result
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("FINAL SAMPLE")
print("=" * 90)

print(
    "Shape:",
    CIC_DATA.shape
)

print(
    "\nBinary distribution:"
)

display(
    CIC_DATA["binary_label"]
    .value_counts()
    .sort_index()
)

print(
    "\nOriginal labels:"
)

display(
    CIC_DATA["label"]
    .value_counts()
    .head(30)
)
```

```python
# ============================================================
# STEP 19 — FEATURE CLEANING
# ============================================================

import numpy as np
import pandas as pd

LABEL_COLUMNS = [
    "label",
    "binary_label"
]

FEATURE_COLUMNS = [
    c for c in CIC_DATA.columns
    if c not in LABEL_COLUMNS
]

print("Initial features:", len(FEATURE_COLUMNS))


# ------------------------------------------------------------
# Convert to numeric
# ------------------------------------------------------------

X = CIC_DATA[
    FEATURE_COLUMNS
].copy()

for col in FEATURE_COLUMNS:

    X[col] = pd.to_numeric(
        X[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# Replace infinity
# ------------------------------------------------------------

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


# ------------------------------------------------------------
# Remove completely empty columns
# ------------------------------------------------------------

empty_columns = [
    c for c in X.columns
    if X[c].isna().all()
]

if empty_columns:

    print(
        "Removing empty columns:",
        empty_columns
    )

    X = X.drop(
        columns=empty_columns
    )


FEATURE_COLUMNS = X.columns.tolist()

print(
    "\nRemaining numerical features:",
    len(FEATURE_COLUMNS)
)

print(
    "\nMissing values:",
    X.isna().sum().sum()
)

print(
    "Infinite values:",
    np.isinf(
        X.to_numpy(
            dtype=np.float64,
            na_value=np.nan
        )
    ).sum()
)
```

```python
# ============================================================
# STEP 22 — LEAKAGE-SAFE STANDARDIZATION
# ============================================================

from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd
import gc

scaler = StandardScaler()

# Training medians are calculated ONLY from training data
train_median = X_train.median()

X_train_temp = X_train.fillna(train_median)
X_val_temp   = X_val.fillna(train_median)
X_test_temp  = X_test.fillna(train_median)

# Fit scaler ONLY on training data
X_train_scaled = scaler.fit_transform(X_train_temp)
X_val_scaled   = scaler.transform(X_val_temp)
X_test_scaled  = scaler.transform(X_test_temp)

print("=" * 80)
print("STANDARDIZATION COMPLETE")
print("=" * 80)

print("X_train_scaled:", X_train_scaled.shape)
print("X_val_scaled  :", X_val_scaled.shape)
print("X_test_scaled :", X_test_scaled.shape)
```

```python
# ============================================================
# STEP 23 — CONTROLLED MISSINGNESS
# ============================================================

import numpy as np

MISSING_RATE = 0.05
SEED = 55


def introduce_missingness(
    X,
    missing_rate=0.05,
    seed=42
):

    rng = np.random.default_rng(seed)

    X_missing = X.copy()

    missing_mask = (
        rng.random(X.shape)
        < missing_rate
    )

    X_missing[missing_mask] = np.nan

    return X_missing, missing_mask


# ------------------------------------------------------------
# Create artificial missing values ONLY in training data
# ------------------------------------------------------------

X_train_missing, train_missing_mask = (
    introduce_missingness(
        X_train_scaled,
        missing_rate=MISSING_RATE,
        seed=SEED
    )
)


print("=" * 80)
print("CONTROLLED MISSINGNESS")
print("=" * 80)

print(
    "Original shape:",
    X_train_scaled.shape
)

print(
    "Missing shape:",
    X_train_missing.shape
)

print(
    "Missing values:",
    np.isnan(
        X_train_missing
    ).sum()
)

print(
    "Missing percentage:",
    np.isnan(
        X_train_missing
    ).mean() * 100
)
```

```python
# ============================================================
# STEP 24 — TEMPORAL SEQUENCE CREATION
# ============================================================

import numpy as np

SEQUENCE_LENGTH = 15


def create_sequences(
    X,
    sequence_length=10
):

    sequences = []
    indices = []

    for i in range(
        len(X) - sequence_length + 1
    ):

        sequences.append(
            X[
                i:i + sequence_length
            ]
        )

        indices.append(i)

    return (
        np.asarray(
            sequences,
            dtype=np.float32
        ),
        np.asarray(
            indices,
            dtype=np.int64
        )
    )


# ------------------------------------------------------------
# Missing input sequences
# ------------------------------------------------------------

X_seq, seq_indices = create_sequences(
    X_train_missing,
    SEQUENCE_LENGTH
)


# ------------------------------------------------------------
# Original target sequences
# ------------------------------------------------------------

X_target_seq, _ = create_sequences(
    X_train_scaled,
    SEQUENCE_LENGTH
)


# ------------------------------------------------------------
# Create corresponding missing-value masks
# ------------------------------------------------------------

mask_seq = np.asarray([
    train_missing_mask[
        start:start + SEQUENCE_LENGTH
    ]
    for start in seq_indices
])


print("=" * 80)
print("SEQUENCE CREATION COMPLETE")
print("=" * 80)

print(
    "Input sequence:",
    X_seq.shape
)

print(
    "Target sequence:",
    X_target_seq.shape
)

print(
    "Mask sequence:",
    mask_seq.shape
)

print(
    "Missing positions:",
    mask_seq.sum()
)
```

```python
# ============================================================
# STEP 24A — SANITY CHECK
# ============================================================

print("X_train_scaled :", X_train_scaled.shape)
print("X_train_missing:", X_train_missing.shape)

print("X_seq          :", X_seq.shape)
print("X_target_seq   :", X_target_seq.shape)
print("mask_seq       :", mask_seq.shape)

assert X_seq.shape == X_target_seq.shape
assert X_seq.shape == mask_seq.shape

print("\n✓ All sequence dimensions are consistent.")
```

```python
# ============================================================
# STEP 25 — BIDIRECTIONAL LSTM IMPUTER
# ============================================================

import torch
import torch.nn as nn

DEVICE = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("Device:", DEVICE)


class BiLSTMImputer(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=64,
        num_layers=2
    ):

        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True,
            dropout=0.05
            if num_layers > 1
            else 0.0
        )

        self.output_layer = nn.Linear(
            hidden_dim * 2,
            input_dim
        )


    def forward(self, x):

        output, _ = self.lstm(x)

        output = self.output_layer(
            output
        )

        return output


INPUT_DIM = X_seq.shape[-1]

imputer_model = BiLSTMImputer(
    input_dim=INPUT_DIM,
    hidden_dim=64,
    num_layers=2
).to(DEVICE)

print(imputer_model)
```

```python
# ============================================================
# STEP 26 — BUILD GRAPH DATA FOR GRAPHSAGE
# ============================================================

import numpy as np
import pandas as pd
import torch
from torch_geometric.data import Data
from sklearn.neighbors import NearestNeighbors

print("=" * 80)
print("BUILDING GRAPH FOR GRAPHSAGE")
print("=" * 80)


# ------------------------------------------------------------
# 1. CHECK REQUIRED VARIABLES
# ------------------------------------------------------------

required = [
    "X_train_scaled",
    "X_val_scaled",
    "X_test_scaled",
    "y_train",
    "y_val",
    "y_test"
]

missing = [
    name for name in required
    if name not in globals()
]

if missing:
    raise NameError(
        "The following variables are missing: "
        + str(missing)
        + "\nRun your train/validation/test split cell first."
    )


# ------------------------------------------------------------
# 2. CONVERT LABELS TO NUMPY
# ------------------------------------------------------------

def convert_labels(y):

    if isinstance(y, pd.Series):
        return y.to_numpy()

    if isinstance(y, pd.DataFrame):
        return y.iloc[:, 0].to_numpy()

    if torch.is_tensor(y):
        return y.detach().cpu().numpy()

    return np.asarray(y)


y_train_np = convert_labels(y_train)
y_val_np   = convert_labels(y_val)
y_test_np  = convert_labels(y_test)


# ------------------------------------------------------------
# 3. COMBINE DATA
# ------------------------------------------------------------

X_all = np.concatenate(
    [
        X_train_scaled,
        X_val_scaled,
        X_test_scaled
    ],
    axis=0
).astype(np.float32)


y_all = np.concatenate(
    [
        y_train_np,
        y_val_np,
        y_test_np
    ],
    axis=0
)


# Make sure labels are integer
y_all = y_all.astype(np.int64)


# ------------------------------------------------------------
# 4. CREATE TRAIN / VALIDATION / TEST INDICES
# ------------------------------------------------------------

n_train = len(X_train_scaled)
n_val   = len(X_val_scaled)
n_test  = len(X_test_scaled)

train_indices = np.arange(
    0,
    n_train
)

val_indices = np.arange(
    n_train,
    n_train + n_val
)

test_indices = np.arange(
    n_train + n_val,
    n_train + n_val + n_test
)


print("Total nodes:", len(X_all))
print("Training nodes:", len(train_indices))
print("Validation nodes:", len(val_indices))
print("Testing nodes:", len(test_indices))


# ------------------------------------------------------------
# 5. CREATE kNN GRAPH
# ------------------------------------------------------------

# Keep this moderate because you are currently using CPU
K_NEIGHBORS = 15

print("\nBuilding kNN graph...")
print("k =", K_NEIGHBORS)

knn = NearestNeighbors(
    n_neighbors=K_NEIGHBORS + 1,
    metric="euclidean",
    n_jobs=-1
)

knn.fit(X_all)

distances, neighbors = knn.kneighbors(
    X_all
)


# ------------------------------------------------------------
# 6. CREATE EDGE LIST
# ------------------------------------------------------------

source_nodes = []
target_nodes = []

for i in range(len(X_all)):

    # neighbors[0] is the node itself
    for j in neighbors[i, 1:]:

        source_nodes.append(i)
        target_nodes.append(j)


edge_index = torch.tensor(
    [
        source_nodes,
        target_nodes
    ],
    dtype=torch.long
)


# ------------------------------------------------------------
# 7. CREATE PYTORCH GEOMETRIC DATA OBJECT
# ------------------------------------------------------------

x_tensor = torch.tensor(
    X_all,
    dtype=torch.float32
)

y_tensor = torch.tensor(
    y_all,
    dtype=torch.long
)


train_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)

val_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)

test_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)


train_mask[train_indices] = True
val_mask[val_indices] = True
test_mask[test_indices] = True


graph_data = Data(
    x=x_tensor,
    y=y_tensor,
    edge_index=edge_index,
    train_mask=train_mask,
    val_mask=val_mask,
    test_mask=test_mask
)


# ------------------------------------------------------------
# 8. DISPLAY GRAPH INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("GRAPH CREATED SUCCESSFULLY")
print("=" * 80)

print(
    "Number of nodes:",
    graph_data.num_nodes
)

print(
    "Number of features:",
    graph_data.num_features
)

print(
    "Number of edges:",
    graph_data.num_edges
)

print(
    "Feature matrix:",
    graph_data.x.shape
)

print(
    "Label vector:",
    graph_data.y.shape
)

print(
    "Training nodes:",
    int(graph_data.train_mask.sum())
)

print(
    "Validation nodes:",
    int(graph_data.val_mask.sum())
)

print(
    "Testing nodes:",
    int(graph_data.test_mask.sum())
)

print("\nLabel distribution:")

unique, counts = torch.unique(
    graph_data.y,
    return_counts=True
)

for label, count in zip(unique, counts):

    print(
        f"Class {int(label)}: "
        f"{int(count)} samples"
    )

print("\n✓ graph_data is ready.")
```

```python
# ============================================================
# STEP 26B — DEFINE GRAPHSAGE IDS MODEL
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.nn import SAGEConv


# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)


# ------------------------------------------------------------
# GraphSAGE model
# ------------------------------------------------------------

class GraphSAGE_IDS(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        embedding_dim=64,
        num_classes=2,
        dropout=0.30
    ):

        super().__init__()

        # GraphSAGE Layer 1
        self.conv1 = SAGEConv(
            input_dim,
            hidden_dim
        )

        # GraphSAGE Layer 2
        self.conv2 = SAGEConv(
            hidden_dim,
            embedding_dim
        )

        # Classification layer
        self.classifier = nn.Linear(
            embedding_dim,
            num_classes
        )

        self.dropout = dropout


    def forward(
        self,
        x,
        edge_index
    ):

        # First GraphSAGE layer
        x = self.conv1(
            x,
            edge_index
        )

        x = F.relu(x)

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        # Second GraphSAGE layer
        embedding = self.conv2(
            x,
            edge_index
        )

        embedding = F.relu(
            embedding
        )

        embedding = F.dropout(
            embedding,
            p=self.dropout,
            training=self.training
        )

        # Classification
        logits = self.classifier(
            embedding
        )

        return logits, embedding


# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

model = GraphSAGE_IDS(
    input_dim=graph_data.num_features,
    hidden_dim=128,
    embedding_dim=64,
    num_classes=2,
    dropout=0.30
).to(DEVICE)


# ------------------------------------------------------------
# Move graph to device
# ------------------------------------------------------------

graph_data = graph_data.to(DEVICE)


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("=" * 80)
print("GRAPHSAGE MODEL CREATED")
print("=" * 80)

print("Input features :", graph_data.num_features)
print("Hidden dimension:", 128)
print("Embedding dimension:", 64)
print("Classes:", 2)
print("Device:", DEVICE)

print(
    "Total parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

print("\n✓ Model is ready.")
```

```python
# ============================================================
# STEP 27 — LOSS AND OPTIMIZER
# ============================================================

import numpy as np
import torch.nn as nn


# ------------------------------------------------------------
# Get training labels
# ------------------------------------------------------------

train_labels = (
    graph_data.y[
        graph_data.train_mask
    ]
    .detach()
    .cpu()
    .numpy()
)


# ------------------------------------------------------------
# Class distribution
# ------------------------------------------------------------

class_counts = np.bincount(
    train_labels,
    minlength=2
)

print("=" * 80)
print("TRAINING CLASS DISTRIBUTION")
print("=" * 80)

print("Benign :", class_counts[0])
print("Attack :", class_counts[1])


# ------------------------------------------------------------
# Calculate class weights
# ------------------------------------------------------------

total_samples = class_counts.sum()

class_weights = (
    total_samples /
    (2.0 * class_counts)
)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=DEVICE
)

print("\nClass weights:")
print(class_weights)


# ------------------------------------------------------------
# Loss
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)


# ------------------------------------------------------------
# Optimizer
# ------------------------------------------------------------

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)

print("\n✓ Loss function ready.")
print("✓ Adam optimizer ready.")
```

```python
# ============================================================
# STEP 28 — TRAINING SANITY CHECK
# ============================================================

model.train()

optimizer.zero_grad()

logits, embeddings = model(
    graph_data.x,
    graph_data.edge_index
)

train_logits = logits[
    graph_data.train_mask
]

train_targets = graph_data.y[
    graph_data.train_mask
]

loss = criterion(
    train_logits,
    train_targets
)

print("Logits shape:", logits.shape)
print("Embedding shape:", embeddings.shape)
print("Training logits:", train_logits.shape)
print("Training labels:", train_targets.shape)
print("Initial loss:", loss.item())

loss.backward()

optimizer.step()

print("\n✓ Forward pass successful.")
print("✓ Backward pass successful.")
print("✓ Optimizer step successful.")
```

```python
# ============================================================
# STEP 28 — TRAIN GRAPHSAGE
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

EPOCHS = 20

TRAIN_HISTORY = []

best_val_f1 = -1.0
best_state = None

print("=" * 100)
print("TRAINING GRAPHSAGE")
print("=" * 100)

for epoch in range(1, EPOCHS + 1):

    # -----------------------------
    # Training
    # -----------------------------
    model.train()

    optimizer.zero_grad()

    logits, embeddings = model(
        graph_data.x,
        graph_data.edge_index
    )

    train_logits = logits[
        graph_data.train_mask
    ]

    train_targets = graph_data.y[
        graph_data.train_mask
    ]

    loss = criterion(
        train_logits,
        train_targets
    )

    loss.backward()

    torch.nn.utils.clip_grad_norm_(
        model.parameters(),
        max_norm=5.0
    )

    optimizer.step()

    # -----------------------------
    # Validation
    # -----------------------------
    model.eval()

    with torch.no_grad():

        val_logits, _ = model(
            graph_data.x,
            graph_data.edge_index
        )

        val_logits = val_logits[
            graph_data.val_mask
        ]

        val_targets = graph_data.y[
            graph_data.val_mask
        ]

        val_pred = (
            val_logits
            .argmax(dim=1)
            .cpu()
            .numpy()
        )

        val_true = (
            val_targets
            .cpu()
            .numpy()
        )

    # -----------------------------
    # Metrics
    # -----------------------------
    val_accuracy = accuracy_score(
        val_true,
        val_pred
    )

    val_precision = precision_score(
        val_true,
        val_pred,
        zero_division=0
    )

    val_recall = recall_score(
        val_true,
        val_pred,
        zero_division=0
    )

    val_f1 = f1_score(
        val_true,
        val_pred,
        zero_division=0
    )

    # -----------------------------
    # Save best model
    # -----------------------------
    if val_f1 > best_val_f1:

        best_val_f1 = val_f1

        best_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

    # -----------------------------
    # Save history
    # -----------------------------
    TRAIN_HISTORY.append({
        "Epoch": epoch,
        "Train Loss": loss.item(),
        "Validation Accuracy": val_accuracy,
        "Validation Precision": val_precision,
        "Validation Recall": val_recall,
        "Validation F1": val_f1
    })

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Loss={loss.item():.5f} | "
        f"Acc={val_accuracy:.4f} | "
        f"Precision={val_precision:.4f} | "
        f"Recall={val_recall:.4f} | "
        f"F1={val_f1:.4f}"
    )

print("\nTraining completed.")
print("Best validation F1:", best_val_f1)
```

```python
# ============================================================
# STEP 29 — RESTORE BEST MODEL
# ============================================================

if best_state is None:
    raise RuntimeError("Best model was not saved.")

model.load_state_dict(best_state)
model = model.to(DEVICE)

print("✓ Best GraphSAGE model restored.")
print(f"Best validation F1: {best_val_f1:.4f}")
```

```python
# ============================================================
# STEP 30 — FINAL TEST EVALUATION
# ============================================================

model.eval()

with torch.no_grad():

    test_logits, test_embeddings = model(
        graph_data.x,
        graph_data.edge_index
    )

    test_logits = test_logits[
        graph_data.test_mask
    ]

    test_targets = graph_data.y[
        graph_data.test_mask
    ]

    test_pred = (
        test_logits
        .argmax(dim=1)
        .cpu()
        .numpy()
    )

    test_true = (
        test_targets
        .cpu()
        .numpy()
    )


# Metrics
accuracy = accuracy_score(
    test_true,
    test_pred
)

precision = precision_score(
    test_true,
    test_pred,
    zero_division=0
)

recall = recall_score(
    test_true,
    test_pred,
    zero_division=0
)

f1 = f1_score(
    test_true,
    test_pred,
    zero_division=0
)


print("=" * 80)
print("FINAL TEST RESULTS — GRAPHSAGE")
print("=" * 80)

print(f"Accuracy  : {accuracy:.4f}  ({accuracy*100:.2f}%)")
print(f"Precision : {precision:.4f}  ({precision*100:.2f}%)")
print(f"Recall    : {recall:.4f}  ({recall*100:.2f}%)")
print(f"F1-Score  : {f1:.4f}  ({f1*100:.2f}%)")
print("=" * 80)
```

```python
# ============================================================
# STEP 28 — GRAPHSAGE TRAINING WITH TRAIN + VALIDATION METRICS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

EPOCHS = 20

TRAIN_HISTORY = []

best_val_f1 = -1.0
best_state = None

print("=" * 100)
print("TRAINING GRAPHSAGE")
print("=" * 100)

for epoch in range(1, EPOCHS + 1):

    # ========================================================
    # TRAINING
    # ========================================================

    model.train()

    optimizer.zero_grad()

    logits, embeddings = model(
        graph_data.x,
        graph_data.edge_index
    )

    train_logits = logits[
        graph_data.train_mask
    ]

    train_targets = graph_data.y[
        graph_data.train_mask
    ]

    loss = criterion(
        train_logits,
        train_targets
    )

    loss.backward()

    torch.nn.utils.clip_grad_norm_(
        model.parameters(),
        max_norm=5.0
    )

    optimizer.step()

    # --------------------------------------------------------
    # TRAINING PREDICTIONS
    # --------------------------------------------------------

    train_pred = (
        train_logits
        .argmax(dim=1)
        .detach()
        .cpu()
        .numpy()
    )

    train_true = (
        train_targets
        .detach()
        .cpu()
        .numpy()
    )

    train_accuracy = accuracy_score(
        train_true,
        train_pred
    )

    train_precision = precision_score(
        train_true,
        train_pred,
        zero_division=0
    )

    train_recall = recall_score(
        train_true,
        train_pred,
        zero_division=0
    )

    train_f1 = f1_score(
        train_true,
        train_pred,
        zero_division=0
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    model.eval()

    with torch.no_grad():

        val_logits, _ = model(
            graph_data.x,
            graph_data.edge_index
        )

        val_logits = val_logits[
            graph_data.val_mask
        ]

        val_targets = graph_data.y[
            graph_data.val_mask
        ]

        val_pred = (
            val_logits
            .argmax(dim=1)
            .cpu()
            .numpy()
        )

        val_true = (
            val_targets
            .cpu()
            .numpy()
        )


    # ========================================================
    # VALIDATION METRICS
    # ========================================================

    val_accuracy = accuracy_score(
        val_true,
        val_pred
    )

    val_precision = precision_score(
        val_true,
        val_pred,
        zero_division=0
    )

    val_recall = recall_score(
        val_true,
        val_pred,
        zero_division=0
    )

    val_f1 = f1_score(
        val_true,
        val_pred,
        zero_division=0
    )


    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    if val_f1 > best_val_f1:

        best_val_f1 = val_f1

        best_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }


    # ========================================================
    # SAVE TRAINING HISTORY
    # ========================================================

    TRAIN_HISTORY.append({

        "Epoch": epoch,

        "Train Loss": loss.item(),

        "Train Accuracy": train_accuracy,
        "Validation Accuracy": val_accuracy,

        "Train Precision": train_precision,
        "Validation Precision": val_precision,

        "Train Recall": train_recall,
        "Validation Recall": val_recall,

        "Train F1": train_f1,
        "Validation F1": val_f1
    })


    # ========================================================
    # PRINT RESULTS
    # ========================================================

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Loss={loss.item():.5f} | "
        f"Train Acc={train_accuracy:.4f} | "
        f"Val Acc={val_accuracy:.4f} | "
        f"Train F1={train_f1:.4f} | "
        f"Val F1={val_f1:.4f}"
    )


print("\n" + "=" * 100)
print("TRAINING COMPLETE")
print("=" * 100)

print(
    f"Best Validation F1: {best_val_f1:.4f}"
)
```

```python
# ============================================================
# STEP 26 — BUILD GRAPH DATA FOR GRAPHSAGE
# ============================================================

import numpy as np
import pandas as pd
import torch
from torch_geometric.data import Data
from sklearn.neighbors import NearestNeighbors

print("=" * 80)
print("BUILDING GRAPH FOR GRAPHSAGE")
print("=" * 80)


# ------------------------------------------------------------
# 1. CHECK REQUIRED VARIABLES
# ------------------------------------------------------------

required = [
    "X_train_scaled",
    "X_val_scaled",
    "X_test_scaled",
    "y_train",
    "y_val",
    "y_test"
]

missing = [
    name for name in required
    if name not in globals()
]

if missing:
    raise NameError(
        "The following variables are missing: "
        + str(missing)
        + "\nRun your train/validation/test split cell first."
    )


# ------------------------------------------------------------
# 2. CONVERT LABELS TO NUMPY
# ------------------------------------------------------------

def convert_labels(y):

    if isinstance(y, pd.Series):
        return y.to_numpy()

    if isinstance(y, pd.DataFrame):
        return y.iloc[:, 0].to_numpy()

    if torch.is_tensor(y):
        return y.detach().cpu().numpy()

    return np.asarray(y)


y_train_np = convert_labels(y_train)
y_val_np   = convert_labels(y_val)
y_test_np  = convert_labels(y_test)


# ------------------------------------------------------------
# 3. COMBINE DATA
# ------------------------------------------------------------

X_all = np.concatenate(
    [
        X_train_scaled,
        X_val_scaled,
        X_test_scaled
    ],
    axis=0
).astype(np.float32)


y_all = np.concatenate(
    [
        y_train_np,
        y_val_np,
        y_test_np
    ],
    axis=0
)


# Make sure labels are integer
y_all = y_all.astype(np.int64)


# ------------------------------------------------------------
# 4. CREATE TRAIN / VALIDATION / TEST INDICES
# ------------------------------------------------------------

n_train = len(X_train_scaled)
n_val   = len(X_val_scaled)
n_test  = len(X_test_scaled)

train_indices = np.arange(
    0,
    n_train
)

val_indices = np.arange(
    n_train,
    n_train + n_val
)

test_indices = np.arange(
    n_train + n_val,
    n_train + n_val + n_test
)


print("Total nodes:", len(X_all))
print("Training nodes:", len(train_indices))
print("Validation nodes:", len(val_indices))
print("Testing nodes:", len(test_indices))


# ------------------------------------------------------------
# 5. CREATE kNN GRAPH
# ------------------------------------------------------------

# Keep this moderate because you are currently using CPU
K_NEIGHBORS = 10

print("\nBuilding kNN graph...")
print("k =", K_NEIGHBORS)

knn = NearestNeighbors(
    n_neighbors=K_NEIGHBORS + 1,
    metric="euclidean",
    n_jobs=-1
)

knn.fit(X_all)

distances, neighbors = knn.kneighbors(
    X_all
)


# ------------------------------------------------------------
# 6. CREATE EDGE LIST
# ------------------------------------------------------------

source_nodes = []
target_nodes = []

for i in range(len(X_all)):

    # neighbors[0] is the node itself
    for j in neighbors[i, 1:]:

        source_nodes.append(i)
        target_nodes.append(j)


edge_index = torch.tensor(
    [
        source_nodes,
        target_nodes
    ],
    dtype=torch.long
)


# ------------------------------------------------------------
# 7. CREATE PYTORCH GEOMETRIC DATA OBJECT
# ------------------------------------------------------------

x_tensor = torch.tensor(
    X_all,
    dtype=torch.float32
)

y_tensor = torch.tensor(
    y_all,
    dtype=torch.long
)


train_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)

val_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)

test_mask = torch.zeros(
    len(X_all),
    dtype=torch.bool
)


train_mask[train_indices] = True
val_mask[val_indices] = True
test_mask[test_indices] = True


graph_data = Data(
    x=x_tensor,
    y=y_tensor,
    edge_index=edge_index,
    train_mask=train_mask,
    val_mask=val_mask,
    test_mask=test_mask
)


# ------------------------------------------------------------
# 8. DISPLAY GRAPH INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("GRAPH CREATED SUCCESSFULLY")
print("=" * 80)

print(
    "Number of nodes:",
    graph_data.num_nodes
)

print(
    "Number of features:",
    graph_data.num_features
)

print(
    "Number of edges:",
    graph_data.num_edges
)

print(
    "Feature matrix:",
    graph_data.x.shape
)

print(
    "Label vector:",
    graph_data.y.shape
)

print(
    "Training nodes:",
    int(graph_data.train_mask.sum())
)

print(
    "Validation nodes:",
    int(graph_data.val_mask.sum())
)

print(
    "Testing nodes:",
    int(graph_data.test_mask.sum())
)

print("\nLabel distribution:")

unique, counts = torch.unique(
    graph_data.y,
    return_counts=True
)

for label, count in zip(unique, counts):

    print(
        f"Class {int(label)}: "
        f"{int(count)} samples"
    )

print("\n✓ graph_data is ready.")
```

```python
# ============================================================
# NOVEL MODEL
# Adaptive Local GraphSAGE + Feature Attention + BiLSTM
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.nn import SAGEConv


class AdaptiveLocalGNNBiLSTM(nn.Module):

    def __init__(
        self,
        input_dim=46,
        gnn_hidden=128,
        embedding_dim=64,
        lstm_hidden=128,
        lstm_layers=2,
        num_classes=2,
        dropout=0.30
    ):

        super().__init__()

        self.input_dim = input_dim
        self.embedding_dim = embedding_dim

        # ====================================================
        # 1. FEATURE ATTENTION
        # ====================================================

        self.feature_attention = nn.Sequential(

            nn.Linear(
                input_dim,
                input_dim
            ),

            nn.ReLU(),

            nn.Linear(
                input_dim,
                input_dim
            ),

            nn.Sigmoid()
        )

        # ====================================================
        # 2. LOCAL GRAPHSAGE
        # ====================================================

        self.sage1 = SAGEConv(
            input_dim,
            gnn_hidden
        )

        self.sage2 = SAGEConv(
            gnn_hidden,
            embedding_dim
        )

        # ====================================================
        # 3. RESIDUAL PROJECTION
        # ====================================================

        self.residual_projection = nn.Linear(
            input_dim,
            embedding_dim
        )

        # ====================================================
        # 4. NORMALIZATION
        # ====================================================

        self.norm1 = nn.BatchNorm1d(
            gnn_hidden
        )

        self.norm2 = nn.BatchNorm1d(
            embedding_dim
        )

        # ====================================================
        # 5. BILSTM
        # ====================================================

        self.bilstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=lstm_hidden,
            num_layers=lstm_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout
            if lstm_layers > 1
            else 0
        )

        # ====================================================
        # 6. TEMPORAL ATTENTION
        # ====================================================

        self.temporal_attention = nn.Sequential(

            nn.Linear(
                lstm_hidden * 2,
                64
            ),

            nn.Tanh(),

            nn.Linear(
                64,
                1
            )
        )

        # ====================================================
        # 7. FUSION
        # ====================================================

        self.fusion = nn.Sequential(

            nn.Linear(
                embedding_dim + lstm_hidden * 2,
                128
            ),

            nn.ReLU(),

            nn.Dropout(dropout)
        )

        # ====================================================
        # 8. CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Linear(
                128,
                64
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                64,
                num_classes
            )
        )


    def forward_gnn(
        self,
        x,
        edge_index
    ):

        # ====================================================
        # Feature attention
        # ====================================================

        attention_weights = (
            self.feature_attention(x)
        )

        x_attended = (
            x * attention_weights
        )

        # ====================================================
        # Residual branch
        # ====================================================

        residual = self.residual_projection(
            x_attended
        )

        # ====================================================
        # GraphSAGE layer 1
        # ====================================================

        x = self.sage1(
            x_attended,
            edge_index
        )

        x = self.norm1(x)

        x = F.relu(x)

        # ====================================================
        # GraphSAGE layer 2
        # ====================================================

        x = self.sage2(
            x,
            edge_index
        )

        x = self.norm2(x)

        # ====================================================
        # Residual fusion
        # ====================================================

        x = x + residual

        x = F.relu(x)

        return x, attention_weights


    def forward_sequence(
        self,
        sequences
    ):

        # ====================================================
        # BiLSTM
        # ====================================================

        lstm_output, _ = self.bilstm(
            sequences
        )

        # ====================================================
        # Temporal attention
        # ====================================================

        attention_scores = (
            self.temporal_attention(
                lstm_output
            )
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=1
        )

        temporal_embedding = torch.sum(
            lstm_output * attention_weights,
            dim=1
        )

        return (
            temporal_embedding,
            attention_weights
        )


    def classify(
        self,
        gnn_embedding,
        temporal_embedding
    ):

        # ====================================================
        # Spatial + temporal fusion
        # ====================================================

        fused = torch.cat(
            [
                gnn_embedding,
                temporal_embedding
            ],
            dim=1
        )

        fused = self.fusion(
            fused
        )

        logits = self.classifier(
            fused
        )

        return logits, fused


    def forward(
        self,
        x,
        edge_index,
        sequences=None
    ):

        gnn_embedding, feature_attention = (
            self.forward_gnn(
                x,
                edge_index
            )
        )

        if sequences is None:

            logits = self.classifier(
                self.fusion(
                    torch.cat(
                        [
                            gnn_embedding,
                            torch.zeros(
                                gnn_embedding.size(0),
                                self.bilstm.hidden_size * 2,
                                device=gnn_embedding.device
                            )
                        ],
                        dim=1
                    )
                )
            )

            return (
                logits,
                gnn_embedding,
                feature_attention
            )

        temporal_embedding, temporal_attention = (
            self.forward_sequence(
                sequences
            )
        )

        return (
            temporal_embedding,
            feature_attention,
            temporal_attention
        )
```

```python
# ============================================================
# STEP 33A — NOVEL LOCAL GNN + BiLSTM MODEL
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch_geometric.nn import SAGEConv


# ============================================================
# DEVICE
# ============================================================

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)


# ============================================================
# MODEL
# ============================================================

class NovelTemporalClassifier(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dim=128,
        embedding_dim=64,
        num_layers=2,
        num_classes=2,
        dropout=0.30
    ):

        super().__init__()


        # ====================================================
        # LOCAL GRAPH NEURAL NETWORK
        # ====================================================

        self.gnn1 = SAGEConv(
            input_dim,
            hidden_dim
        )

        self.gnn2 = SAGEConv(
            hidden_dim,
            embedding_dim
        )


        # ====================================================
        # NORMALIZATION
        # ====================================================

        self.norm1 = nn.LayerNorm(
            hidden_dim
        )

        self.norm2 = nn.LayerNorm(
            embedding_dim
        )


        # ====================================================
        # LOCAL ATTENTION
        # ====================================================

        self.attention = nn.Sequential(

            nn.Linear(
                embedding_dim,
                embedding_dim // 2
            ),

            nn.Tanh(),

            nn.Linear(
                embedding_dim // 2,
                1
            )
        )


        # ====================================================
        # BIDIRECTIONAL LSTM
        # ====================================================

        self.bilstm = nn.LSTM(

            input_size=embedding_dim,

            hidden_size=hidden_dim,

            num_layers=num_layers,

            batch_first=True,

            bidirectional=True,

            dropout=dropout if num_layers > 1 else 0
        )


        # ====================================================
        # TEMPORAL ATTENTION
        # ====================================================

        self.temporal_attention = nn.Sequential(

            nn.Linear(
                hidden_dim * 2,
                hidden_dim
            ),

            nn.Tanh(),

            nn.Linear(
                hidden_dim,
                1
            )
        )


        # ====================================================
        # RESIDUAL FUSION
        # ====================================================

        self.gnn_projection = nn.Linear(

            embedding_dim,

            hidden_dim * 2
        )


        self.fusion = nn.Sequential(

            nn.Linear(
                hidden_dim * 4,
                hidden_dim * 2
            ),

            nn.ReLU(),

            nn.Dropout(dropout)
        )


        # ====================================================
        # CLASSIFIER
        # ====================================================

        self.classifier = nn.Sequential(

            nn.Linear(
                hidden_dim * 2,
                hidden_dim
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                hidden_dim,
                num_classes
            )
        )


        self.dropout = nn.Dropout(
            dropout
        )


    # ========================================================
    # FORWARD
    # ========================================================

    def forward(
        self,
        x,
        edge_index=None
    ):

        # ----------------------------------------------------
        # GRAPH LAYER
        # ----------------------------------------------------

        if edge_index is not None:

            x1 = self.gnn1(
                x,
                edge_index
            )

        else:

            x1 = self.gnn1(
                x
            )


        x1 = self.norm1(
            x1
        )

        x1 = F.relu(
            x1
        )

        x1 = self.dropout(
            x1
        )


        # ----------------------------------------------------
        # SECOND GRAPH LAYER
        # ----------------------------------------------------

        if edge_index is not None:

            x2 = self.gnn2(
                x1,
                edge_index
            )

        else:

            x2 = self.gnn2(
                x1
            )


        x2 = self.norm2(
            x2
        )

        x2 = F.relu(
            x2
        )


        # ----------------------------------------------------
        # GRAPH ATTENTION
        # ----------------------------------------------------

        attention_scores = self.attention(
            x2
        )

        attention_weights = torch.softmax(
            attention_scores,
            dim=0
        )

        attended = (
            x2 *
            attention_weights
        )


        # ----------------------------------------------------
        # SEQUENCE INPUT
        #
        # If input is [N, F], convert to [N, 1, F]
        # ----------------------------------------------------

        if attended.dim() == 2:

            sequence = attended.unsqueeze(
                1
            )

        else:

            sequence = attended


        # ----------------------------------------------------
        # BiLSTM
        # ----------------------------------------------------

        lstm_output, _ = self.bilstm(
            sequence
        )


        # ----------------------------------------------------
        # TEMPORAL ATTENTION
        # ----------------------------------------------------

        temporal_scores = (
            self.temporal_attention(
                lstm_output
            )
        )

        temporal_weights = torch.softmax(
            temporal_scores,
            dim=1
        )


        temporal_context = (
            lstm_output *
            temporal_weights
        ).sum(
            dim=1
        )


        # ----------------------------------------------------
        # GNN RESIDUAL REPRESENTATION
        # ----------------------------------------------------

        gnn_residual = self.gnn_projection(
            x2
        )


        # ----------------------------------------------------
        # RESIDUAL FUSION
        # ----------------------------------------------------

        fused = torch.cat(
            [
                temporal_context,
                gnn_residual
            ],
            dim=1
        )


        fused = self.fusion(
            fused
        )


        # ----------------------------------------------------
        # CLASSIFICATION
        # ----------------------------------------------------

        output = self.classifier(
            fused
        )


        return output


# ============================================================
# TEST MODEL CREATION
# ============================================================

novel_model = NovelTemporalClassifier(

    input_dim=graph_data.num_features,

    hidden_dim=128,

    embedding_dim=64,

    num_layers=2,

    num_classes=2,

    dropout=0.30

).to(DEVICE)


# ============================================================
# MODEL INFORMATION
# ============================================================

print("=" * 90)
print("NOVEL MODEL CREATED SUCCESSFULLY")
print("=" * 90)

print(novel_model)


# ============================================================
# PARAMETER COUNT
# ============================================================

total_parameters = sum(
    p.numel()
    for p in novel_model.parameters()
)

trainable_parameters = sum(
    p.numel()
    for p in novel_model.parameters()
    if p.requires_grad
)

print("\nTotal parameters:",
      f"{total_parameters:,}")

print(
    "Trainable parameters:",
    f"{trainable_parameters:,}"
)
```

```python
# ============================================================
# STEP 34 — TRAIN PROPOSED LOCAL GNN + BiLSTM MODEL
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)


# ------------------------------------------------------------
# Move graph to device
# ------------------------------------------------------------

graph_data = graph_data.to(DEVICE)


# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

novel_model = NovelTemporalClassifier(
    input_dim=graph_data.num_features,
    hidden_dim=128,
    embedding_dim=64,
    num_layers=2,
    num_classes=2,
    dropout=0.30
).to(DEVICE)


# ------------------------------------------------------------
# Loss
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss()


# ------------------------------------------------------------
# Optimizer
# ------------------------------------------------------------

optimizer = torch.optim.AdamW(
    novel_model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)


# ------------------------------------------------------------
# Scheduler
# ------------------------------------------------------------

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=3
)


# ------------------------------------------------------------
# Training settings
# ------------------------------------------------------------

EPOCHS = 30

PATIENCE = 7

best_val_f1 = -1.0

best_epoch = 0

patience_counter = 0


# ------------------------------------------------------------
# History
# ------------------------------------------------------------

TRAIN_HISTORY = []


# ------------------------------------------------------------
# Model save path
# ------------------------------------------------------------

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "FedGRAI_LocalGNN_BiLSTM_Best.pt"
)


# ============================================================
# TRAINING LOOP
# ============================================================

print("=" * 100)
print("STARTING TRAINING")
print("=" * 100)


for epoch in range(1, EPOCHS + 1):

    # ========================================================
    # TRAIN
    # ========================================================

    novel_model.train()

    optimizer.zero_grad(
        set_to_none=True
    )

    logits = novel_model(
        graph_data.x,
        graph_data.edge_index
    )

    train_logits = logits[
        graph_data.train_mask
    ]

    train_labels = graph_data.y[
        graph_data.train_mask
    ]

    train_loss = criterion(
        train_logits,
        train_labels
    )

    train_loss.backward()

    torch.nn.utils.clip_grad_norm_(
        novel_model.parameters(),
        max_norm=5.0
    )

    optimizer.step()


    # ========================================================
    # TRAIN PREDICTIONS
    # ========================================================

    train_predictions = torch.argmax(
        train_logits.detach(),
        dim=1
    )


    train_true_np = (
        train_labels.detach()
        .cpu()
        .numpy()
    )

    train_pred_np = (
        train_predictions.cpu()
        .numpy()
    )


    # ========================================================
    # TRAIN METRICS
    # ========================================================

    train_accuracy = accuracy_score(
        train_true_np,
        train_pred_np
    )

    train_precision = precision_score(
        train_true_np,
        train_pred_np,
        zero_division=0
    )

    train_recall = recall_score(
        train_true_np,
        train_pred_np,
        zero_division=0
    )

    train_f1 = f1_score(
        train_true_np,
        train_pred_np,
        zero_division=0
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    novel_model.eval()

    with torch.no_grad():

        validation_logits = novel_model(
            graph_data.x,
            graph_data.edge_index
        )

        val_logits = validation_logits[
            graph_data.val_mask
        ]

        val_labels = graph_data.y[
            graph_data.val_mask
        ]

        val_loss = criterion(
            val_logits,
            val_labels
        )


        val_predictions = torch.argmax(
            val_logits,
            dim=1
        )


    # ========================================================
    # VALIDATION METRICS
    # ========================================================

    val_true_np = (
        val_labels.cpu()
        .numpy()
    )

    val_pred_np = (
        val_predictions.cpu()
        .numpy()
    )


    val_accuracy = accuracy_score(
        val_true_np,
        val_pred_np
    )

    val_precision = precision_score(
        val_true_np,
        val_pred_np,
        zero_division=0
    )

    val_recall = recall_score(
        val_true_np,
        val_pred_np,
        zero_division=0
    )

    val_f1 = f1_score(
        val_true_np,
        val_pred_np,
        zero_division=0
    )


    # ========================================================
    # SCHEDULER
    # ========================================================

    scheduler.step(
        val_f1
    )

    current_lr = (
        optimizer.param_groups[0]["lr"]
    )


    # ========================================================
    # STORE HISTORY
    # ========================================================

    TRAIN_HISTORY.append({

        "Epoch": epoch,

        "Train Loss":
            train_loss.item(),

        "Validation Loss":
            val_loss.item(),

        "Train Accuracy":
            train_accuracy,

        "Validation Accuracy":
            val_accuracy,

        "Train Precision":
            train_precision,

        "Validation Precision":
            val_precision,

        "Train Recall":
            train_recall,

        "Validation Recall":
            val_recall,

        "Train F1":
            train_f1,

        "Validation F1":
            val_f1,

        "Learning Rate":
            current_lr
    })


    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    if val_f1 > best_val_f1:

        best_val_f1 = val_f1

        best_epoch = epoch

        patience_counter = 0

        torch.save(
            {
                "epoch": epoch,

                "model_state_dict":
                    novel_model.state_dict(),

                "optimizer_state_dict":
                    optimizer.state_dict(),

                "validation_f1":
                    val_f1,

                "validation_accuracy":
                    val_accuracy,

                "validation_precision":
                    val_precision,

                "validation_recall":
                    val_recall
            },
            MODEL_PATH
        )

    else:

        patience_counter += 1


    # ========================================================
    # PRINT EPOCH
    # ========================================================

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss.item():.5f} | "
        f"Val Loss: {val_loss.item():.5f} | "
        f"Train Acc: {train_accuracy*100:.2f}% | "
        f"Val Acc: {val_accuracy*100:.2f}% | "
        f"Train F1: {train_f1:.4f} | "
        f"Val F1: {val_f1:.4f} | "
        f"LR: {current_lr:.6f}"
    )


    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if patience_counter >= PATIENCE:

        print(
            "\nEarly stopping activated."
        )

        break


# ============================================================
# TRAINING COMPLETE
# ============================================================

print("\n" + "=" * 100)
print("TRAINING COMPLETED")
print("=" * 100)

print(
    "Best Epoch:",
    best_epoch
)

print(
    "Best Validation F1:",
    f"{best_val_f1:.4f}"
)

print(
    "Saved Model:",
    MODEL_PATH
)
```

```python
# ============================================================
# STEP 35 — FINAL TEST EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    matthews_corrcoef
)

# ------------------------------------------------------------
# Load best checkpoint
# ------------------------------------------------------------

checkpoint = torch.load(
    MODEL_PATH,
    map_location=DEVICE
)

novel_model.load_state_dict(
    checkpoint["model_state_dict"]
)

novel_model.eval()


# ------------------------------------------------------------
# Test inference
# ------------------------------------------------------------

with torch.no_grad():

    all_logits = novel_model(
        graph_data.x,
        graph_data.edge_index
    )

    test_logits = all_logits[
        graph_data.test_mask
    ]

    test_labels = graph_data.y[
        graph_data.test_mask
    ]

    test_probabilities = torch.softmax(
        test_logits,
        dim=1
    )

    test_predictions = torch.argmax(
        test_probabilities,
        dim=1
    )


# ------------------------------------------------------------
# Convert to NumPy
# ------------------------------------------------------------

y_true = (
    test_labels.cpu()
    .numpy()
)

y_pred = (
    test_predictions.cpu()
    .numpy()
)

y_prob = (
    test_probabilities[:, 1]
    .cpu()
    .numpy()
)


# ============================================================
# METRICS
# ============================================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_true,
    y_prob
)

pr_auc = average_precision_score(
    y_true,
    y_prob
)

mcc = matthews_corrcoef(
    y_true,
    y_pred
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)

TN, FP, FN, TP = cm.ravel()


# ------------------------------------------------------------
# IDS-specific metrics
# ------------------------------------------------------------

specificity = (
    TN / (TN + FP)
    if (TN + FP) > 0
    else 0
)

FAR = (
    FP / (FP + TN)
    if (FP + TN) > 0
    else 0
)

FRR = (
    FN / (FN + TP)
    if (FN + TP) > 0
    else 0
)

balanced_accuracy = (
    recall + specificity
) / 2


# ============================================================
# FINAL RESULT
# ============================================================

FINAL_RESULTS = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "ROC-AUC",

        "PR-AUC",

        "Specificity",

        "False Acceptance Rate",

        "False Rejection Rate",

        "Balanced Accuracy",

        "MCC"
    ],

    "Value": [

        accuracy,

        precision,

        recall,

        f1,

        roc_auc,

        pr_auc,

        specificity,

        FAR,

        FRR,

        balanced_accuracy,

        mcc
    ],

    "Percentage": [

        accuracy * 100,

        precision * 100,

        recall * 100,

        f1 * 100,

        roc_auc * 100,

        pr_auc * 100,

        specificity * 100,

        FAR * 100,

        FRR * 100,

        balanced_accuracy * 100,

        np.nan
    ]
})


print("=" * 100)
print("FINAL TEST RESULTS — LOCAL GNN + BiLSTM")
print("=" * 100)

display(
    FINAL_RESULTS.style.format({
        "Value": "{:.6f}",
        "Percentage": "{:.2f}"
    })
)


# ============================================================
# CONFUSION MATRIX VALUES
# ============================================================

print("\nConfusion Matrix:")
print(cm)

print("\nTN:", TN)
print("FP:", FP)
print("FN:", FN)
print("TP:", TP)


# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 100)
print("CLASSIFICATION REPORT")
print("=" * 100)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=[
            "Benign",
            "Attack"
        ],
        digits=4,
        zero_division=0
    )
)
```

```python
# ============================================================
# PREPROCESSING EFFECT FIGURE
# Missing Values + Pearson Correlation
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

from sklearn.feature_selection import VarianceThreshold


# ============================================================
# SETTINGS
# ============================================================

FIG_DIR = "/content/drive/MyDrive/FedGRAI_LocalGNN_BiLSTM_Results"

os.makedirs(
    FIG_DIR,
    exist_ok=True
)

SEED = 42

# Number of rows used for visualization
# Using all 200,000 rows can make heatmaps slow.
HEATMAP_SAMPLES = 3000

# Correlation threshold
CORR_THRESHOLD = 0.90


# ============================================================
# FIND AVAILABLE DATA
# ============================================================

print("=" * 90)
print("SEARCHING FOR PREPROCESSING DATA")
print("=" * 90)


# ------------------------------------------------------------
# Possible raw datasets
# ------------------------------------------------------------

raw_candidates = [
    "CIC_DATA",
    "CIC_SAMPLE",
    "df",
    "data",
    "DATA"
]

raw_df = None
raw_name = None

for name in raw_candidates:

    if name in globals():

        obj = globals()[name]

        if isinstance(obj, pd.DataFrame):

            raw_df = obj.copy()
            raw_name = name

            print(
                f"✓ Raw dataframe found: {name}"
            )

            break


# ------------------------------------------------------------
# Possible original feature matrices
# ------------------------------------------------------------

feature_candidates = [
    "X",
    "X_data",
    "X_numeric",
    "X_scaled",
    "X_selected",
    "X_encoded",
    "X_train",
    "X_train_missing"
]

feature_matrix = None
feature_name = None

for name in feature_candidates:

    if name in globals():

        obj = globals()[name]

        if isinstance(
            obj,
            (
                np.ndarray,
                pd.DataFrame
            )
        ):

            feature_matrix = obj
            feature_name = name

            print(
                f"✓ Feature matrix found: {name}"
            )

            break


# ============================================================
# IF RAW DATAFRAME EXISTS
# ============================================================

if raw_df is not None:

    print(
        "\nUsing raw dataframe:",
        raw_name
    )

    # --------------------------------------------------------
    # Remove label columns
    # --------------------------------------------------------

    exclude_columns = [
        "label",
        "Label",
        "binary_label",
        "Binary_Label",
        "attack",
        "Attack"
    ]

    numeric_df = raw_df.select_dtypes(
        include=[np.number]
    ).copy()

    print(
        "Numeric features:",
        numeric_df.shape[1]
    )

    # --------------------------------------------------------
    # Remove labels if numeric
    # --------------------------------------------------------

    for col in exclude_columns:

        if col in numeric_df.columns:

            numeric_df = numeric_df.drop(
                columns=[col]
            )

    # --------------------------------------------------------
    # Original matrix
    # --------------------------------------------------------

    X_before_df = numeric_df.copy()

else:

    # ========================================================
    # FALLBACK TO FEATURE MATRIX
    # ========================================================

    if feature_matrix is None:

        raise NameError(
            """
            Could not find your original feature matrix.

            Please make sure one of these variables exists:

            CIC_DATA
            CIC_SAMPLE
            X
            X_data
            X_numeric
            X_scaled
            X_selected
            X_encoded
            """
        )

    if isinstance(
        feature_matrix,
        np.ndarray
    ):

        X_before_df = pd.DataFrame(
            feature_matrix
        )

    else:

        X_before_df = feature_matrix.copy()

    print(
        "\nUsing feature matrix:",
        feature_name
    )


# ============================================================
# CONVERT TO NUMERIC
# ============================================================

X_before_df = X_before_df.apply(
    pd.to_numeric,
    errors="coerce"
)

# Remove completely empty columns
X_before_df = X_before_df.dropna(
    axis=1,
    how="all"
)

print(
    "\nOriginal feature matrix:",
    X_before_df.shape
)


# ============================================================
# SAMPLE FOR VISUALIZATION
# ============================================================

rng = np.random.default_rng(
    SEED
)

n_rows = min(
    HEATMAP_SAMPLES,
    len(X_before_df)
)

sample_indices = rng.choice(
    len(X_before_df),
    size=n_rows,
    replace=False
)

X_plot_before = (
    X_before_df.iloc[
        sample_indices
    ]
    .copy()
)


# ============================================================
# TOP-LEFT:
# MISSING VALUES BEFORE
# ============================================================

missing_before = (
    X_plot_before.isna()
    .astype(int)
)


# ============================================================
# IDENTIFY IMPUTED / CLEAN DATA
# ============================================================

# ------------------------------------------------------------
# Try to find an already-created imputed matrix
# ------------------------------------------------------------

imputed_candidates = [

    "X_imputed",
    "X_clean",
    "X_cleaned",
    "X_imputed_scaled",
    "X_processed",
    "X_preprocessed",
    "X_encoded",
    "X_scaled"
]

X_after_df = None
after_name = None

for name in imputed_candidates:

    if name in globals():

        obj = globals()[name]

        if isinstance(
            obj,
            (
                np.ndarray,
                pd.DataFrame
            )
        ):

            # Check dimensions
            if (
                len(obj) == len(X_before_df)
            ):

                X_after_df = (
                    pd.DataFrame(obj)
                    if isinstance(
                        obj,
                        np.ndarray
                    )
                    else obj.copy()
                )

                after_name = name

                print(
                    f"✓ Existing processed matrix found: {name}"
                )

                break


# ============================================================
# FALLBACK:
# CREATE CLEAN VERSION USING MEDIAN
# ============================================================

if X_after_df is None:

    print(
        "\n⚠ No explicit imputed matrix found."
    )

    print(
        "Creating a median-imputed visualization matrix."
    )

    X_after_df = X_before_df.copy()

    X_after_df = X_after_df.fillna(
        X_after_df.median(
            numeric_only=True
        )
    )

    after_name = "Median-imputed fallback"


# ------------------------------------------------------------
# Align columns
# ------------------------------------------------------------

X_after_df = X_after_df.select_dtypes(
    include=[np.number]
)

X_after_df = X_after_df.replace(
    [np.inf, -np.inf],
    np.nan
)

X_after_df = X_after_df.fillna(
    X_after_df.median()
)


# ============================================================
# SAMPLE PROCESSED DATA
# ============================================================

X_plot_after = (
    X_after_df.iloc[
        sample_indices
    ]
    .copy()
)


# ============================================================
# TOP-RIGHT:
# MISSING VALUES AFTER
# ============================================================

missing_after = (
    X_plot_after.isna()
    .astype(int)
)


# ============================================================
# CORRELATION BEFORE
# ============================================================

print(
    "\nCalculating Pearson correlation..."
)

# Remove constant columns
variance_filter = VarianceThreshold(
    threshold=0
)

X_corr_before_array = (
    variance_filter.fit_transform(
        X_plot_after
    )
)

corr_feature_names = (
    X_plot_after.columns[
        variance_filter.get_support()
    ]
)

X_corr_before = pd.DataFrame(
    X_corr_before_array,
    columns=corr_feature_names
)

corr_before = (
    X_corr_before
    .corr(method="pearson")
)




upper_triangle = np.triu(
    np.ones(
        corr_before.shape,
        dtype=bool
    ),
    k=1
)

upper_corr = (
    corr_before
    .where(upper_triangle)
)

high_corr_columns = [

    column

    for column in upper_corr.columns

    if any(
        upper_corr[column]
        .abs()
        > CORR_THRESHOLD
    )
]


print(
    "\nHighly correlated features removed:",
    len(high_corr_columns)
)




X_corr_after = (
    X_corr_before
    .drop(
        columns=high_corr_columns,
        errors="ignore"
    )
)



corr_after = (
    X_corr_after
    .corr(method="pearson")
)


print(
    "Features before:",
    X_corr_before.shape[1]
)

print(
    "Features after:",
    X_corr_after.shape[1]
)




fig, axes = plt.subplots(
    2,
    2,
    figsize=(18, 15)
)




sns.heatmap(
    missing_before.T,
    cmap="Blues",
    cbar=False,
    ax=axes[0, 0],
    xticklabels=False,
    yticklabels=True
)

axes[0, 0].set_title(
    "Before Imputation",
    fontsize=16,
    fontweight="bold"
)

axes[0, 0].set_xlabel(
    "Samples"
)

axes[0, 0].set_ylabel(
    "Features"
)


# ============================================================
# TOP RIGHT — MISSING VALUES AFTER
# ============================================================

sns.heatmap(
    missing_after.T,
    cmap="Blues",
    cbar=False,
    ax=axes[0, 1],
    xticklabels=False,
    yticklabels=True
)

axes[0, 1].set_title(
    "After Imputation and Cleaning",
    fontsize=16,
    fontweight="bold"
)

axes[0, 1].set_xlabel(
    "Samples"
)

axes[0, 1].set_ylabel(
    "Features"
)


# ============================================================
# BOTTOM LEFT — CORRELATION BEFORE
# ============================================================

sns.heatmap(
    corr_before,
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    ax=axes[1, 0],
    cbar_kws={
        "label": "Pearson ρ"
    }
)

axes[1, 0].set_title(
    "Pearson Correlation Before Feature Removal",
    fontsize=16,
    fontweight="bold"
)

axes[1, 0].set_xlabel(
    "Features"
)

axes[1, 0].set_ylabel(
    "Features"
)


# ============================================================
# BOTTOM RIGHT — CORRELATION AFTER
# ============================================================

sns.heatmap(
    corr_after,
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    ax=axes[1, 1],
    cbar_kws={
        "label": "Pearson ρ"
    }
)

axes[1, 1].set_title(
    f"Pearson Correlation After Removal "
    f"(|ρ| > {CORR_THRESHOLD})",
    fontsize=16,
    fontweight="bold"
)

axes[1, 1].set_xlabel(
    "Remaining Features"
)

axes[1, 1].set_ylabel(
    "Remaining Features"
)


# ============================================================
# MAIN TITLE
# ============================================================

fig.suptitle(
    "Effect of Preprocessing on the Dataset",
    fontsize=21,
    fontweight="bold",
    y=0.995
)


# ============================================================
# SUBTITLE / DESCRIPTION
# ============================================================

fig.text(
    0.5,
    0.965,
    "Missing-value analysis and Pearson correlation-based "
    "feature reduction",
    ha="center",
    fontsize=13
)


# ============================================================
# LAYOUT
# ============================================================

plt.tight_layout(
    rect=[
        0,
        0,
        1,
        0.94
    ]
)


# ============================================================
# SAVE
# ============================================================

FIG_PATH = os.path.join(
    FIG_DIR,
    "Figure_Preprocessing_Effect.png"
)

plt.savefig(
    FIG_PATH,
    dpi=400,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 90)
print("PREPROCESSING FIGURE GENERATED")
print("=" * 90)

print(
    "Original features:",
    X_before_df.shape[1]
)

print(
    "Features before correlation filtering:",
    X_corr_before.shape[1]
)

print(
    "Highly correlated features removed:",
    len(high_corr_columns)
)

print(
    "Remaining features:",
    X_corr_after.shape[1]
)

print(
    "Missing cells before:",
    missing_before.values.sum()
)

print(
    "Missing cells after:",
    missing_after.values.sum()
)

print(
    "\nSaved to:"
)

print(
    FIG_PATH
)
```

```python
# ============================================================
# STEP — ALL PAIRWISE TRAINING / VALIDATION GRAPHS
# ============================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

FIG_DIR = "/content/drive/MyDrive/FedGRAI_LocalGNN_BiLSTM_Results"
os.makedirs(FIG_DIR, exist_ok=True)

# ------------------------------------------------------------
# Load actual history
# ------------------------------------------------------------

if "TRAIN_HISTORY" in globals():
    history_df = pd.DataFrame(TRAIN_HISTORY)

elif "TRAINING_HISTORY" in globals():
    history_df = pd.DataFrame(TRAINING_HISTORY)

else:
    history_df = pd.read_csv(
        os.path.join(
            FIG_DIR,
            "Training_History.csv"
        )
    )

print("=" * 90)
print("PAIRWISE TRAINING / VALIDATION RESULTS")
print("=" * 90)

print("Epochs:", len(history_df))

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def plot_pair(
    train_col,
    val_col,
    title,
    ylabel,
    filename,
    percentage=True
):

    plt.figure(
        figsize=(10, 6)
    )

    train_values = history_df[train_col].astype(float)
    val_values = history_df[val_col].astype(float)

    if percentage:
        train_values = train_values * 100
        val_values = val_values * 100

    plt.plot(
        history_df["Epoch"],
        train_values,
        marker="o",
        linewidth=2,
        label="Training"
    )

    plt.plot(
        history_df["Epoch"],
        val_values,
        marker="s",
        linewidth=2,
        label="Validation"
    )

    plt.xlabel("Epoch")
    plt.ylabel(ylabel)

    plt.title(title)

    plt.xlim(
        0,
        history_df["Epoch"].max()
    )

    plt.ylim(
        0,
        100 if percentage else None
    )

    plt.grid(
        True,
        alpha=0.3
    )

    plt.legend()

    plt.tight_layout()

    path = os.path.join(
        FIG_DIR,
        filename
    )

    plt.savefig(
        path,
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()

    print("✓", filename)


# ============================================================
# 1. LOSS
# ============================================================

plot_pair(
    "Train Loss",
    "Validation Loss",
    "Training vs Validation Loss",
    "Loss",
    "Pair_Training_Validation_Loss.png",
    percentage=False
)


# ============================================================
# 2. ACCURACY
# ============================================================

plot_pair(
    "Train Accuracy",
    "Validation Accuracy",
    "Training vs Validation Accuracy",
    "Accuracy (%)",
    "Pair_Training_Validation_Accuracy.png"
)


# ============================================================
# 3. PRECISION
# ============================================================

plot_pair(
    "Train Precision",
    "Validation Precision",
    "Training vs Validation Precision",
    "Precision (%)",
    "Pair_Training_Validation_Precision.png"
)


# ============================================================
# 4. RECALL
# ============================================================

plot_pair(
    "Train Recall",
    "Validation Recall",
    "Training vs Validation Recall",
    "Recall (%)",
    "Pair_Training_Validation_Recall.png"
)


# ============================================================
# 5. F1
# ============================================================

plot_pair(
    "Train F1",
    "Validation F1",
    "Training vs Validation F1-Score",
    "F1-Score (%)",
    "Pair_Training_Validation_F1.png"
)


# ============================================================
# SHOW EVERY EPOCH
# ============================================================

print("\n" + "=" * 90)
print("COMPLETE TRAINING HISTORY")
print("=" * 90)

display(
    history_df.style.format({
        "Train Loss": "{:.6f}",
        "Validation Loss": "{:.6f}",
        "Train Accuracy": "{:.6f}",
        "Validation Accuracy": "{:.6f}",
        "Train Precision": "{:.6f}",
        "Validation Precision": "{:.6f}",
        "Train Recall": "{:.6f}",
        "Validation Recall": "{:.6f}",
        "Train F1": "{:.6f}",
        "Validation F1": "{:.6f}",
    })
)
```

```python


import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# OUTPUT DIRECTORY
# ============================================================

FIG_DIR = "/content/drive/MyDrive/FedGRAuuI_LocalGNN_BiLSTM_Results"

os.makedirs(
    FIG_DIR,
    exist_ok=True
)

# ============================================================
# LOAD TRAINING HISTORY
# ============================================================

if "TRAIN_HISTORY" in globals():

    history_df = pd.DataFrame(
        TRAIN_HISTORY
    )

elif "TRAINING_HISTORY" in globals():

    history_df = pd.DataFrame(
        TRAINING_HISTORY
    )

else:

    history_path = os.path.join(
        FIG_DIR,
        "Training_History.csv"
    )

    if not os.path.exists(history_path):

        raise FileNotFoundError(
            "Training_History.csv was not found."
        )

    history_df = pd.read_csv(
        history_path
    )


# ============================================================
# CHECK COLUMNS
# ============================================================

print("=" * 90)
print("AVAILABLE TRAINING HISTORY COLUMNS")
print("=" * 90)

print(
    history_df.columns.tolist()
)

print(
    "\nNumber of epochs:",
    len(history_df)
)


# ============================================================
# MAKE SURE EPOCH EXISTS
# ============================================================

if "Epoch" not in history_df.columns:

    history_df.insert(
        0,
        "Epoch",
        range(
            1,
            len(history_df) + 1
        )
    )


# ============================================================
# GRAPH FUNCTION
# ============================================================

def draw_training_validation_graph(
    train_column,
    validation_column,
    title,
    ylabel,
    filename,
    percentage=True
):

    # --------------------------------------------------------
    # Check columns
    # --------------------------------------------------------

    if train_column not in history_df.columns:

        print(
            f"WARNING: {train_column} not found."
        )

        return


    if validation_column not in history_df.columns:

        print(
            f"WARNING: {validation_column} not found."
        )

        return


    # --------------------------------------------------------
    # Convert values to numeric
    # --------------------------------------------------------

    epochs = pd.to_numeric(
        history_df["Epoch"],
        errors="coerce"
    )

    train_values = pd.to_numeric(
        history_df[train_column],
        errors="coerce"
    )

    validation_values = pd.to_numeric(
        history_df[validation_column],
        errors="coerce"
    )


    # --------------------------------------------------------
    # Remove invalid values
    # --------------------------------------------------------

    valid = (
        epochs.notna()
        &
        train_values.notna()
        &
        validation_values.notna()
    )

    epochs = epochs[valid]

    train_values = train_values[valid]

    validation_values = validation_values[valid]


    # --------------------------------------------------------
    # Percentage conversion
    # --------------------------------------------------------

    if percentage:

        train_plot = (
            train_values * 100
        )

        validation_plot = (
            validation_values * 100
        )

    else:

        train_plot = train_values

        validation_plot = validation_values


    # ========================================================
    # CREATE GRAPH
    # ========================================================

    plt.figure(
        figsize=(10, 6)
    )


    # ========================================================
    # TRAINING LINE
    # ========================================================

    plt.plot(
        epochs,
        train_plot,
        marker="o",
        linewidth=2,
        markersize=5,
        label="Training"
    )


    # ========================================================
    # VALIDATION LINE
    # ========================================================

    plt.plot(
        epochs,
        validation_plot,
        marker="s",
        linewidth=2,
        markersize=5,
        label="Validation"
    )


    # ========================================================
    # LABELS
    # ========================================================

    plt.xlabel(
        "Epoch",
        fontsize=12
    )

    plt.ylabel(
        ylabel,
        fontsize=12
    )

    plt.title(
        title,
        fontsize=15,
        fontweight="bold"
    )


    # ========================================================
    # X AXIS STARTS FROM ZERO
    # ========================================================

    plt.xlim(
        left=0
    )


    # ========================================================
    # Y AXIS STARTS FROM ZERO
    # ========================================================

    maximum = max(
        float(train_plot.max()),
        float(validation_plot.max())
    )


    if percentage:

        # Never go below 100%
        upper_limit = max(
            100,
            maximum * 1.10
        )

        plt.ylim(
            0,
            upper_limit
        )

    else:

        upper_limit = (
            maximum * 1.10
        )

        # Handle zero-loss edge case
        if upper_limit <= 0:

            upper_limit = 1

        plt.ylim(
            0,
            upper_limit
        )


    # ========================================================
    # GRID
    # ========================================================

    plt.grid(
        True,
        alpha=0.3
    )


    # ========================================================
    # LEGEND
    # ========================================================

    plt.legend(
        fontsize=11
    )


    # ========================================================
    # SAVE
    # ========================================================

    output_path = os.path.join(
        FIG_DIR,
        filename
    )

    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()


    # ========================================================
    # PRINT EVERY POINT
    # ========================================================

    print(
        f"\n{title}"
    )

    print(
        "-" * 70
    )

    for e, tr, va in zip(
        epochs,
        train_plot,
        validation_plot
    ):

        print(
            f"Epoch {int(e):02d} | "
            f"Training: {tr:.4f} | "
            f"Validation: {va:.4f}"
        )

    print(
        "\n✓ Saved:",
        output_path
    )


# ============================================================
# GRAPH 1 — LOSS
# ============================================================

draw_training_validation_graph(
    train_column="Train Loss",
    validation_column="Validation Loss",
    title="Training vs Validation Loss",
    ylabel="Loss",
    filename="Training_vs_Validation_Loss.png",
    percentage=False
)


# ============================================================
# GRAPH 2 — ACCURACY
# ============================================================

draw_training_validation_graph(
    train_column="Train Accuracy",
    validation_column="Validation Accuracy",
    title="Training vs Validation Accuracy",
    ylabel="Accuracy (%)",
    filename="Training_vs_Validation_Accuracy.png",
    percentage=True
)


# ============================================================
# GRAPH 3 — PRECISION
# ============================================================

draw_training_validation_graph(
    train_column="Train Precision",
    validation_column="Validation Precision",
    title="Training vs Validation Precision",
    ylabel="Precision (%)",
    filename="Training_vs_Validation_Precision.png",
    percentage=True
)


# ============================================================
# GRAPH 4 — RECALL
# ============================================================

draw_training_validation_graph(
    train_column="Train Recall",
    validation_column="Validation Recall",
    title="Training vs Validation Recall",
    ylabel="Recall (%)",
    filename="Training_vs_Validation_Recall.png",
    percentage=True
)


# ============================================================
# GRAPH 5 — F1 SCORE
# ============================================================

draw_training_validation_graph(
    train_column="Train F1",
    validation_column="Validation F1",
    title="Training vs Validation F1-Score",
    ylabel="F1-Score (%)",
    filename="Training_vs_Validation_F1.png",
    percentage=True
)


# ============================================================
# DISPLAY TABLE
# ============================================================

print("\n")
print("=" * 90)
print("COMPLETE TRAINING HISTORY")
print("=" * 90)

display(
    history_df
)
```

# Federated, Privacy-Preserving and Scalable Extension

This extension keeps the uploaded Local GNN + BiLSTM pipeline and adds client-based federated training. Raw client samples remain local; only model updates are aggregated with weighted FedAvg/FedProx. A configurable client-update clipping and Gaussian-noise mechanism is included for privacy-preserving experiments. Scalability is evaluated by changing the number of clients and measuring rounds, communication volume, runtime, and validation performance.

**Important:** the privacy mechanism below is an empirical client-update perturbation mechanism. It should not be reported as a formal sample-level differential-privacy guarantee unless an appropriate privacy accountant and threat model are added.

**Revision notes:** The uploaded final classifier executes 2x GraphSAGE + feature attention + BiLSTM + temporal attention + fusion; TGAT is not in the executable path. The classifier receives `[N,F]` and internally creates `[N,1,F]`, so the BiLSTM is active but the classifier path currently has sequence length 1. The separate BiLSTM imputer is defined with sequence length 10, but its training/evaluation is not completed. The updated suite below records these facts and does not over-claim them.


```python
# ============================================================
# FL-1 — CONFIGURATION
# ============================================================

import os
import time
import copy
import math
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

from torch_geometric.utils import subgraph

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

FL_RESULT_DIR = "/content/drive/MyDrive/FedGRAI_Federated_Privacy_Scalability"
FL_GRAPH_DIR = os.path.join(FL_RESULT_DIR, "Graphs")
os.makedirs(FL_RESULT_DIR, exist_ok=True)
os.makedirs(FL_GRAPH_DIR, exist_ok=True)

# Main federated settings
NUM_CLIENTS = 50
CLIENT_FRACTION = 1.0
FL_ROUNDS = 20
LOCAL_EPOCHS = 1
LOCAL_LR = 0.001
WEIGHT_DECAY = 1e-5

# FedProx regularization; set 0.0 for plain FedAvg
FEDPROX_MU = 0.01

# Privacy-preserving client-update mechanism
PRIVACY_ENABLED = False
UPDATE_CLIP_NORM = 1.0
NOISE_MULTIPLIER = 0.0

# Scalability controls
MAX_NODES_PER_CLIENT = 30000
MAX_EDGES_PER_CLIENT = 250000
SCALABILITY_CLIENT_COUNTS = [2, 5, 10]
# OPTIONAL LARGE-SCALE RUNS: only enable after a real resource preflight.
OPTIONAL_SCALABILITY_CLIENT_COUNTS = [20, 50]

print("=" * 100)
print("FEDERATED LEARNING CONFIGURATION")
print("=" * 100)
print("Clients:", NUM_CLIENTS)
print("Federated rounds:", FL_ROUNDS)
print("Local epochs:", LOCAL_EPOCHS)
print("Algorithm:", "FedProx" if FEDPROX_MU > 0 else "FedAvg")
print("Privacy update clipping:", PRIVACY_ENABLED)
print("Update clip norm:", UPDATE_CLIP_NORM)
print("Gaussian noise multiplier:", NOISE_MULTIPLIER)
print("Max nodes/client:", MAX_NODES_PER_CLIENT)
print("Device:", DEVICE if "DEVICE" in globals() else "cuda/cpu")
```

```python
# ============================================================
# FL-2 — BUILD CLIENT PARTITIONS WITHOUT SHARING RAW DATA
# ============================================================

def build_federated_clients(
    graph,
    num_clients=50,
    max_nodes_per_client=None,
    seed=42
):
    """Create disjoint training-node partitions.

    Only local node indices are assigned to clients. The original
    graph/data remains in the notebook and is never exported.
    """
    rng = np.random.default_rng(seed)

    train_nodes = torch.where(
        graph.train_mask.detach().cpu()
    )[0].numpy()

    rng.shuffle(train_nodes)

    if max_nodes_per_client is not None:
        total_capacity = num_clients * max_nodes_per_client
        train_nodes = train_nodes[:min(len(train_nodes), total_capacity)]

    partitions = np.array_split(train_nodes, num_clients)

    clients = []
    for cid, nodes_np in enumerate(partitions):
        nodes_np = np.asarray(nodes_np, dtype=np.int64)
        if len(nodes_np) == 0:
            continue

        nodes = torch.tensor(nodes_np, dtype=torch.long)

        # Induced local subgraph: cross-client nodes are not used in
        # local message passing.
        local_edge_index, _ = subgraph(
            nodes,
            graph.edge_index.detach().cpu(),
            relabel_nodes=True,
            num_nodes=graph.num_nodes
        )

        clients.append({
            "client_id": cid,
            "global_nodes": nodes,
            "edge_index": local_edge_index,
            "num_samples": len(nodes)
        })

    return clients


def materialize_client(graph, client):
    nodes = client["global_nodes"].to(graph.x.device)
    edge_index = client["edge_index"].to(graph.x.device)

    x_local = graph.x[nodes]
    y_local = graph.y[nodes]

    return x_local, y_local, edge_index


clients = build_federated_clients(
    graph_data,
    num_clients=NUM_CLIENTS,
    max_nodes_per_client=MAX_NODES_PER_CLIENT,
    seed=SEED
)

print("=" * 100)
print("FEDERATED CLIENT PARTITIONS")
print("=" * 100)

client_table = pd.DataFrame([
    {
        "Client": c["client_id"],
        "Training Samples": c["num_samples"],
        "Local Edges": int(c["edge_index"].shape[1])
    }
    for c in clients
])

display(client_table)

print("Total federated training samples:",
      int(client_table["Training Samples"].sum()))
```

```python
# ============================================================
# FL-3 — PRIVACY-PRESERVING UPDATE FUNCTIONS
# ============================================================

def clone_state_dict(model):
    return {
        k: v.detach().cpu().clone()
        for k, v in model.state_dict().items()
    }


def state_dict_l2_distance(a, b):
    total = 0.0
    for k in a:
        da = a[k].float()
        db = b[k].float()
        total += torch.sum((da - db) ** 2).item()
    return math.sqrt(total)


def clip_and_noise_update(
    local_state,
    global_state,
    clip_norm=1.0,
    noise_multiplier=0.0
):
    """Clip a client model update and optionally perturb it.

    This is client-update perturbation, not a formal example-level DP
    guarantee. The noisy update is what is sent to the server.
    """
    delta = {}
    sq_norm = 0.0

    for k in global_state:
        d = local_state[k].float() - global_state[k].float()
        delta[k] = d
        sq_norm += torch.sum(d * d).item()

    update_norm = math.sqrt(sq_norm)
    scale = min(1.0, clip_norm / (update_norm + 1e-12))

    noisy_state = {}
    for k in global_state:
        clipped = delta[k] * scale

        if noise_multiplier > 0:
            noise = torch.randn_like(clipped) * (
                noise_multiplier * clip_norm
            )
            clipped = clipped + noise

        noisy_state[k] = (
            global_state[k].float() + clipped
        ).to(global_state[k].dtype)

    return noisy_state, update_norm, scale


def aggregate_client_states(
    client_states,
    client_weights,
    global_state
):
    total_weight = float(sum(client_weights))
    new_state = {}

    for k in global_state:
        accumulator = torch.zeros_like(
            global_state[k],
            dtype=torch.float32
        )

        for state, weight in zip(
            client_states,
            client_weights
        ):
            accumulator += (
                state[k].float() *
                (weight / total_weight)
            )

        new_state[k] = accumulator.to(
            global_state[k].dtype
        )

    return new_state


print("Privacy-preserving update functions are ready.")
```

```python
# ============================================================
# FL-4 — LOCAL CLIENT TRAINING + FEDAVG/FEDPROX
# ============================================================

def make_federated_model():
    model = NovelTemporalClassifier(
        input_dim=graph_data.num_features,
        hidden_dim=128,
        embedding_dim=64,
        num_layers=2,
        num_classes=2,
        dropout=0.30
    ).to(DEVICE)
    return model


def train_one_client(
    global_state,
    client,
    local_epochs=1,
    lr=0.001,
    fedprox_mu=0.01,
    privacy_enabled=True,
    clip_norm=1.0,
    noise_multiplier=0.0
):
    local_model = make_federated_model()
    local_model.load_state_dict(global_state)
    local_model.train()

    x_local, y_local, edge_index_local = materialize_client(
        graph_data,
        client
    )

    # Class weights computed locally
    counts = torch.bincount(
        y_local,
        minlength=2
    ).float()

    safe_counts = torch.clamp(counts, min=1.0)
    weights = (
        counts.sum() / (2.0 * safe_counts)
    ).to(DEVICE)

    criterion_local = nn.CrossEntropyLoss(
        weight=weights
    )

    optimizer_local = torch.optim.AdamW(
        local_model.parameters(),
        lr=lr,
        weight_decay=WEIGHT_DECAY
    )

    global_on_device = {
        k: v.to(DEVICE)
        for k, v in global_state.items()
    }

    start = time.perf_counter()

    for _ in range(local_epochs):
        optimizer_local.zero_grad(set_to_none=True)

        logits = local_model(
            x_local,
            edge_index_local
        )

        loss = criterion_local(
            logits,
            y_local
        )

        # FedProx proximal term
        if fedprox_mu > 0:
            prox = torch.zeros(
                (),
                device=DEVICE
            )
            for name, param in local_model.named_parameters():
                prox += torch.sum(
                    (param - global_on_device[name]) ** 2
                )
            loss = loss + 0.5 * fedprox_mu * prox

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            local_model.parameters(),
            max_norm=5.0
        )

        optimizer_local.step()

    elapsed = time.perf_counter() - start

    local_state = clone_state_dict(local_model)

    if privacy_enabled:
        transmitted_state, raw_norm, clip_scale = clip_and_noise_update(
            local_state,
            global_state,
            clip_norm=clip_norm,
            noise_multiplier=noise_multiplier
        )
    else:
        transmitted_state = local_state
        raw_norm = state_dict_l2_distance(
            local_state,
            global_state
        )
        clip_scale = 1.0

    with torch.no_grad():
        local_model.eval()
        local_logits = local_model(
            x_local,
            edge_index_local
        )
        local_pred = torch.argmax(
            local_logits,
            dim=1
        )

    y_np = y_local.detach().cpu().numpy()
    p_np = local_pred.detach().cpu().numpy()

    return {
        "state": transmitted_state,
        "samples": client["num_samples"],
        "loss": float(loss.detach().cpu()),
        "accuracy": accuracy_score(y_np, p_np),
        "f1": f1_score(y_np, p_np, zero_division=0),
        "elapsed": elapsed,
        "raw_update_norm": raw_norm,
        "clip_scale": clip_scale
    }


def evaluate_global_model(
    model,
    graph,
    mask
):
    model.eval()

    with torch.no_grad():
        logits = model(
            graph.x,
            graph.edge_index
        )
        selected_logits = logits[mask]
        labels = graph.y[mask]
        probabilities = torch.softmax(
            selected_logits,
            dim=1
        )[:, 1]
        predictions = torch.argmax(
            selected_logits,
            dim=1
        )

    y_true_local = labels.cpu().numpy()
    y_pred_local = predictions.cpu().numpy()
    y_prob_local = probabilities.cpu().numpy()

    return {
        "loss": float(
            nn.CrossEntropyLoss()(
                selected_logits,
                labels
            ).cpu()
        ),
        "accuracy": accuracy_score(
            y_true_local,
            y_pred_local
        ),
        "precision": precision_score(
            y_true_local,
            y_pred_local,
            zero_division=0
        ),
        "recall": recall_score(
            y_true_local,
            y_pred_local,
            zero_division=0
        ),
        "f1": f1_score(
            y_true_local,
            y_pred_local,
            zero_division=0
        ),
        "y_true": y_true_local,
        "y_pred": y_pred_local,
        "y_prob": y_prob_local
    }


print("Client training and evaluation functions are ready.")
```

```python
# ============================================================
# FL-5 — FEDERATED TRAINING
# ============================================================

federated_model = make_federated_model()
global_state = clone_state_dict(federated_model)

FL_HISTORY = []
FL_CLIENT_LOG = []

num_total_clients = len(clients)

for round_id in range(1, FL_ROUNDS + 1):

    round_start = time.perf_counter()

    sample_size = max(
        1,
        math.ceil(
            CLIENT_FRACTION * num_total_clients
        )
    )

    rng = np.random.default_rng(
        SEED + round_id
    )

    selected_ids = rng.choice(
        num_total_clients,
        size=min(sample_size, num_total_clients),
        replace=False
    )

    selected_clients = [
        clients[int(i)]
        for i in selected_ids
    ]

    local_states = []
    local_weights = []
    local_losses = []
    local_times = []
    update_norms = []
    clip_scales = []

    for client in selected_clients:

        result = train_one_client(
            global_state=global_state,
            client=client,
            local_epochs=LOCAL_EPOCHS,
            lr=LOCAL_LR,
            fedprox_mu=FEDPROX_MU,
            privacy_enabled=PRIVACY_ENABLED,
            clip_norm=UPDATE_CLIP_NORM,
            noise_multiplier=NOISE_MULTIPLIER
        )

        local_states.append(result["state"])
        local_weights.append(result["samples"])
        local_losses.append(result["loss"])
        local_times.append(result["elapsed"])
        update_norms.append(result["raw_update_norm"])
        clip_scales.append(result["clip_scale"])

        FL_CLIENT_LOG.append({
            "Round": round_id,
            "Client": client["client_id"],
            "Samples": client["num_samples"],
            "Local Loss": result["loss"],
            "Local Accuracy": result["accuracy"],
            "Local F1": result["f1"],
            "Update Norm": result["raw_update_norm"],
            "Clip Scale": result["clip_scale"],
            "Local Time (s)": result["elapsed"]
        })

    global_state = aggregate_client_states(
        local_states,
        local_weights,
        global_state
    )

    federated_model.load_state_dict(
        global_state
    )

    validation = evaluate_global_model(
        federated_model,
        graph_data,
        graph_data.val_mask
    )

    round_time = time.perf_counter() - round_start

    # Approximate transmitted payload:
    # one model state per selected client, in bytes.
    model_bytes = sum(
        tensor.numel() * tensor.element_size()
        for tensor in global_state.values()
    )

    upload_bytes = model_bytes * len(selected_clients)
    download_bytes = model_bytes * len(selected_clients)

    FL_HISTORY.append({
        "Round": round_id,
        "Selected Clients": len(selected_clients),
        "Validation Loss": validation["loss"],
        "Validation Accuracy": validation["accuracy"],
        "Validation Precision": validation["precision"],
        "Validation Recall": validation["recall"],
        "Validation F1": validation["f1"],
        "Mean Local Loss": float(np.mean(local_losses)),
        "Mean Update Norm": float(np.mean(update_norms)),
        "Mean Clip Scale": float(np.mean(clip_scales)),
        "Round Time (s)": round_time,
        "Model Size (MB)": model_bytes / (1024**2),
        "Upload (MB)": upload_bytes / (1024**2),
        "Download (MB)": download_bytes / (1024**2),
        "Total Communication (MB)": (
            upload_bytes + download_bytes
        ) / (1024**2)
    })

    print(
        f"Round {round_id:02d}/{FL_ROUNDS} | "
        f"Clients={len(selected_clients)} | "
        f"Val Acc={validation['accuracy']*100:.2f}% | "
        f"Val F1={validation['f1']*100:.2f}% | "
        f"Time={round_time:.2f}s"
    )

FL_HISTORY_DF = pd.DataFrame(FL_HISTORY)
FL_CLIENT_LOG_DF = pd.DataFrame(FL_CLIENT_LOG)

FL_HISTORY_DF.to_csv(
    os.path.join(FL_RESULT_DIR, "Federated_Round_History.csv"),
    index=False
)

FL_CLIENT_LOG_DF.to_csv(
    os.path.join(FL_RESULT_DIR, "Federated_Client_Log.csv"),
    index=False
)

torch.save(
    {
        "model_state_dict": federated_model.state_dict(),
        "config": {
            "num_clients": NUM_CLIENTS,
            "rounds": FL_ROUNDS,
            "local_epochs": LOCAL_EPOCHS,
            "fedprox_mu": FEDPROX_MU,
            "privacy_enabled": PRIVACY_ENABLED,
            "clip_norm": UPDATE_CLIP_NORM,
            "noise_multiplier": NOISE_MULTIPLIER
        }
    },
    os.path.join(
        FL_RESULT_DIR,
        "Federated_Privacy_Preserving_Model.pt"
    )
)

print("\nFederated training completed.")
```

```python
# ============================================================
# FL-6 — FINAL FEDERATED TEST EVALUATION
# ============================================================

fl_test = evaluate_global_model(
    federated_model,
    graph_data,
    graph_data.test_mask
)

fl_cm = confusion_matrix(
    fl_test["y_true"],
    fl_test["y_pred"]
)

fl_roc_auc = roc_auc_score(
    fl_test["y_true"],
    fl_test["y_prob"]
)

fl_pr_auc = average_precision_score(
    fl_test["y_true"],
    fl_test["y_prob"]
)

tn, fp, fn, tp = fl_cm.ravel()

fl_specificity = (
    tn / (tn + fp)
    if (tn + fp) > 0 else 0
)

fl_far = (
    fp / (fp + tn)
    if (fp + tn) > 0 else 0
)

fl_frr = (
    fn / (fn + tp)
    if (fn + tp) > 0 else 0
)

FL_FINAL_RESULTS = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC",
        "PR-AUC",
        "Specificity",
        "False Acceptance Rate",
        "False Rejection Rate",
        "Balanced Accuracy",
        "MCC"
    ],
    "Value": [
        fl_test["accuracy"],
        fl_test["precision"],
        fl_test["recall"],
        fl_test["f1"],
        fl_roc_auc,
        fl_pr_auc,
        fl_specificity,
        fl_far,
        fl_frr,
        (fl_test["recall"] + fl_specificity) / 2,
        matthews_corrcoef(fl_test["y_true"], fl_test["y_pred"])
    ]
})

FL_FINAL_RESULTS["Percentage"] = (
    FL_FINAL_RESULTS["Value"] * 100
)

FL_FINAL_RESULTS.to_csv(
    os.path.join(
        FL_RESULT_DIR,
        "Federated_Final_Test_Results.csv"
    ),
    index=False
)

pd.DataFrame(
    fl_cm,
    index=["Actual Benign", "Actual Attack"],
    columns=["Predicted Benign", "Predicted Attack"]
).to_csv(
    os.path.join(
        FL_RESULT_DIR,
        "Federated_Confusion_Matrix.csv"
    )
)

print("=" * 100)
print("FINAL FEDERATED TEST RESULTS")
print("=" * 100)
display(
    FL_FINAL_RESULTS.style.format({
        "Value": "{:.6f}",
        "Percentage": "{:.2f}"
    })
)
print("\nConfusion Matrix:")
print(fl_cm)
```

```python
# ============================================================
# FL-8 — SCALABILITY EXPERIMENT
# Varies client count and records model quality, runtime and
# communication. Set SCALABILITY_CLIENT_COUNTS to expand.
# ============================================================

def run_scalability_experiment(
    client_counts,
    rounds=20,
    local_epochs=1,
    max_nodes_per_client=10000
):
    rows = []

    for n_clients in client_counts:

        print(
            f"\n--- Scalability run: {n_clients} clients ---"
        )

        exp_clients = build_federated_clients(
            graph_data,
            num_clients=n_clients,
            max_nodes_per_client=max_nodes_per_client,
            seed=SEED
        )

        exp_model = make_federated_model()
        exp_state = clone_state_dict(exp_model)

        start_all = time.perf_counter()

        for round_id in range(1, rounds + 1):

            round_start = time.perf_counter()

            states = []
            weights = []

            for client in exp_clients:
                result = train_one_client(
                    global_state=exp_state,
                    client=client,
                    local_epochs=local_epochs,
                    lr=LOCAL_LR,
                    fedprox_mu=FEDPROX_MU,
                    privacy_enabled=PRIVACY_ENABLED,
                    clip_norm=UPDATE_CLIP_NORM,
                    noise_multiplier=NOISE_MULTIPLIER
                )
                states.append(result["state"])
                weights.append(result["samples"])

            exp_state = aggregate_client_states(
                states,
                weights,
                exp_state
            )
            exp_model.load_state_dict(exp_state)

            val = evaluate_global_model(
                exp_model,
                graph_data,
                graph_data.val_mask
            )

            model_bytes = sum(
                t.numel() * t.element_size()
                for t in exp_state.values()
            )

            round_time = time.perf_counter() - round_start

            rows.append({
                "Clients": n_clients,
                "Round": round_id,
                "Validation Accuracy": val["accuracy"],
                "Validation F1": val["f1"],
                "Round Time (s)": round_time,
                "Communication (MB)": (
                    2 * len(exp_clients) * model_bytes
                ) / (1024**2),
                "Training Samples Used": sum(
                    weights
                )
            })

        total_time = time.perf_counter() - start_all

        rows[-1]["Total Experiment Time (s)"] = total_time

    result_df = pd.DataFrame(rows)
    result_df.to_csv(
        os.path.join(
            FL_RESULT_DIR,
            "Scalability_Experiment.csv"
        ),
        index=False
    )
    return result_df


SCALABILITY_RESULTS = run_scalability_experiment(
    SCALABILITY_CLIENT_COUNTS,
    rounds=3,
    local_epochs=1,
    max_nodes_per_client=10000
)

display(SCALABILITY_RESULTS)

# ------------------------------------------------------------
# Scalability graph 1 — clients vs communication
# ------------------------------------------------------------

agg_scalability = (
    SCALABILITY_RESULTS
    .groupby("Clients", as_index=False)
    .agg({
        "Validation Accuracy": "mean",
        "Validation F1": "mean",
        "Round Time (s)": "mean",
        "Communication (MB)": "mean"
    })
)

plt.figure(figsize=(9, 6))
plt.plot(
    agg_scalability["Clients"],
    agg_scalability["Communication (MB)"],
    marker="o",
    linewidth=2,
    label="Communication / Round"
)
plt.xlabel("Number of Clients")
plt.ylabel("Communication (MB)")
plt.title("Scalability — Client Count vs Communication")
plt.grid(True, alpha=0.3)
plt.legend()
save_plot("FL_15_Scalability_Communication.png")

# ------------------------------------------------------------
# Scalability graph 2 — clients vs runtime
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))
plt.plot(
    agg_scalability["Clients"],
    agg_scalability["Round Time (s)"],
    marker="o",
    linewidth=2,
    label="Mean Round Time"
)
plt.xlabel("Number of Clients")
plt.ylabel("Mean Round Time (seconds)")
plt.title("Scalability — Client Count vs Runtime")
plt.grid(True, alpha=0.3)
plt.legend()
save_plot("FL_16_Scalability_Runtime.png")

# ------------------------------------------------------------
# Scalability graph 3 — clients vs validation accuracy
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))
plt.plot(
    agg_scalability["Clients"],
    agg_scalability["Validation Accuracy"] * 100,
    marker="o",
    linewidth=2,
    label="Validation Accuracy"
)
plt.xlabel("Number of Clients")
plt.ylabel("Mean Validation Accuracy (%)")
plt.title("Scalability — Client Count vs Validation Accuracy")
plt.grid(True, alpha=0.3)
plt.ylim(0, 105)
plt.legend()
save_plot("FL_17_Scalability_Accuracy.png")

# ------------------------------------------------------------
# Scalability graph 4 — clients vs validation F1
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))
plt.plot(
    agg_scalability["Clients"],
    agg_scalability["Validation F1"] * 100,
    marker="o",
    linewidth=2,
    label="Validation F1"
)
plt.xlabel("Number of Clients")
plt.ylabel("Mean Validation F1 (%)")
plt.title("Scalability — Client Count vs Validation F1")
plt.grid(True, alpha=0.3)
plt.ylim(0, 105)
plt.legend()
save_plot("FL_18_Scalability_F1.png")

print("\nScalability experiment completed.")
```

```python
# ============================================================
# FL-9 — PRIVACY SENSITIVITY EXPERIMENT
# Compares several noise levels using a short federated run.
# ============================================================

def run_privacy_sensitivity(
    noise_levels=(0.0, 0.005, 0.01, 0.02),
    rounds=3
):
    rows = []

    base_clients = build_federated_clients(
        graph_data,
        num_clients=NUM_CLIENTS,
        max_nodes_per_client=10000,
        seed=SEED
    )

    for noise in noise_levels:

        exp_model = make_federated_model()
        exp_state = clone_state_dict(exp_model)

        for round_id in range(1, rounds + 1):

            states = []
            weights = []

            for client in base_clients:
                result = train_one_client(
                    global_state=exp_state,
                    client=client,
                    local_epochs=1,
                    lr=LOCAL_LR,
                    fedprox_mu=FEDPROX_MU,
                    privacy_enabled=(noise > 0),
                    clip_norm=UPDATE_CLIP_NORM,
                    noise_multiplier=noise
                )
                states.append(result["state"])
                weights.append(result["samples"])

            exp_state = aggregate_client_states(
                states,
                weights,
                exp_state
            )
            exp_model.load_state_dict(exp_state)

        test_eval = evaluate_global_model(
            exp_model,
            graph_data,
            graph_data.test_mask
        )

        rows.append({
            "Noise Multiplier": noise,
            "Test Accuracy": test_eval["accuracy"],
            "Test Precision": test_eval["precision"],
            "Test Recall": test_eval["recall"],
            "Test F1": test_eval["f1"]
        })

    df = pd.DataFrame(rows)
    df.to_csv(
        os.path.join(
            FL_RESULT_DIR,
            "Privacy_Sensitivity_Experiment.csv"
        ),
        index=False
    )
    return df


PRIVACY_RESULTS = run_privacy_sensitivity()
display(PRIVACY_RESULTS)

plt.figure(figsize=(10, 6))
for col, label, marker in [
    ("Test Accuracy", "Accuracy", "o"),
    ("Test Precision", "Precision", "s"),
    ("Test Recall", "Recall", "^"),
    ("Test F1", "F1", "D")
]:
    plt.plot(
        PRIVACY_RESULTS["Noise Multiplier"],
        PRIVACY_RESULTS[col] * 100,
        marker=marker,
        linewidth=2,
        label=label
    )

plt.xlabel("Gaussian Noise Multiplier")
plt.ylabel("Test Score (%)")
plt.title("Privacy Sensitivity — Noise vs Model Performance")
plt.grid(True, alpha=0.3)
plt.ylim(0, 105)
plt.legend()
save_plot("FL_19_Privacy_Sensitivity.png")
```

```python

EXPERIMENT_CONFIG = {
    "dataset": "CIC-IoT-2023",
    "dataset_path": "/content/drive/MyDrive/IDS dataset",

    "final_num_clients": 100,
    "final_algorithm": "FedProx",
    "final_partition": "mild_non_iid",

    "rounds": 20,
    "local_epochs": 30,

    "local_lr": LOCAL_LR,
    "fedprox_mu": FEDPROX_MU,
    "update_clip_norm": UPDATE_CLIP_NORM,

    "primary_noise_multiplier": 0.0,

    "seeds": [42, 43, 44, 45, 46],

    "scalability_client_counts": [100],
    "optional_large_client_counts": [],

    "privacy_noise_multipliers": [0.0],
    "mild_noniid_dirichlet_alpha": 0.5,
    "strong_noniid_dirichlet_alpha": 0.1,

    "correlation_threshold": 0.90,

    "run_optional_large_scalability": False,
    "run_ablation": False,
    "run_repeated_seeds": True,
    "run_imputer_reconstruction": False,

    # Disable unrelated additional experiments
    "run_external_bot_iot": False,
    "run_gnnexplainer": False,
    "run_formal_dp": False,
    "run_tgat_comparison": False,

    "dp_delta": 1e-5,
    "gnnexplainer_cases_per_bucket": 5,
    "gnnexplainer_epochs": 100,
}
SEED = 42
LOCAL_LR = 0.001
FEDPROX_MU = 0.01
if EXPERIMENT_CONFIG["run_repeated_seeds"]:


# ============================================================
# 100-CLIENT FEDPROX: FIVE INDEPENDENT RANDOM SEEDS
# 20 GLOBAL ROUNDS | 30 LOCAL EPOCHS
# ============================================================

if EXPERIMENT_CONFIG["run_repeated_seeds"]:

    repeated_rows = []
    all_seed_histories = []

    for seed in EXPERIMENT_CONFIG["seeds"]:

        print("\n" + "=" * 70)
        print(f"RUNNING FEDPROX: 100 CLIENTS | SEED {seed}")
        print("=" * 70)

        run = run_federated_experiment(
            num_clients=100,
            algorithm="FedProx",
            partition_profile="mild_non_iid",
            rounds=20,
            local_epochs=30,
            seed=seed,
            noise_multiplier=0.0,
            dirichlet_alpha=0.5,
            record_client_log=True
        )

        # Save final test metrics for this seed
        final_metrics = run["final"].copy()
        final_metrics["Seed"] = seed
        final_metrics["Clients"] = 100

        repeated_rows.append(final_metrics)

        # Save all 20 rounds for this seed
        history = run["history"].copy()
        history["Seed"] = seed
        history["Clients"] = 100

        all_seed_histories.append(history)

        # Save individual seed files
        save_csv(
            history,
            f"FedProx_100Clients_Seed_{seed}_Round_History.csv"
        )

        save_csv(
            run["client_log"],
            f"FedProx_100Clients_Seed_{seed}_Client_Log.csv"
        )

        print("\nFINAL TEST METRICS")
        print(pd.DataFrame([final_metrics]).to_string(index=False))

        del run

    # --------------------------------------------------------
    # COMBINE ALL FIVE SEEDS
    # --------------------------------------------------------

    REPEATED_RESULTS = pd.DataFrame(repeated_rows)

    ALL_SEED_HISTORY = pd.concat(
        all_seed_histories,
        ignore_index=True
    )

    save_csv(
        REPEATED_RESULTS,
        "FedProx_100Clients_Five_Seeds_Final_Results.csv"
    )

    save_csv(
        ALL_SEED_HISTORY,
        "FedProx_100Clients_Five_Seeds_Round_History.csv"
    )

    # --------------------------------------------------------
    # MEAN AND STANDARD DEVIATION
    # --------------------------------------------------------

    metric_columns = [
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "ROC-AUC",
        "PR-AUC",
        "Specificity",
        "FAR_FPR",
        "FRR_FNR",
        "Balanced_Accuracy",
        "MCC"
    ]

    SUMMARY = REPEATED_RESULTS[
        metric_columns
    ].agg(["mean", "std"]).T.reset_index()

    SUMMARY.columns = [
        "Metric",
        "Mean",
        "Standard_Deviation"
    ]

    save_csv(
        SUMMARY,
        "FedProx_100Clients_Five_Seeds_Mean_STD.csv"
    )

    print("\nFIVE-SEED FINAL RESULTS")
    display(REPEATED_RESULTS)

    print("\nMEAN AND STANDARD DEVIATION")
    display(SUMMARY)

else:
    print("Repeated-seed experiment disabled.")


# ============================================================
# PUBLICATION GRAPHS: 100 CLIENTS / FIVE SEEDS
# ============================================================

SEED_GRAPH_DIR = os.path.join(
    FL_GRAPH_DIR,
    "100_Clients_Five_Seeds"
)

os.makedirs(SEED_GRAPH_DIR, exist_ok=True)

# Metric columns in the actual training history
round_metrics = {
    "Validation Accuracy": "Validation_Accuracy",
    "Validation Precision": "Validation_Precision",
    "Validation Recall": "Validation_Recall",
    "Validation F1": "Validation_F1",
    "Validation Loss": "Validation_Loss",
    "Mean Local Loss": "Mean_Local_Loss",
    "Communication": "Cumulative_Communication_MB",
    "Round Runtime": "Round_Runtime_s"
}

for title, column in round_metrics.items():

    plt.figure(figsize=(11, 7))

    for seed in EXPERIMENT_CONFIG["seeds"]:

        subset = ALL_SEED_HISTORY[
            ALL_SEED_HISTORY["Seed"] == seed
        ].sort_values("Round")

        values = subset[column].values

        # Convert classification metrics to percentages
        if column in [
            "Validation_Accuracy",
            "Validation_Precision",
            "Validation_Recall",
            "Validation_F1"
        ]:
            values = values * 100

        plt.plot(
            subset["Round"],
            values,
            marker="o",
            markersize=4,
            linewidth=2,
            label=f"Seed {seed}"
        )

    plt.xlabel("Global Communication Rounds")

    if column in [
        "Validation_Accuracy",
        "Validation_Precision",
        "Validation_Recall",
        "Validation_F1"
    ]:
        plt.ylabel("Performance (%)")
    else:
        plt.ylabel(title)

    plt.title(
        f"FedGraph-IDS: {title} - 100 Clients",
        fontsize=14,
        fontweight="bold"
    )

    plt.xticks(range(1, 21))
    plt.grid(True, linestyle="--", alpha=0.4)
    plt.legend()
    plt.tight_layout()

    filename = title.replace(" ", "_") + ".png"

    plt.savefig(
        os.path.join(SEED_GRAPH_DIR, filename),
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# ============================================================
# FINAL TEST METRIC COMPARISON
# ============================================================

final_metrics_to_plot = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC-AUC",
    "PR-AUC",
    "Balanced_Accuracy"
]

for metric in final_metrics_to_plot:

    plt.figure(figsize=(9, 6))

    plt.bar(
        REPEATED_RESULTS["Seed"].astype(str),
        REPEATED_RESULTS[metric] * 100
    )

    plt.xlabel("Random Seed")
    plt.ylabel("Performance (%)")

    plt.title(
        f"FedGraph-IDS: Final {metric} - 100 Clients"
    )

    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            SEED_GRAPH_DIR,
            f"Final_{metric}.png"
        ),
        dpi=400,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


# MCC uses its original 0-1 scale
plt.figure(figsize=(9, 6))

plt.bar(
    REPEATED_RESULTS["Seed"].astype(str),
    REPEATED_RESULTS["MCC"]
)

plt.xlabel("Random Seed")
plt.ylabel("MCC")
plt.title("FedGraph-IDS: Final MCC - 100 Clients")
plt.ylim(0, 1)

plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()

plt.savefig(
    os.path.join(SEED_GRAPH_DIR, "Final_MCC.png"),
    dpi=400,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("All five-seed graphs generated successfully.")
print("Graph directory:", SEED_GRAPH_DIR)

```

```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# FEDGRAPH-IDS: 200 CLIENT FEDPROX EXPERIMENT
# MANUAL PERFORMANCE VISUALIZATION
# ============================================================

OUTPUT = "FedGraph_IDS_200_Clients_Results"
os.makedirs(OUTPUT, exist_ok=True)

CLIENTS = 200
ROUNDS = 20
LOCAL_EPOCHS = 30
SEEDS = [42, 43, 44, 45, 46]

print("=" * 65)
print("FEDGRAPH-IDS: 200 CLIENT FEDPROX EXPERIMENT")
print("=" * 65)

print("Dataset: CIC-IoT-2023")
print("Clients:", CLIENTS)
print("Algorithm: FedProx")
print("Global rounds:", ROUNDS)
print("Local epochs:", LOCAL_EPOCHS)
print("Random seeds:", SEEDS)
print("Learning rate: 0.001")
print("FedProx mu: 0.01")
print("Dirichlet alpha: 0.5")
print("Noise multiplier: 0.0")

# ============================================================
# 1. MANUAL ROUND-WISE ACCURACY
# Illustrative values: replace with actual experiment results.
# ============================================================

accuracy = {

    42: [95.2,95.7,96.2,96.6,97.0,97.3,97.5,97.7,97.9,98.1,
         98.2,98.3,98.4,98.5,98.6,98.7,98.8,98.9,99.0,99.05],

    43: [95.0,95.5,96.0,96.4,96.8,97.1,97.4,97.6,97.8,98.0,
         98.1,98.2,98.3,98.4,98.5,98.6,98.7,98.8,98.9,99.0],

    44: [95.3,95.8,96.3,96.7,97.0,97.3,97.6,97.8,98.0,98.2,
         98.3,98.4,98.5,98.6,98.7,98.8,98.9,99.0,99.05,99.10],

    45: [95.1,95.6,96.1,96.5,96.9,97.2,97.5,97.7,97.9,98.1,
         98.2,98.3,98.4,98.5,98.6,98.7,98.8,98.9,99.0,99.05],

    46: [95.2,95.7,96.2,96.6,96.9,97.2,97.5,97.7,97.9,98.1,
         98.2,98.4,98.5,98.6,98.7,98.8,98.9,99.0,99.05,99.10]
}

# ============================================================
# 2. OTHER EVALUATION METRICS
# These are illustrative derived trajectories.
# Enter independently measured metrics for real experiments.
# ============================================================

precision = {
    s: [min(99.95, x + 0.30) for x in accuracy[s]]
    for s in SEEDS
}

recall = {
    s: [max(0, x - 0.90) for x in accuracy[s]]
    for s in SEEDS
}

f1_score = {
    s: [
        2*p*r/(p+r)
        for p, r in zip(precision[s], recall[s])
    ]
    for s in SEEDS
}

balanced_accuracy = {
    s: [max(0, x - 0.55) for x in accuracy[s]]
    for s in SEEDS
}

mcc = {
    s: [
        min(0.999, max(0, (x - 50)/50))
        for x in accuracy[s]
    ]
    for s in SEEDS
}

training_loss = {
    s: np.linspace(
        0.22 + i*0.008,
        0.010 + i*0.002,
        ROUNDS
    ).tolist()
    for i, s in enumerate(SEEDS)
}

# ============================================================
# 3. COMBINE METRICS
# ============================================================

all_metrics = {
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-score": f1_score,
    "Balanced Accuracy": balanced_accuracy,
    "MCC": mcc,
    "Training Loss": training_loss
}

# ============================================================
# 4. ROUND-WISE GRAPH GENERATOR
# ============================================================

def draw_round_graph(metric_name, metric_data):

    plt.figure(figsize=(12, 7))

    for seed in SEEDS:

        plt.plot(
            range(1, ROUNDS + 1),
            metric_data[seed],
            marker="o",
            markersize=4,
            linewidth=2,
            label=f"Seed {seed}"
        )

    plt.xlabel("Global Communication Rounds", fontsize=12)

    if metric_name == "MCC":
        plt.ylabel("Matthews Correlation Coefficient")
    elif metric_name == "Training Loss":
        plt.ylabel("Training Loss")
    else:
        plt.ylabel("Performance (%)")

    plt.title(
        f"FedGraph-IDS: {metric_name} - 200 Clients",
        fontsize=15,
        fontweight="bold"
    )

    plt.xticks(range(1, ROUNDS + 1))
    plt.grid(True, linestyle="--", alpha=0.4)
    plt.legend(title="Random Seeds")

    plt.tight_layout()

    filename = f"{metric_name.replace(' ', '_')}_200_Clients.png"

    plt.savefig(
        os.path.join(OUTPUT, filename),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print("Generated:", filename)


# ============================================================
# 5. GENERATE ALL ROUND-WISE GRAPHS
# ============================================================

for metric_name, metric_data in all_metrics.items():
    draw_round_graph(metric_name, metric_data)


# ============================================================
# 6. FINAL METRICS FOR EACH SEED
# ============================================================

final_results = []

for seed in SEEDS:

    final_results.append({

        "Seed": seed,
        "Accuracy": accuracy[seed][-1],
        "Precision": precision[seed][-1],
        "Recall": recall[seed][-1],
        "F1-score": f1_score[seed][-1],
        "Balanced Accuracy": balanced_accuracy[seed][-1],
        "MCC": mcc[seed][-1]

    })

df = pd.DataFrame(final_results)

print("\nFINAL EVALUATION METRICS")
print(df.round(4).to_string(index=False))

df.to_csv(
    os.path.join(OUTPUT, "Final_Seed_Metrics.csv"),
    index=False
)


# ============================================================
# 7. MEAN AND STANDARD DEVIATION
# ============================================================

summary = df.drop(columns=["Seed"]).agg(["mean", "std"]).T

print("\nMEAN AND STANDARD DEVIATION")
print(summary.round(4))

summary.to_csv(
    os.path.join(OUTPUT, "Mean_Standard_Deviation.csv")
)


# ============================================================
# 8. FINAL PERFORMANCE COMPARISON
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-score",
    "Balanced Accuracy"
]

x = np.arange(len(SEEDS))
width = 0.15

plt.figure(figsize=(14, 7))

for i, metric in enumerate(metrics):

    plt.bar(
        x + i * width,
        df[metric],
        width,
        label=metric
    )

plt.xlabel("Random Seeds")
plt.ylabel("Performance (%)")

plt.title(
    "FedGraph-IDS: Final Evaluation Metrics - 200 Clients",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(x + width * 2, SEEDS)
plt.ylim(0, 110)

plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT, "Final_Metrics_Comparison.png"),
    dpi=300
)

plt.show()
plt.close()


# ============================================================
# 9. MCC COMPARISON
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    df["Seed"].astype(str),
    df["MCC"]
)

plt.xlabel("Random Seeds")
plt.ylabel("Matthews Correlation Coefficient")

plt.title(
    "FedGraph-IDS: MCC Comparison - 200 Clients",
    fontsize=14
)

plt.ylim(0, 1.1)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT, "MCC_Comparison.png"),
    dpi=300
)

plt.show()
plt.close()


# ============================================================
# 10. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("200 CLIENT GRAPH GENERATION COMPLETED")
print("=" * 65)

print("Clients:", CLIENTS)
print("Global rounds:", ROUNDS)
print("Local epochs:", LOCAL_EPOCHS)
print("Random seeds:", SEEDS)

print("\nFiles saved at:")
print(os.path.abspath(OUTPUT))

print("\nNOTE:")
print("All manually entered trajectories are illustrative.")
print("They must not be presented as verified experimental results.")
```

```python



import os, time, math, json, random, copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    balanced_accuracy_score, matthews_corrcoef, roc_curve,
    precision_recall_curve
)
from sklearn.neighbors import NearestNeighbors
from torch_geometric.data import Data
from torch_geometric.nn import GATConv


BACKUP_DIR = os.path.join(FL_RESULT_DIR, "Backups")
os.makedirs(BACKUP_DIR, exist_ok=True)


EXPERIMENT_CONFIG = {
    "dataset": "CIC-IoT-2023 (training); BoT-IoT (independent testing only)",
    "dataset_path": "/content/drive/MyDrive/IDS dataset",
    "final_num_clients": 400,
    "final_algorithm": "FedProx",
    "final_partition": "mild_non_iid",
    "rounds": 20,
    "local_epochs": 30,
    "local_lr": LOCAL_LR,
    "fedprox_mu": FEDPROX_MU,
    "update_clip_norm": UPDATE_CLIP_NORM,
    "primary_noise_multiplier": 0.0,
    "seeds": [42, 43, 44, 45, 46],
    "scalability_client_counts": [100, 200, 400],
    "optional_large_client_counts": [200, 400],
    "privacy_noise_multipliers": [0.0, 0.005, 0.01, 0.02],
    "mild_noniid_dirichlet_alpha": 0.5,
    "strong_noniid_dirichlet_alpha": 0.1,
    "correlation_threshold": 0.90,
    "run_optional_large_scalability": False,
    "run_ablation": False,
    "run_repeated_seeds": True,
    "run_imputer_reconstruction": False,
    "run_external_bot_iot": True,
    "run_gnnexplainer": True,
    "run_formal_dp": True,
    "run_tgat_comparison": True,
    "dp_delta": 1e-5,
    "gnnexplainer_cases_per_bucket": 5,
    "gnnexplainer_epochs": 100,
}


print(json.dumps(EXPERIMENT_CONFIG, indent=2))


def save_csv(df, filename):
    path = os.path.join(FL_RESULT_DIR, filename)
    df.to_csv(path, index=False)
    print(f"Saved: {path}")
    return path

with open(os.path.join(FL_RESULT_DIR, "Experiment_Config.json"), "w") as f:
    json.dump(EXPERIMENT_CONFIG, f, indent=2)


def compute_binary_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()

    specificity = tn / (tn + fp) if (tn + fp) else 0.0
    far_fpr = fp / (fp + tn) if (fp + tn) else 0.0
    frr_fnr = fn / (fn + tp) if (fn + tp) else 0.0


    try:
        roc_auc = roc_auc_score(y_true, y_prob)
    except ValueError:
        roc_auc = np.nan
    try:
        pr_auc = average_precision_score(y_true, y_prob)
    except ValueError:
        pr_auc = np.nan

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Specificity": specificity,
        "FAR_FPR": far_fpr,
        "FRR_FNR": frr_fnr,
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "MCC": matthews_corrcoef(y_true, y_pred),
        "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
    }

def evaluate_model_metrics(model, graph, mask):
    model.eval()
    with torch.no_grad():
        out = model(graph.x, graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        logits = logits[mask]
        labels = graph.y[mask]
        probs = torch.softmax(logits, dim=1)[:, 1]
        preds = torch.argmax(logits, dim=1)

    y_true = labels.detach().cpu().numpy()
    y_pred = preds.detach().cpu().numpy()
    y_prob = probs.detach().cpu().numpy()

    metrics = compute_binary_metrics(y_true, y_pred, y_prob)
    metrics["Validation Loss"] = float(nn.CrossEntropyLoss()(logits, labels).detach().cpu())
    metrics["y_true"] = y_true
    metrics["y_pred"] = y_pred
    metrics["y_prob"] = y_prob
    return metrics


ARCHITECTURE_VERIFICATION = pd.DataFrame([
    ["GraphSAGE layer 1", "NovelTemporalClassifier.gnn1", True, "Active in final classifier"],
    ["GraphSAGE layer 2", "NovelTemporalClassifier.gnn2", True, "Active in final classifier"],
    ["TGAT", "No TGAT class/call found", False, "Do not claim TGAT in final architecture"],
    ["Feature attention", "NovelTemporalClassifier.attention", True, "Active before BiLSTM"],
    ["BiLSTM in classifier", "NovelTemporalClassifier.bilstm", True, "Executes with sequence length 1"],
    ["Temporal attention", "NovelTemporalClassifier.temporal_attention", True, "Executes over length-1 sequence"],
    ["BiLSTM imputer", "BiLSTMImputer", False, "Defined, but no completed training/evaluation found"],
], columns=["Component", "Code Path", "Active", "Verification"])
save_csv(ARCHITECTURE_VERIFICATION, "Architecture_Verification.csv")
display(ARCHITECTURE_VERIFICATION)


def partition_train_nodes(graph, num_clients, profile="iid", seed=42,
                          dirichlet_alpha=0.5, min_samples=1):
    """
    IID:
        class-balanced random allocation, so every client receives
        approximately the same class proportions.

    Mild/strong non-IID:
        class-conditional Dirichlet allocation. Lower alpha => stronger
        heterogeneity. The split is deterministic for a fixed seed.
    """
    rng = np.random.default_rng(seed)
    train_nodes = torch.where(graph.train_mask.detach().cpu())[0].numpy()
    y = graph.y.detach().cpu().numpy()

    if profile == "iid":

        client_lists = [[] for _ in range(num_clients)]
        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            for j, node in enumerate(cls_nodes):
                client_lists[j % num_clients].append(int(node))
    else:
        alpha = dirichlet_alpha
        client_lists = [[] for _ in range(num_clients)]

        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            proportions = rng.dirichlet(np.full(num_clients, alpha))
            counts = rng.multinomial(len(cls_nodes), proportions)

            start = 0
            for cid, count in enumerate(counts):
                client_lists[cid].extend(
                    cls_nodes[start:start+count].tolist()
                )
                start += count


        if any(len(v) < min_samples for v in client_lists):
            return partition_train_nodes(
                graph, num_clients, profile, seed + 1,
                dirichlet_alpha, min_samples
            )

    clients = []
    for cid, nodes_list in enumerate(client_lists):
        nodes_np = np.asarray(nodes_list, dtype=np.int64)
        rng.shuffle(nodes_np)
        nodes = torch.tensor(nodes_np, dtype=torch.long)

        local_edge_index, _ = subgraph(
            nodes,
            graph.edge_index.detach().cpu(),
            relabel_nodes=True,
            num_nodes=graph.num_nodes
        )

        counts = np.bincount(y[nodes_np], minlength=2)

        clients.append({
            "client_id": cid,
            "global_nodes": nodes,
            "edge_index": local_edge_index,
            "num_samples": len(nodes_np),
            "class_0": int(counts[0]),
            "class_1": int(counts[1]),
        })

    return clients

def clients_to_distribution_df(clients, profile, seed):
    rows = []
    for c in clients:
        total = max(c["num_samples"], 1)
        rows.append({
            "Seed": seed,
            "Partition": profile,
            "Client": c["client_id"],
            "Samples": c["num_samples"],
            "Benign_Count": c["class_0"],
            "Attack_Count": c["class_1"],
            "Benign_Fraction": c["class_0"] / total,
            "Attack_Fraction": c["class_1"] / total,
        })
    return pd.DataFrame(rows)

distribution_frames = []
for profile, alpha in [
    ("iid", None),
    ("mild_non_iid", EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]),
    ("strong_non_iid", EXPERIMENT_CONFIG["strong_noniid_dirichlet_alpha"]),
]:
    cset = partition_train_nodes(
        graph_data, 5, profile=profile, seed=SEED,
        dirichlet_alpha=alpha if alpha is not None else 0.5
    )
    distribution_frames.append(
        clients_to_distribution_df(cset, profile, SEED)
    )

CLIENT_DISTRIBUTION_DF = pd.concat(distribution_frames, ignore_index=True)


CURRENT_PARTITION_DIAGNOSIS = pd.DataFrame([{
    "Current_Uploaded_Partition": "random shuffled train nodes + np.array_split",
    "Class_Balance_Enforced": False,
    "Controlled_NonIID_Enforced": False,
    "Revised_Setups": "IID, mild_non_iid, strong_non_iid",
    "Conclusion": "Current setup should not be described as proper non-IID."
}])
save_csv(CURRENT_PARTITION_DIAGNOSIS, "Client_Distribution_Diagnosis.csv")
save_csv(CLIENT_DISTRIBUTION_DF, "Client_Distribution.csv")

print("\nClient-distribution verification:")
display(CLIENT_DISTRIBUTION_DF)

# ------------------------------------------------------------
# RECOMMENDATION 6 — MODEL FACTORY
# ------------------------------------------------------------
def set_all_seeds(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def make_experiment_model():
    return NovelTemporalClassifier(
        input_dim=graph_data.num_features,
        hidden_dim=128,
        embedding_dim=64,
        num_layers=2,
        num_classes=2,
        dropout=0.30
    ).to(DEVICE)

def count_model_parameters(model):
    return sum(p.numel() for p in model.parameters())

# ------------------------------------------------------------
# RECOMMENDATION 7 — FEDERATED TRAINING ENGINE
# ------------------------------------------------------------
def train_one_client_v2(global_state, client, algorithm="FedProx",
                        local_epochs=1, lr=0.001, fedprox_mu=0.01,
                        clip_norm=1.0, noise_multiplier=0.0):
    local_model = make_experiment_model()
    local_model.load_state_dict(global_state)
    local_model.train()

    x_local, y_local, edge_local = materialize_client(
        graph_data, client
    )

    counts = torch.bincount(y_local, minlength=2).float()
    safe = torch.clamp(counts, min=1.0)
    class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=lr,
        weight_decay=WEIGHT_DECAY
    )

    global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
    start = time.perf_counter()

    for _ in range(local_epochs):
        optimizer.zero_grad(set_to_none=True)
        logits = local_model(x_local, edge_local)
        if isinstance(logits, (tuple, list)):
            logits = logits[0]

        loss = criterion(logits, y_local)

        if algorithm.lower() == "fedprox" and fedprox_mu > 0:
            prox = torch.zeros((), device=DEVICE)
            for name, param in local_model.named_parameters():
                prox += torch.sum(
                    (param - global_device[name]) ** 2
                )
            loss = loss + 0.5 * fedprox_mu * prox

        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            local_model.parameters(), max_norm=5.0
        )
        optimizer.step()

    elapsed = time.perf_counter() - start
    local_state = clone_state_dict(local_model)


    if noise_multiplier > 0:
        transmitted, raw_norm, clip_scale = clip_and_noise_update(
            local_state, global_state,
            clip_norm=clip_norm,
            noise_multiplier=noise_multiplier
        )
    else:
        transmitted = local_state
        raw_norm = state_dict_l2_distance(local_state, global_state)
        clip_scale = 1.0

    with torch.no_grad():
        local_model.eval()
        pred_logits = local_model(x_local, edge_local)
        if isinstance(pred_logits, (tuple, list)):
            pred_logits = pred_logits[0]
        pred = torch.argmax(pred_logits, dim=1)

    return {
        "state": transmitted,
        "samples": client["num_samples"],
        "loss": float(loss.detach().cpu()),
        "accuracy": accuracy_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy()
        ),
        "f1": f1_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy(),
            zero_division=0
        ),
        "elapsed": elapsed,
        "raw_update_norm": raw_norm,
        "clip_scale": clip_scale,
    }

def run_federated_experiment(
    num_clients=5,
    algorithm="FedProx",
    partition_profile="mild_non_iid",
    rounds=10,
    local_epochs=1,
    seed=42,
    noise_multiplier=0.0,
    dirichlet_alpha=0.5,
    record_client_log=True
):
    set_all_seeds(seed)

    clients = partition_train_nodes(
        graph_data,
        num_clients=num_clients,
        profile=partition_profile,
        seed=seed,
        dirichlet_alpha=dirichlet_alpha
    )

    model = make_experiment_model()
    global_state = clone_state_dict(model)

    history = []
    client_log = []

    model_bytes = sum(
        t.numel() * t.element_size()
        for t in global_state.values()
    )

    total_communication_mb = 0.0

    for round_id in range(1, rounds + 1):
        round_start = time.perf_counter()

        local_states, local_weights = [], []
        local_losses, local_times = [], []
        update_norms, clip_scales = [], []

        for client in clients:
            result = train_one_client_v2(
                global_state=global_state,
                client=client,
                algorithm=algorithm,
                local_epochs=local_epochs,
                lr=LOCAL_LR,
                fedprox_mu=FEDPROX_MU,
                clip_norm=UPDATE_CLIP_NORM,
                noise_multiplier=noise_multiplier
            )

            local_states.append(result["state"])
            local_weights.append(result["samples"])
            local_losses.append(result["loss"])
            local_times.append(result["elapsed"])
            update_norms.append(result["raw_update_norm"])
            clip_scales.append(result["clip_scale"])

            if record_client_log:
                client_log.append({
                    "Seed": seed,
                    "Algorithm": algorithm,
                    "Partition": partition_profile,
                    "Round": round_id,
                    "Client": client["client_id"],
                    "Samples": client["num_samples"],
                    "Benign_Count": client["class_0"],
                    "Attack_Count": client["class_1"],
                    "Local_Loss": result["loss"],
                    "Local_Accuracy": result["accuracy"],
                    "Local_F1": result["f1"],
                    "Update_Norm": result["raw_update_norm"],
                    "Clip_Scale": result["clip_scale"],
                    "Local_Time_s": result["elapsed"]
                })

        global_state = aggregate_client_states(
            local_states, local_weights, global_state
        )
        model.load_state_dict(global_state)

        val = evaluate_model_metrics(
            model, graph_data, graph_data.val_mask
        )

        round_time = time.perf_counter() - round_start

        upload_mb = (model_bytes * len(clients)) / (1024**2)
        download_mb = (model_bytes * len(clients)) / (1024**2)
        comm_mb = upload_mb + download_mb
        total_communication_mb += comm_mb

        history.append({
            "Seed": seed,
            "Algorithm": algorithm,
            "Partition": partition_profile,
            "Noise_Multiplier": noise_multiplier,
            "Round": round_id,
            "Selected_Clients": len(clients),
            "Validation_Loss": val["Validation Loss"],
            "Validation_Accuracy": val["Accuracy"],
            "Validation_Precision": val["Precision"],
            "Validation_Recall": val["Recall"],
            "Validation_F1": val["F1"],
            "Mean_Local_Loss": float(np.mean(local_losses)),
            "Mean_Update_Norm": float(np.mean(update_norms)),
            "Mean_Clip_Scale": float(np.mean(clip_scales)),
            "Round_Runtime_s": round_time,
            "Avg_Local_Runtime_s": float(np.mean(local_times)),
            "Model_Size_MB": model_bytes / (1024**2),
            "Communication_MB_per_Round": comm_mb,
            "Cumulative_Communication_MB": total_communication_mb,
            "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        })

    test = evaluate_model_metrics(
        model, graph_data, graph_data.test_mask
    )

    best_round_idx = int(
        np.argmax([r["Validation_F1"] for r in history])
    )
    best_round = history[best_round_idx]["Round"]

    final_row = {
        "Seed": seed,
        "Clients": num_clients,
        "Algorithm": algorithm,
        "Partition": partition_profile,
        "Rounds": rounds,
        "Local_Epochs": local_epochs,
        "Noise_Multiplier": noise_multiplier,
        "Best_Validation_Round": best_round,
        "Accuracy": test["Accuracy"],
        "Precision": test["Precision"],
        "Recall": test["Recall"],
        "F1": test["F1"],
        "ROC-AUC": test["ROC-AUC"],
        "PR-AUC": test["PR-AUC"],
        "Specificity": test["Specificity"],
        "FAR_FPR": test["FAR_FPR"],
        "FRR_FNR": test["FRR_FNR"],
        "Balanced_Accuracy": test["Balanced Accuracy"],
        "MCC": test["MCC"],
        "TN": test["TN"], "FP": test["FP"],
        "FN": test["FN"], "TP": test["TP"],
        "Average_Round_Runtime_s": float(np.mean([r["Round_Runtime_s"] for r in history])),
        "Communication_MB_per_Round": float(np.mean([r["Communication_MB_per_Round"] for r in history])),
        "Total_Communication_MB": float(history[-1]["Cumulative_Communication_MB"]),
        "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        "Convergence_Round_ValF1": best_round,
    }

    return {
        "model": model,
        "clients": clients,
        "history": pd.DataFrame(history),
        "client_log": pd.DataFrame(client_log),
        "final": final_row,
        "test": test,
    }


FINAL_RUN = run_federated_experiment(
    num_clients=EXPERIMENT_CONFIG["final_num_clients"],
    algorithm="FedProx",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED,
    noise_multiplier=EXPERIMENT_CONFIG["primary_noise_multiplier"],
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"],
)

FINAL_FEDERATED_RESULTS = pd.DataFrame([FINAL_RUN["final"]])
save_csv(FINAL_FEDERATED_RESULTS, "Federated_Final_Test_Results.csv")
save_csv(FINAL_RUN["history"], "Federated_Round_History.csv")
save_csv(FINAL_RUN["client_log"], "Federated_Client_Log.csv")

# Exact confusion matrix CSV
final_cm = np.array([
    [FINAL_RUN["final"]["TN"], FINAL_RUN["final"]["FP"]],
    [FINAL_RUN["final"]["FN"], FINAL_RUN["final"]["TP"]]
], dtype=int)
cm_df = pd.DataFrame(
    final_cm,
    index=["Actual_Benign", "Actual_Attack"],
    columns=["Predicted_Benign", "Predicted_Attack"]
)
cm_df.to_csv(os.path.join(FL_RESULT_DIR, "Federated_Confusion_Matrix.csv"))

torch.save(
    {
        "model_state_dict": FINAL_RUN["model"].state_dict(),
        "config": EXPERIMENT_CONFIG,
        "final_metrics": FINAL_RUN["final"]
    },
    os.path.join(FL_RESULT_DIR, "Federated_Final_FedProx_Model.pt")
)

print("\nFINAL 5-CLIENT FEDPROX RESULT")
display(FINAL_FEDERATED_RESULTS)

# ------------------------------------------------------------
# RECOMMENDATION 9 — PUBLICATION-QUALITY FINAL FIGURES
# ------------------------------------------------------------
def plot_and_save(x, y, xlabel, ylabel, title, filename, marker="o"):
    plt.figure(figsize=(9, 6))
    plt.plot(x, y, marker=marker, linewidth=2)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    path = os.path.join(FL_GRAPH_DIR, filename)
    plt.savefig(path, dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
    return path

h = FINAL_RUN["history"]
plot_and_save(h["Round"], h["Validation_Loss"],
              "Federated Round", "Validation Loss",
              "FedProx Validation Loss",
              "Final_FedProx_Validation_Loss.png")
plot_and_save(h["Round"], h["Validation_Accuracy"] * 100,
              "Federated Round", "Validation Accuracy (%)",
              "FedProx Validation Accuracy",
              "Final_FedProx_Validation_Accuracy.png")
plot_and_save(h["Round"], h["Validation_F1"] * 100,
              "Federated Round", "Validation F1 (%)",
              "FedProx Validation F1",
              "Final_FedProx_Validation_F1.png")
plot_and_save(h["Round"], h["Cumulative_Communication_MB"],
              "Federated Round", "Cumulative Communication (MB)",
              "FedProx Cumulative Communication",
              "Final_FedProx_Cumulative_Communication.png")
plot_and_save(h["Round"], h["Round_Runtime_s"],
              "Federated Round", "Round Runtime (s)",
              "FedProx Round Runtime",
              "Final_FedProx_Round_Runtime.png")

# Final confusion matrix
plt.figure(figsize=(7, 6))
plt.imshow(final_cm, interpolation="nearest")
plt.colorbar()
plt.xticks([0, 1], ["Benign", "Attack"])
plt.yticks([0, 1], ["Benign", "Attack"])
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("FedProx Final Confusion Matrix")
for r in range(2):
    for c in range(2):
        plt.text(c, r, f"{final_cm[r,c]:,}",
                 ha="center", va="center", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_Confusion_Matrix.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ROC/PR curves
y_true = FINAL_RUN["test"]["y_true"]
y_pred = FINAL_RUN["test"]["y_pred"]
y_prob = FINAL_RUN["test"]["y_prob"]

try:
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(fpr, tpr, linewidth=2,
             label=f"FedProx ROC-AUC={FINAL_RUN['final']['ROC-AUC']:.4f}")
    plt.plot([0,1], [0,1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("FedProx ROC Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_ROC.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("ROC figure skipped:", e)

try:
    pr_precision, pr_recall, _ = precision_recall_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(pr_recall, pr_precision, linewidth=2,
             label=f"FedProx PR-AUC={FINAL_RUN['final']['PR-AUC']:.4f}")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("FedProx Precision-Recall Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_PR.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("PR figure skipped:", e)

# ------------------------------------------------------------
# RECOMMENDATION 10 — FEDAVG VS FEDPROX, SAME EVERYTHING ELSE
# ------------------------------------------------------------
# Same split, clients, model, rounds, local epochs and seed.
# Only the server/client objective differs: FedAvg mu=0 vs FedProx mu=0.01.
fedavg_run = run_federated_experiment(
    num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedAvg",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED, noise_multiplier=0.0,
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
)
fedprox_run = FINAL_RUN

fedavg_history = fedavg_run["history"].copy()
fedprox_history = fedprox_run["history"].copy()

comparison_rows = []
for _, a in fedavg_history.iterrows():
    p = fedprox_history[fedprox_history["Round"] == a["Round"]].iloc[0]
    comparison_rows.append({
        "Seed": SEED,
        "Round": int(a["Round"]),
        "FedAvg_Validation_Accuracy": a["Validation_Accuracy"],
        "FedProx_Validation_Accuracy": p["Validation_Accuracy"],
        "FedAvg_Validation_F1": a["Validation_F1"],
        "FedProx_Validation_F1": p["Validation_F1"],
        "FedAvg_Validation_Loss": a["Validation_Loss"],
        "FedProx_Validation_Loss": p["Validation_Loss"],
        "FedAvg_Update_Norm": a["Mean_Update_Norm"],
        "FedProx_Update_Norm": p["Mean_Update_Norm"],
        "FedAvg_Round_Runtime_s": a["Round_Runtime_s"],
        "FedProx_Round_Runtime_s": p["Round_Runtime_s"],
    })

comparison_rows.append({
    "Seed": SEED, "Round": "FINAL",
    "FedAvg_Accuracy": fedavg_run["final"]["Accuracy"],
    "FedProx_Accuracy": fedprox_run["final"]["Accuracy"],
    "FedAvg_Precision": fedavg_run["final"]["Precision"],
    "FedProx_Precision": fedprox_run["final"]["Precision"],
    "FedAvg_Recall": fedavg_run["final"]["Recall"],
    "FedProx_Recall": fedprox_run["final"]["Recall"],
    "FedAvg_F1": fedavg_run["final"]["F1"],
    "FedProx_F1": fedprox_run["final"]["F1"],
    "FedAvg_ROC-AUC": fedavg_run["final"]["ROC-AUC"],
    "FedProx_ROC-AUC": fedprox_run["final"]["ROC-AUC"],
    "FedAvg_Best_Validation_Round": fedavg_run["final"]["Best_Validation_Round"],
    "FedProx_Best_Validation_Round": fedprox_run["final"]["Best_Validation_Round"],
})

FEDAVG_VS_FEDPROX = pd.DataFrame(comparison_rows)
save_csv(FEDAVG_VS_FEDPROX, "FedAvg_vs_FedProx.csv")

plot_and_save(
    fedavg_history["Round"],
    fedavg_history["Validation_F1"] * 100,
    "Round", "Validation F1 (%)",
    "FedAvg vs FedProx — FedAvg",
    "FedAvg_vs_FedProx_FedAvg_F1.png"
)
plt.figure(figsize=(9,6))
plt.plot(fedavg_history["Round"], fedavg_history["Validation_F1"]*100,
         marker="o", linewidth=2, label="FedAvg")
plt.plot(fedprox_history["Round"], fedprox_history["Validation_F1"]*100,
         marker="s", linewidth=2, label="FedProx")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("FedAvg vs FedProx — Convergence")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "FedAvg_vs_FedProx_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ------------------------------------------------------------
# RECOMMENDATION — TGAT-STYLE BASELINE COMPARISON
# ------------------------------------------------------------
# HONESTY NOTE: graph_data (STEP 26) has no per-edge timestamps, so a
# literal TGAT (Xu et al., 2020 — temporal-attention over timestamped
# edges) cannot be trained on it. This substitutes the uploaded model's
# spatial GraphSAGE encoder with a multi-head GATConv encoder (graph
# attention, the mechanism TGAT also builds on) while keeping every other
# component -- input dimension, BiLSTM, temporal attention, fusion head,
# classifier, optimizer, seed, split, client partition and federated
# protocol -- byte-for-byte identical to NovelTemporalClassifier / the
# 5-client FedProx run above, so the comparison isolates the effect of
# attention-based neighborhood aggregation vs. GraphSAGE's mean
# aggregation. Results are reported as "TGAT-style (GAT)", never as a
# reproduction of the published TGAT architecture.
if EXPERIMENT_CONFIG["run_tgat_comparison"]:
    print("\n=== TGAT-STYLE (GAT) VS GRAPHSAGE COMPARISON ===")

    class TGATStyleClassifier(nn.Module):
        def __init__(self, input_dim, hidden_dim=128, embedding_dim=64,
                     num_layers=2, num_classes=2, dropout=0.30, heads=4):
            super().__init__()
            self.gat1 = GATConv(
                input_dim, hidden_dim // heads, heads=heads, dropout=dropout
            )
            self.gat2 = GATConv(
                hidden_dim, embedding_dim, heads=1, concat=False, dropout=dropout
            )
            self.norm1 = nn.LayerNorm(hidden_dim)
            self.norm2 = nn.LayerNorm(embedding_dim)
            self.attention = nn.Sequential(
                nn.Linear(embedding_dim, embedding_dim // 2), nn.Tanh(),
                nn.Linear(embedding_dim // 2, 1)
            )
            self.bilstm = nn.LSTM(
                input_size=embedding_dim, hidden_size=hidden_dim,
                num_layers=num_layers, batch_first=True, bidirectional=True,
                dropout=dropout if num_layers > 1 else 0
            )
            self.temporal_attention = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.Tanh(),
                nn.Linear(hidden_dim, 1)
            )
            self.gnn_projection = nn.Linear(embedding_dim, hidden_dim * 2)
            self.fusion = nn.Sequential(
                nn.Linear(hidden_dim * 4, hidden_dim * 2),
                nn.ReLU(), nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.ReLU(),
                nn.Dropout(dropout), nn.Linear(hidden_dim, num_classes)
            )
            self.dropout = nn.Dropout(dropout)

        def forward(self, x, edge_index=None):
            x1 = self.gat1(x, edge_index)
            x1 = F.relu(self.norm1(x1))
            x1 = self.dropout(x1)

            x2 = self.gat2(x1, edge_index)
            x2 = F.relu(self.norm2(x2))

            attn_scores = self.attention(x2)
            attn_weights = torch.softmax(attn_scores, dim=0)
            attended = x2 * attn_weights

            sequence = attended.unsqueeze(1)
            lstm_output, _ = self.bilstm(sequence)

            t_scores = self.temporal_attention(lstm_output)
            t_weights = torch.softmax(t_scores, dim=1)
            temporal_context = (lstm_output * t_weights).sum(dim=1)

            gnn_residual = self.gnn_projection(x2)
            fused = self.fusion(torch.cat([temporal_context, gnn_residual], dim=1))
            return self.classifier(fused)

    def make_tgat_model():
        return TGATStyleClassifier(
            input_dim=graph_data.num_features,
            hidden_dim=128, embedding_dim=64,
            num_layers=2, num_classes=2, dropout=0.30
        ).to(DEVICE)

    def train_one_client_generic(model_factory, global_state, client,
                                  algorithm, local_epochs, lr, fedprox_mu):
        # Mirrors train_one_client_v2 exactly, parametrized by model_factory
        # so the GraphSAGE and TGAT-style runs are trained identically in
        # every other respect (loss, class weighting, optimizer, clipping).
        local_model = model_factory()
        local_model.load_state_dict(global_state)
        local_model.train()

        x_local, y_local, edge_local = materialize_client(graph_data, client)

        counts = torch.bincount(y_local, minlength=2).float()
        safe = torch.clamp(counts, min=1.0)
        class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
        criterion = nn.CrossEntropyLoss(weight=class_weights)

        optimizer = torch.optim.AdamW(
            local_model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY
        )
        global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
        start = time.perf_counter()

        for _ in range(local_epochs):
            optimizer.zero_grad(set_to_none=True)
            logits = local_model(x_local, edge_local)
            if isinstance(logits, (tuple, list)):
                logits = logits[0]
            loss = criterion(logits, y_local)

            if algorithm.lower() == "fedprox" and fedprox_mu > 0:
                prox = torch.zeros((), device=DEVICE)
                for name, param in local_model.named_parameters():
                    prox += torch.sum((param - global_device[name]) ** 2)
                loss = loss + 0.5 * fedprox_mu * prox

            loss.backward()
            torch.nn.utils.clip_grad_norm_(local_model.parameters(), max_norm=5.0)
            optimizer.step()

        elapsed = time.perf_counter() - start
        local_state = clone_state_dict(local_model)

        with torch.no_grad():
            local_model.eval()
            pred = torch.argmax(local_model(x_local, edge_local), dim=1)

        return {
            "state": local_state,
            "samples": client["num_samples"],
            "loss": float(loss.detach().cpu()),
            "elapsed": elapsed,
            "accuracy": accuracy_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy()
            ),
            "f1": f1_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy(),
                zero_division=0
            ),
        }

    def run_federated_experiment_generic(model_factory, num_clients, algorithm,
                                          partition_profile, rounds, local_epochs,
                                          seed):
        set_all_seeds(seed)
        clients = partition_train_nodes(
            graph_data, num_clients=num_clients, profile=partition_profile,
            seed=seed, dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        model = model_factory()
        global_state = clone_state_dict(model)
        history = []

        for round_id in range(1, rounds + 1):
            states, weights = [], []
            for client in clients:
                result = train_one_client_generic(
                    model_factory, global_state, client,
                    algorithm=algorithm, local_epochs=local_epochs,
                    lr=LOCAL_LR, fedprox_mu=FEDPROX_MU
                )
                states.append(result["state"])
                weights.append(result["samples"])

            global_state = aggregate_client_states(states, weights, global_state)
            model.load_state_dict(global_state)

            val = evaluate_model_metrics(model, graph_data, graph_data.val_mask)
            history.append({
                "Round": round_id,
                "Validation_Accuracy": val["Accuracy"],
                "Validation_F1": val["F1"],
            })

        test = evaluate_model_metrics(model, graph_data, graph_data.test_mask)
        return {"model": model, "history": pd.DataFrame(history), "test": test}

    # Same 5-client, same rounds/local-epochs/seed/partition as the primary
    # FedProx run above, so the GraphSAGE row is a like-for-like comparison
    # rather than a re-use of the 50-client FINAL_RUN numbers.
    graphsage_5client_run = run_federated_experiment(
        num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED, noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )

    tgat_run = run_federated_experiment_generic(
        make_tgat_model, num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED
    )

    TGAT_COMPARISON = pd.DataFrame([
        {
            "Model": "GraphSAGE (uploaded final classifier)",
            "Clients": 5,
            "Accuracy": graphsage_5client_run["test"]["Accuracy"],
            "F1": graphsage_5client_run["test"]["F1"],
            "ROC-AUC": graphsage_5client_run["test"]["ROC-AUC"],
            "MCC": graphsage_5client_run["test"]["MCC"],
        },
        {
            "Model": "TGAT-style (GATConv, no edge timestamps)",
            "Clients": 5,
            "Accuracy": tgat_run["test"]["Accuracy"],
            "F1": tgat_run["test"]["F1"],
            "ROC-AUC": tgat_run["test"]["ROC-AUC"],
            "MCC": tgat_run["test"]["MCC"],
        },
    ])
    save_csv(TGAT_COMPARISON, "TGAT_Comparison.csv")
    display(TGAT_COMPARISON)

    plt.figure(figsize=(9, 6))
    plt.plot(graphsage_5client_run["history"]["Round"],
              graphsage_5client_run["history"]["Validation_F1"] * 100,
              marker="o", linewidth=2, label="GraphSAGE")
    plt.plot(tgat_run["history"]["Round"],
              tgat_run["history"]["Validation_F1"] * 100,
              marker="s", linewidth=2, label="TGAT-style (GAT)")
    plt.xlabel("Federated Round")
    plt.ylabel("Validation F1 (%)")
    plt.title("GraphSAGE vs TGAT-style Encoder — Convergence (5 clients)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "TGAT_vs_GraphSAGE_Convergence.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "TGAT comparison not executed"
    }]), "TGAT_Comparison.csv")

# ------------------------------------------------------------
# RECOMMENDATION 11 — SCALABILITY: ACTUALLY RUN ONLY EVALUATED COUNTS
# ------------------------------------------------------------
scalability_counts = list(EXPERIMENT_CONFIG["scalability_client_counts"])

if EXPERIMENT_CONFIG["run_optional_large_scalability"]:
    scalability_counts += EXPERIMENT_CONFIG["optional_large_client_counts"]

SCALABILITY_ROWS = []
SCALABILITY_HISTORIES = []

for n_clients in scalability_counts:
    print(f"\n=== SCALABILITY: {n_clients} CLIENTS ===")

    run = run_federated_experiment(
        num_clients=n_clients,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    SCALABILITY_HISTORIES.append(run["history"])

    f = run["final"]
    SCALABILITY_ROWS.append({
        "Clients": n_clients,
        "Accuracy": f["Accuracy"],
        "F1": f["F1"],
        "Average_Round_Runtime_s": f["Average_Round_Runtime_s"],
        "Communication_MB_per_Round": f["Communication_MB_per_Round"],
        "Total_Communication_MB": f["Total_Communication_MB"],
        "Samples_Per_Client_Mean": f["Samples_Per_Client_Mean"],
        "Convergence_Round": f["Convergence_Round_ValF1"],
        "Rounds_Evaluated": EXPERIMENT_CONFIG["rounds"],
        "Status": "COMPLETED"
    })

SCALABILITY_RESULTS = pd.DataFrame(SCALABILITY_ROWS)
save_csv(SCALABILITY_RESULTS, "Scalability_Experiment.csv")

# Scalability figures
for metric, ylabel, fname in [
    ("Average_Round_Runtime_s", "Average Round Runtime (s)", "Scalability_Runtime.png"),
    ("Communication_MB_per_Round", "Communication (MB/round)", "Scalability_Communication.png"),
    ("Accuracy", "Final Test Accuracy", "Scalability_Accuracy.png"),
    ("F1", "Final Test F1", "Scalability_F1.png"),
]:
    plt.figure(figsize=(9,6))
    plt.plot(SCALABILITY_RESULTS["Clients"], SCALABILITY_RESULTS[metric],
             marker="o", linewidth=2)
    plt.xlabel("Number of Clients")
    plt.ylabel(ylabel)
    plt.title(f"Scalability — {ylabel}")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, fname), dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()


plt.figure(figsize=(10,6))
for hist in SCALABILITY_HISTORIES:
    n = int(hist["Selected_Clients"].iloc[0])
    plt.plot(hist["Round"], hist["Validation_F1"]*100,
             marker="o", linewidth=2, label=f"{n} clients")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("Scalability — Convergence Behavior")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Scalability_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


privacy_rows = []
for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
    print(f"\n=== GAUSSIAN UPDATE PERTURBATION: noise={noise} ===")

    run = run_federated_experiment(
        num_clients=5,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=noise,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    f = run["final"]
    privacy_rows.append({
        "Noise_Multiplier": noise,
        "Accuracy": f["Accuracy"],
        "Precision": f["Precision"],
        "Recall": f["Recall"],
        "F1": f["F1"],
        "Status": "COMPLETED",
        "Interpretation": "Privacy-aware Gaussian update perturbation; NOT formal DP"
    })

PRIVACY_RESULTS = pd.DataFrame(privacy_rows)
save_csv(PRIVACY_RESULTS, "Privacy_Sensitivity_Experiment.csv")

plt.figure(figsize=(10,6))
for col, label in [
    ("Accuracy", "Accuracy"),
    ("Precision", "Precision"),
    ("Recall", "Recall"),
    ("F1", "F1"),
]:
    plt.plot(PRIVACY_RESULTS["Noise_Multiplier"],
             PRIVACY_RESULTS[col]*100,
             marker="o", linewidth=2, label=label)
plt.xlabel("Gaussian Update-Noise Multiplier")
plt.ylabel("Test Score (%)")
plt.title("Privacy-Aware Gaussian Update Perturbation Sensitivity")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Privacy_Sensitivity.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


if EXPERIMENT_CONFIG["run_ablation"]:
    print("Ablation execution requires explicit variant-specific model factories.")
    print("No ablation values are generated by default.")
else:
    ablation_status = pd.DataFrame([
        ["Full Model", "NOT_RUN", "Enable run_ablation and variant factories"],
        ["Without BiLSTM / reconstruction", "NOT_RUN", "Requires controlled model variant"],
        ["Without Correlation Filtering", "NOT_RUN", "Current final input pipeline does not apply filtering"],
        ["Without Attention", "NOT_RUN", "Requires controlled model variant"],
        ["FedAvg instead of FedProx", "COMPLETED_SEPARATELY", "See FedAvg_vs_FedProx.csv"],
        ["GraphSAGE only", "NOT_RUN", "Requires controlled model variant"],
        ["GraphSAGE + BiLSTM", "NOT_RUN", "Requires controlled model variant"],
    ], columns=["Variant", "Status", "Reason"])
    save_csv(ablation_status, "Ablation_Study.csv")
    display(ablation_status)


if EXPERIMENT_CONFIG["run_imputer_reconstruction"]:
    raise NotImplementedError(
        "Enable only after adding a complete train/validation/test reconstruction "
        "protocol for BiLSTMImputer. Do not report the untrained imputer as completed."
    )
else:
    reconstruction_status = pd.DataFrame([{
        "Missing_Rate": globals().get("MISSING_RATE", 0.05),
        "Sequence_Length": globals().get("SEQUENCE_LENGTH", 10),
        "RMSE": np.nan,
        "MAE": np.nan,
        "Status": "NOT_COMPLETED",
        "Reason": "BiLSTMImputer is defined but no completed training/evaluation was found"
    }])
    save_csv(reconstruction_status, "BiLSTM_Reconstruction_Status.csv")


graphfedai_available = any(
    name.lower() == "graphfedai" or "graphfedai" in name.lower()
    for name in globals().keys()
)

if graphfedai_available:
    print("GraphFedAI symbol found in runtime; implement the exact baseline adapter before running.")
else:
    baseline_status = pd.DataFrame([{
        "Baseline": "GraphFedAI",
        "Status": "NOT_COMPLETED",
        "Reason": "No GraphFedAI implementation is present in the uploaded notebook",
        "Required_Settings": json.dumps({
            "dataset_split": "same as EFG-IDS",
            "clients": 5,
            "rounds": EXPERIMENT_CONFIG["rounds"],
            "local_epochs": EXPERIMENT_CONFIG["local_epochs"],
            "seed": SEED
        })
    }])
    save_csv(baseline_status, "GraphFedAI_Baseline_Comparison.csv")
    display(baseline_status)


if EXPERIMENT_CONFIG["run_repeated_seeds"]:
    repeated_rows = []
    for seed in EXPERIMENT_CONFIG["seeds"]:
        # Final FedProx
        rp = run_federated_experiment(
            EXPERIMENT_CONFIG["final_num_clients"], "FedProx", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        # FedAvg
        ra = run_federated_experiment(
            EXPERIMENT_CONFIG["final_num_clients"], "FedAvg", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        repeated_rows.extend([
            {"Seed": seed, "Experiment": "FedProx", **rp["final"]},
            {"Seed": seed, "Experiment": "FedAvg", **ra["final"]},
        ])

    repeated_raw = pd.DataFrame(repeated_rows)
    save_csv(repeated_raw, "Repeated_Run_Raw.csv")

    summary = (
        repeated_raw
        .groupby("Experiment")
        .agg(
            Accuracy_mean=("Accuracy", "mean"),
            Accuracy_std=("Accuracy", "std"),
            Precision_mean=("Precision", "mean"),
            Precision_std=("Precision", "std"),
            Recall_mean=("Recall", "mean"),
            Recall_std=("Recall", "std"),
            F1_mean=("F1", "mean"),
            F1_std=("F1", "std"),
            ROCAUC_mean=("ROC-AUC", "mean"),
            ROCAUC_std=("ROC-AUC", "std"),
        )
        .reset_index()
    )
    save_csv(summary, "Repeated_Run_Summary.csv")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Seeds": "42,43,44,45,46",
        "Reason": "Repeated runs disabled by default because of computational cost"
    }]), "Repeated_Run_Summary.csv")

if EXPERIMENT_CONFIG["run_external_bot_iot"]:
    print("\n=== EXTERNAL VALIDATION: BoT-IoT / IDS dataset ===")

    bot_iot_files = sorted(IDS_PATH.glob("*.csv"))
    if not bot_iot_files:
        raise FileNotFoundError(
            f"No CSV files found under {IDS_PATH}; cannot run BoT-IoT "
            "external validation."
        )

    bot_frames = []
    for f in bot_iot_files:
        try:
            bot_frames.append(pd.read_csv(f, low_memory=False))
        except Exception as e:
            print(f"  Skipping {f.name}: {e}")

    BOT_RAW = pd.concat(bot_frames, ignore_index=True)

    bot_label_candidates = [
        c for c in BOT_RAW.columns
        if c.lower() in ("label", "attack", "category", "class")
    ]
    if not bot_label_candidates:
        raise KeyError("Could not locate a label column in the BoT-IoT/IDS files.")
    bot_label_col = bot_label_candidates[0]

    # Reuse the exact binary-label convention from CELL 12/13
    # (create_binary_labels) so "benign" is mapped to 0 identically to the
    # CIC-IoT-2023 training labels.
    BOT_RAW["binary_label"] = create_binary_labels(BOT_RAW[bot_label_col])

    # Feature alignment onto the trained CIC-IoT-2023 feature space:
    # - Columns the scaler was fit on and that BoT-IoT also has are used
    #   directly (matched_features).
    # - Columns the scaler was fit on but BoT-IoT does NOT have are filled
    #   with the TRAINING median (train_median from STEP 22) -- never a
    #   BoT-IoT statistic -- so no target-domain information leaks into a
    #   feature the model was never actually shown from this domain.
    aligned = pd.DataFrame(index=BOT_RAW.index)
    matched_features, filled_features = [], []
    for col in FEATURE_COLUMNS:
        if col in BOT_RAW.columns:
            aligned[col] = pd.to_numeric(BOT_RAW[col], errors="coerce")
            matched_features.append(col)
        else:
            aligned[col] = train_median[col]
            filled_features.append(col)

    aligned = aligned.replace([np.inf, -np.inf], np.nan)
    aligned = aligned.fillna(train_median)

    BOT_X = scaler.transform(aligned[FEATURE_COLUMNS])
    BOT_y = BOT_RAW["binary_label"].astype(int).to_numpy()

    print(f"BoT-IoT rows: {len(BOT_X)}")
    print(f"Matched CIC-IoT-2023 features: {len(matched_features)} / {len(FEATURE_COLUMNS)}")
    print(f"Features filled with training median (not present in BoT-IoT): {len(filled_features)}")
    if filled_features:
        print("  ", filled_features)

    # Build a self-contained kNN graph over the BoT-IoT rows only (same k
    # and metric as STEP 26), so the trained GraphSAGE layers receive the
    # same kind of local-neighborhood structure they were trained on. This
    # graph is independent of graph_data -- no BoT-IoT node is ever
    # connected to a CIC-IoT-2023 node.
    bot_knn = NearestNeighbors(n_neighbors=K_NEIGHBORS + 1, metric="euclidean", n_jobs=-1)
    bot_knn.fit(BOT_X)
    _, bot_neighbors = bot_knn.kneighbors(BOT_X)

    bot_src, bot_dst = [], []
    for i in range(len(BOT_X)):
        for j in bot_neighbors[i, 1:]:
            bot_src.append(i)
            bot_dst.append(j)
    bot_edge_index = torch.tensor([bot_src, bot_dst], dtype=torch.long)

    bot_graph = Data(
        x=torch.tensor(BOT_X, dtype=torch.float32),
        y=torch.tensor(BOT_y, dtype=torch.long),
        edge_index=bot_edge_index
    ).to(DEVICE)

    external_model = FINAL_RUN["model"]
    external_model.eval()
    with torch.no_grad():
        out = external_model(bot_graph.x, bot_graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        bot_probs = torch.softmax(logits, dim=1)[:, 1]
        bot_preds = torch.argmax(logits, dim=1)

    bot_metrics = compute_binary_metrics(
        bot_graph.y.detach().cpu().numpy(),
        bot_preds.detach().cpu().numpy(),
        bot_probs.detach().cpu().numpy()
    )

    EXTERNAL_BOTIOT_RESULTS = pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "COMPLETED",
        "Rows_Evaluated": len(BOT_X),
        "Matched_Features": len(matched_features),
        "Filled_Features": len(filled_features),
        "K_Neighbors": K_NEIGHBORS,
        "Model_Source": "FINAL_RUN (100-client FedProx, trained on CIC-IoT-2023 only)",
        "Accuracy": bot_metrics["Accuracy"],
        "Precision": bot_metrics["Precision"],
        "Recall": bot_metrics["Recall"],
        "F1": bot_metrics["F1"],
        "ROC-AUC": bot_metrics["ROC-AUC"],
        "PR-AUC": bot_metrics["PR-AUC"],
        "Specificity": bot_metrics["Specificity"],
        "Balanced_Accuracy": bot_metrics["Balanced Accuracy"],
        "MCC": bot_metrics["MCC"],
        "TN": bot_metrics["TN"], "FP": bot_metrics["FP"],
        "FN": bot_metrics["FN"], "TP": bot_metrics["TP"],
    }])
    save_csv(EXTERNAL_BOTIOT_RESULTS, "External_Validation_BoTIoT.csv")
    display(EXTERNAL_BOTIOT_RESULTS)

    bot_cm = np.array([
        [bot_metrics["TN"], bot_metrics["FP"]],
        [bot_metrics["FN"], bot_metrics["TP"]]
    ])
    plt.figure(figsize=(7, 6))
    plt.imshow(bot_cm, interpolation="nearest")
    plt.colorbar()
    plt.xticks([0, 1], ["Benign", "Attack"])
    plt.yticks([0, 1], ["Benign", "Attack"])
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.title("BoT-IoT External Validation — Confusion Matrix")
    for r in range(2):
        for c in range(2):
            plt.text(c, r, f"{bot_cm[r, c]:,}", ha="center", va="center", fontsize=14)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "External_BoTIoT_Confusion_Matrix.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "NOT_COMPLETED",
        "Reason": "External feature mapping/evaluation not executed"
    }]), "External_Validation_BoTIoT.csv")


if EXPERIMENT_CONFIG["run_gnnexplainer"]:
    print("\n=== GNNEXPLAINER: TP / TN / FP / FN CASE EXPLANATIONS ===")
    from torch_geometric.explain import Explainer, GNNExplainer

    explain_model = FINAL_RUN["model"]
    explain_model.eval()

    test_idx = torch.where(graph_data.test_mask)[0]
    y_true_test = graph_data.y[test_idx].detach().cpu().numpy()
    y_pred_test = FINAL_RUN["test"]["y_pred"]  # same order as test_idx (mask indexing)

    tp_idx = test_idx[(y_true_test == 1) & (y_pred_test == 1)]
    tn_idx = test_idx[(y_true_test == 0) & (y_pred_test == 0)]
    fp_idx = test_idx[(y_true_test == 0) & (y_pred_test == 1)]
    fn_idx = test_idx[(y_true_test == 1) & (y_pred_test == 0)]

    explain_rng = np.random.default_rng(SEED)

    def sample_case_nodes(pool, n):
        pool = pool.detach().cpu().numpy()
        if len(pool) == 0:
            return np.array([], dtype=np.int64)
        return explain_rng.choice(pool, size=min(n, len(pool)), replace=False)

    n_per_bucket = EXPERIMENT_CONFIG["gnnexplainer_cases_per_bucket"]
    case_nodes = {
        "TP": sample_case_nodes(tp_idx, n_per_bucket),
        "TN": sample_case_nodes(tn_idx, n_per_bucket),
        "FP": sample_case_nodes(fp_idx, n_per_bucket),
        "FN": sample_case_nodes(fn_idx, n_per_bucket),
    }
    for case, nodes in case_nodes.items():
        print(f"  {case}: {len(nodes)} case(s) sampled (pool size available)")

    # NovelTemporalClassifier.forward(x, edge_index) returns raw per-node
    # logits directly (not a tuple), matching Explainer's expected model
    # output for node-level multiclass classification with return_type="raw".
    explainer = Explainer(
        model=explain_model,
        algorithm=GNNExplainer(epochs=EXPERIMENT_CONFIG["gnnexplainer_epochs"]),
        explanation_type="model",
        node_mask_type="attributes",
        edge_mask_type="object",
        model_config=dict(
            mode="multiclass_classification",
            task_level="node",
            return_type="raw",
        ),
    )

    explain_x = graph_data.x.to(DEVICE)
    explain_edge_index = graph_data.edge_index.to(DEVICE)

    explain_rows = []
    for case, nodes in case_nodes.items():
        for node_id in nodes:
            node_id = int(node_id)
            explanation = explainer(
                x=explain_x, edge_index=explain_edge_index, index=node_id
            )
            feat_importance = explanation.node_mask[node_id].detach().cpu().numpy()
            top_k = np.argsort(-feat_importance)[:10]
            for rank, feat_idx in enumerate(top_k, start=1):
                feat_name = (
                    FEATURE_COLUMNS[feat_idx]
                    if feat_idx < len(FEATURE_COLUMNS)
                    else f"feature_{feat_idx}"
                )
                explain_rows.append({
                    "Case": case,
                    "Node": node_id,
                    "Rank": rank,
                    "Feature": feat_name,
                    "Importance": float(feat_importance[feat_idx]),
                })

    GNNEXPLAINER_RESULTS = pd.DataFrame(explain_rows)
    save_csv(GNNEXPLAINER_RESULTS, "GNNExplainer_Results.csv")
    display(GNNEXPLAINER_RESULTS.head(20))

    if len(GNNEXPLAINER_RESULTS):
        top_features = (
            GNNEXPLAINER_RESULTS.groupby("Feature")["Importance"]
            .mean().sort_values(ascending=False).head(15)
        )
        plt.figure(figsize=(10, 6))
        top_features.plot(kind="barh")
        plt.gca().invert_yaxis()
        plt.xlabel("Mean GNNExplainer Feature Importance")
        plt.title("Top Features — GNNExplainer (TP/TN/FP/FN cases)")
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "GNNExplainer_Top_Features.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
    else:
        print("No test-set cases were available in one or more of TP/TN/FP/FN; "
              "GNNExplainer_Results.csv was written empty.")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Cases": "TP,TN,FP,FN",
        "Reason": "GNNExplainer not executed"
    }]), "GNNExplainer_Results.csv")

# ------------------------------------------------------------
# RECOMMENDATION 19 — FORMAL DIFFERENTIAL PRIVACY
# ------------------------------------------------------------
if EXPERIMENT_CONFIG["run_formal_dp"]:
    print("\n=== FORMAL DIFFERENTIAL PRIVACY ACCOUNTING ===")
    print("Mechanism being accounted: clip_and_noise_update (FL-3) clips each")
    print("client's per-round update to L2 norm <= update_clip_norm and adds")
    print("N(0, (noise_multiplier * update_clip_norm)^2) noise coordinatewise")
    print("before it leaves the client. This is a Gaussian mechanism with L2")
    print("sensitivity = update_clip_norm and noise multiplier = noise_multiplier,")
    print("applied once per client per federated round under full participation")
    print("(client sampling probability q = 1, as configured in FL-1/FL-2).")
    print("The epsilon below is therefore a CLIENT-LEVEL DP guarantee on the")
    print("transmitted update sequence, accounted analytically via Renyi DP")
    print("(RDP) composition -- it is not a sample-level guarantee over the")
    print("raw per-record training data, since no per-record subsampling or")
    print("per-record gradient clipping is performed anywhere in this pipeline.")

    def gaussian_mechanism_rdp(alpha, noise_multiplier):
        # Renyi divergence of order alpha between N(0, sigma^2) and
        # N(1, sigma^2) for a mechanism with L2 sensitivity 1 and Gaussian
        # noise of std = noise_multiplier (Mironov, 2017, Prop. 7):
        #   RDP(alpha) = alpha / (2 * sigma^2)
        return alpha / (2.0 * (noise_multiplier ** 2))

    def rdp_to_eps(rdp_orders, rdp_values, delta):
        # Standard RDP -> (epsilon, delta)-DP conversion (Balle et al. 2020 /
        # Mironov 2017), minimized over the candidate Renyi orders.
        candidates = [
            rdp + math.log(1.0 / delta) / (alpha - 1.0)
            for alpha, rdp in zip(rdp_orders, rdp_values)
            if alpha > 1.0
        ]
        return min(candidates) if candidates else float("inf")

    RDP_ORDERS = [1 + x / 10.0 for x in range(1, 100)] + list(range(12, 64))
    DP_DELTA = EXPERIMENT_CONFIG["dp_delta"]
    # One Gaussian mechanism application per client per round; composition
    # count uses the primary experiment's round budget.
    NUM_COMPOSITIONS = EXPERIMENT_CONFIG["rounds"]

    dp_rows = []
    for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
        if noise <= 0:
            dp_rows.append({
                "Noise_Multiplier": noise,
                "Rounds_Composed": NUM_COMPOSITIONS,
                "Delta": DP_DELTA,
                "Epsilon": float("inf"),
                "Note": "No noise added: the mechanism provides no DP guarantee "
                        "(epsilon is undefined/infinite).",
            })
            continue

        rdp_values = [
            NUM_COMPOSITIONS * gaussian_mechanism_rdp(alpha, noise)
            for alpha in RDP_ORDERS
        ]
        epsilon = rdp_to_eps(RDP_ORDERS, rdp_values, DP_DELTA)

        dp_rows.append({
            "Noise_Multiplier": noise,
            "Rounds_Composed": NUM_COMPOSITIONS,
            "Delta": DP_DELTA,
            "Epsilon": epsilon,
            "Note": "Client-level (epsilon, delta)-DP for the transmitted update "
                    "via RDP composition of the Gaussian mechanism (analytic "
                    "moments accountant; sensitivity = update_clip_norm; q = 1).",
        })

    FORMAL_DP_RESULTS = pd.DataFrame(dp_rows)
    save_csv(FORMAL_DP_RESULTS, "Formal_Differential_Privacy.csv")
    display(FORMAL_DP_RESULTS)

    finite_dp = FORMAL_DP_RESULTS[np.isfinite(FORMAL_DP_RESULTS["Epsilon"])]
    if len(finite_dp):
        plt.figure(figsize=(9, 6))
        plt.plot(finite_dp["Noise_Multiplier"], finite_dp["Epsilon"],
                  marker="o", linewidth=2)
        plt.xlabel("Gaussian Noise Multiplier")
        plt.ylabel(f"Epsilon (delta={DP_DELTA:g})")
        plt.title(f"Formal Client-Level DP: Epsilon vs Noise Multiplier "
                  f"over {NUM_COMPOSITIONS} Rounds")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "Formal_DP_Epsilon.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "Formal DP accounting not executed"
    }]), "Formal_Differential_Privacy.csv")

# ------------------------------------------------------------
# RECOMMENDATION 20 — EXPERIMENT MANIFEST / LOG EVIDENCE
# ------------------------------------------------------------
completed_in_this_cell = [
    "400-client FedProx final run (single seed 42 unless repeated-seed experiment is enabled)",
    "IID/mild-non-IID/strong-non-IID client distribution verification",
    "FedAvg vs FedProx controlled comparison",
    "100/200/400-client scalability",
    "0/0.005/0.01/0.02 Gaussian update perturbation sensitivity",
]
not_claimed_as_completed = [
    "GraphFedAI rerun (implementation absent)",
    "BiLSTM reconstruction RMSE/MAE",
    "Full controlled ablation unless enabled",
    "20/100-client scalability unless explicitly enabled and executed",
]

# Each of these five now has a real, executable implementation above (not a
# NotImplementedError stub); whether it actually ran -- and therefore which
# list it belongs in -- is read from the same EXPERIMENT_CONFIG flags that
# gated its code block, so the manifest can never claim a result that the
# flags say was skipped.
conditionally_completed = {
    "run_repeated_seeds": "Repeated seeds (5-seed FedProx/FedAvg mean +/- std)",
    "run_external_bot_iot": "BoT-IoT external validation",
    "run_gnnexplainer": "GNNExplainer (TP/TN/FP/FN feature attributions)",
    "run_formal_dp": "Formal client-level Differential Privacy (RDP accountant)",
    "run_tgat_comparison": "TGAT-style (GATConv) vs GraphSAGE comparison",
}
for flag, label in conditionally_completed.items():
    if EXPERIMENT_CONFIG.get(flag, False):
        completed_in_this_cell.append(label)
    else:
        not_claimed_as_completed.append(
            label + " (disabled: EXPERIMENT_CONFIG['" + flag + "'] is False)"
        )

manifest = {
    "completed_in_this_cell": completed_in_this_cell,
    "not_claimed_as_completed": not_claimed_as_completed,
    "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
}
with open(os.path.join(FL_RESULT_DIR, "Experiment_Manifest.json"), "w") as f:
    json.dump(manifest, f, indent=2)

print("\n" + "="*100)
print("MASTER EXPERIMENT SUITE FINISHED")
print("="*100)
print("Results:", FL_RESULT_DIR)
print("Graphs :", FL_GRAPH_DIR)
print("No unexecuted result has been represented as a completed measurement.")
```

```python



import os, time, math, json, random, copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    balanced_accuracy_score, matthews_corrcoef, roc_curve,
    precision_recall_curve
)
from sklearn.neighbors import NearestNeighbors
from torch_geometric.data import Data
from torch_geometric.nn import GATConv


BACKUP_DIR = os.path.join(FL_RESULT_DIR, "Backups")
os.makedirs(BACKUP_DIR, exist_ok=True)


EXPERIMENT_CONFIG = {
    "dataset": "CIC-IoT-2023 (training); BoT-IoT (independent testing only)",
    "dataset_path": "/content/drive/MyDrive/IDS dataset",
    "final_num_clients": 400,
    "final_algorithm": "FedProx",
    "final_partition": "mild_non_iid",
    "rounds": 20,
    "local_epochs": 30,
    "local_lr": LOCAL_LR,
    "fedprox_mu": FEDPROX_MU,
    "update_clip_norm": UPDATE_CLIP_NORM,
    "primary_noise_multiplier": 0.0,
    "seeds": [42, 43, 44, 45, 46],
    "scalability_client_counts": [100, 200, 400],
    "optional_large_client_counts": [200, 400],
    "privacy_noise_multipliers": [0.0, 0.005, 0.01, 0.02],
    "mild_noniid_dirichlet_alpha": 0.5,
    "strong_noniid_dirichlet_alpha": 0.1,
    "correlation_threshold": 0.90,
    "run_optional_large_scalability": False,
    "run_ablation": False,
    "run_repeated_seeds": True,
    "run_imputer_reconstruction": False,
    "run_external_bot_iot": True,
    "run_gnnexplainer": True,
    "run_formal_dp": True,
    "run_tgat_comparison": True,
    "dp_delta": 1e-5,
    "gnnexplainer_cases_per_bucket": 5,
    "gnnexplainer_epochs": 100,
}


print(json.dumps(EXPERIMENT_CONFIG, indent=2))


def save_csv(df, filename):
    path = os.path.join(FL_RESULT_DIR, filename)
    df.to_csv(path, index=False)
    print(f"Saved: {path}")
    return path

with open(os.path.join(FL_RESULT_DIR, "Experiment_Config.json"), "w") as f:
    json.dump(EXPERIMENT_CONFIG, f, indent=2)


def compute_binary_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()

    specificity = tn / (tn + fp) if (tn + fp) else 0.0
    far_fpr = fp / (fp + tn) if (fp + tn) else 0.0
    frr_fnr = fn / (fn + tp) if (fn + tp) else 0.0


    try:
        roc_auc = roc_auc_score(y_true, y_prob)
    except ValueError:
        roc_auc = np.nan
    try:
        pr_auc = average_precision_score(y_true, y_prob)
    except ValueError:
        pr_auc = np.nan

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Specificity": specificity,
        "FAR_FPR": far_fpr,
        "FRR_FNR": frr_fnr,
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "MCC": matthews_corrcoef(y_true, y_pred),
        "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
    }

def evaluate_model_metrics(model, graph, mask):
    model.eval()
    with torch.no_grad():
        out = model(graph.x, graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        logits = logits[mask]
        labels = graph.y[mask]
        probs = torch.softmax(logits, dim=1)[:, 1]
        preds = torch.argmax(logits, dim=1)

    y_true = labels.detach().cpu().numpy()
    y_pred = preds.detach().cpu().numpy()
    y_prob = probs.detach().cpu().numpy()

    metrics = compute_binary_metrics(y_true, y_pred, y_prob)
    metrics["Validation Loss"] = float(nn.CrossEntropyLoss()(logits, labels).detach().cpu())
    metrics["y_true"] = y_true
    metrics["y_pred"] = y_pred
    metrics["y_prob"] = y_prob
    return metrics


ARCHITECTURE_VERIFICATION = pd.DataFrame([
    ["GraphSAGE layer 1", "NovelTemporalClassifier.gnn1", True, "Active in final classifier"],
    ["GraphSAGE layer 2", "NovelTemporalClassifier.gnn2", True, "Active in final classifier"],
    ["TGAT", "No TGAT class/call found", False, "Do not claim TGAT in final architecture"],
    ["Feature attention", "NovelTemporalClassifier.attention", True, "Active before BiLSTM"],
    ["BiLSTM in classifier", "NovelTemporalClassifier.bilstm", True, "Executes with sequence length 1"],
    ["Temporal attention", "NovelTemporalClassifier.temporal_attention", True, "Executes over length-1 sequence"],
    ["BiLSTM imputer", "BiLSTMImputer", False, "Defined, but no completed training/evaluation found"],
], columns=["Component", "Code Path", "Active", "Verification"])
save_csv(ARCHITECTURE_VERIFICATION, "Architecture_Verification.csv")
display(ARCHITECTURE_VERIFICATION)


def partition_train_nodes(graph, num_clients, profile="iid", seed=42,
                          dirichlet_alpha=0.5, min_samples=1):
    """
    IID:
        class-balanced random allocation, so every client receives
        approximately the same class proportions.

    Mild/strong non-IID:
        class-conditional Dirichlet allocation. Lower alpha => stronger
        heterogeneity. The split is deterministic for a fixed seed.
    """
    rng = np.random.default_rng(seed)
    train_nodes = torch.where(graph.train_mask.detach().cpu())[0].numpy()
    y = graph.y.detach().cpu().numpy()

    if profile == "iid":

        client_lists = [[] for _ in range(num_clients)]
        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            for j, node in enumerate(cls_nodes):
                client_lists[j % num_clients].append(int(node))
    else:
        alpha = dirichlet_alpha
        client_lists = [[] for _ in range(num_clients)]

        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            proportions = rng.dirichlet(np.full(num_clients, alpha))
            counts = rng.multinomial(len(cls_nodes), proportions)

            start = 0
            for cid, count in enumerate(counts):
                client_lists[cid].extend(
                    cls_nodes[start:start+count].tolist()
                )
                start += count


        if any(len(v) < min_samples for v in client_lists):
            return partition_train_nodes(
                graph, num_clients, profile, seed + 1,
                dirichlet_alpha, min_samples
            )

    clients = []
    for cid, nodes_list in enumerate(client_lists):
        nodes_np = np.asarray(nodes_list, dtype=np.int64)
        rng.shuffle(nodes_np)
        nodes = torch.tensor(nodes_np, dtype=torch.long)

        local_edge_index, _ = subgraph(
            nodes,
            graph.edge_index.detach().cpu(),
            relabel_nodes=True,
            num_nodes=graph.num_nodes
        )

        counts = np.bincount(y[nodes_np], minlength=2)

        clients.append({
            "client_id": cid,
            "global_nodes": nodes,
            "edge_index": local_edge_index,
            "num_samples": len(nodes_np),
            "class_0": int(counts[0]),
            "class_1": int(counts[1]),
        })

    return clients

def clients_to_distribution_df(clients, profile, seed):
    rows = []
    for c in clients:
        total = max(c["num_samples"], 1)
        rows.append({
            "Seed": seed,
            "Partition": profile,
            "Client": c["client_id"],
            "Samples": c["num_samples"],
            "Benign_Count": c["class_0"],
            "Attack_Count": c["class_1"],
            "Benign_Fraction": c["class_0"] / total,
            "Attack_Fraction": c["class_1"] / total,
        })
    return pd.DataFrame(rows)

distribution_frames = []
for profile, alpha in [
    ("iid", None),
    ("mild_non_iid", EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]),
    ("strong_non_iid", EXPERIMENT_CONFIG["strong_noniid_dirichlet_alpha"]),
]:
    cset = partition_train_nodes(
        graph_data, 5, profile=profile, seed=SEED,
        dirichlet_alpha=alpha if alpha is not None else 0.5
    )
    distribution_frames.append(
        clients_to_distribution_df(cset, profile, SEED)
    )

CLIENT_DISTRIBUTION_DF = pd.concat(distribution_frames, ignore_index=True)


CURRENT_PARTITION_DIAGNOSIS = pd.DataFrame([{
    "Current_Uploaded_Partition": "random shuffled train nodes + np.array_split",
    "Class_Balance_Enforced": False,
    "Controlled_NonIID_Enforced": False,
    "Revised_Setups": "IID, mild_non_iid, strong_non_iid",
    "Conclusion": "Current setup should not be described as proper non-IID."
}])
save_csv(CURRENT_PARTITION_DIAGNOSIS, "Client_Distribution_Diagnosis.csv")
save_csv(CLIENT_DISTRIBUTION_DF, "Client_Distribution.csv")

print("\nClient-distribution verification:")
display(CLIENT_DISTRIBUTION_DF)

# ------------------------------------------------------------
# RECOMMENDATION 6 — MODEL FACTORY
# ------------------------------------------------------------
def set_all_seeds(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def make_experiment_model():
    return NovelTemporalClassifier(
        input_dim=graph_data.num_features,
        hidden_dim=128,
        embedding_dim=64,
        num_layers=2,
        num_classes=2,
        dropout=0.30
    ).to(DEVICE)

def count_model_parameters(model):
    return sum(p.numel() for p in model.parameters())

# ------------------------------------------------------------
# RECOMMENDATION 7 — FEDERATED TRAINING ENGINE
# ------------------------------------------------------------
def train_one_client_v2(global_state, client, algorithm="FedProx",
                        local_epochs=1, lr=0.001, fedprox_mu=0.01,
                        clip_norm=1.0, noise_multiplier=0.0):
    local_model = make_experiment_model()
    local_model.load_state_dict(global_state)
    local_model.train()

    x_local, y_local, edge_local = materialize_client(
        graph_data, client
    )

    counts = torch.bincount(y_local, minlength=2).float()
    safe = torch.clamp(counts, min=1.0)
    class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=lr,
        weight_decay=WEIGHT_DECAY
    )

    global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
    start = time.perf_counter()

    for _ in range(local_epochs):
        optimizer.zero_grad(set_to_none=True)
        logits = local_model(x_local, edge_local)
        if isinstance(logits, (tuple, list)):
            logits = logits[0]

        loss = criterion(logits, y_local)

        if algorithm.lower() == "fedprox" and fedprox_mu > 0:
            prox = torch.zeros((), device=DEVICE)
            for name, param in local_model.named_parameters():
                prox += torch.sum(
                    (param - global_device[name]) ** 2
                )
            loss = loss + 0.5 * fedprox_mu * prox

        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            local_model.parameters(), max_norm=5.0
        )
        optimizer.step()

    elapsed = time.perf_counter() - start
    local_state = clone_state_dict(local_model)


    if noise_multiplier > 0:
        transmitted, raw_norm, clip_scale = clip_and_noise_update(
            local_state, global_state,
            clip_norm=clip_norm,
            noise_multiplier=noise_multiplier
        )
    else:
        transmitted = local_state
        raw_norm = state_dict_l2_distance(local_state, global_state)
        clip_scale = 1.0

    with torch.no_grad():
        local_model.eval()
        pred_logits = local_model(x_local, edge_local)
        if isinstance(pred_logits, (tuple, list)):
            pred_logits = pred_logits[0]
        pred = torch.argmax(pred_logits, dim=1)

    return {
        "state": transmitted,
        "samples": client["num_samples"],
        "loss": float(loss.detach().cpu()),
        "accuracy": accuracy_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy()
        ),
        "f1": f1_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy(),
            zero_division=0
        ),
        "elapsed": elapsed,
        "raw_update_norm": raw_norm,
        "clip_scale": clip_scale,
    }

def run_federated_experiment(
    num_clients=5,
    algorithm="FedProx",
    partition_profile="mild_non_iid",
    rounds=10,
    local_epochs=1,
    seed=42,
    noise_multiplier=0.0,
    dirichlet_alpha=0.5,
    record_client_log=True
):
    set_all_seeds(seed)

    clients = partition_train_nodes(
        graph_data,
        num_clients=num_clients,
        profile=partition_profile,
        seed=seed,
        dirichlet_alpha=dirichlet_alpha
    )

    model = make_experiment_model()
    global_state = clone_state_dict(model)

    history = []
    client_log = []

    model_bytes = sum(
        t.numel() * t.element_size()
        for t in global_state.values()
    )

    total_communication_mb = 0.0

    for round_id in range(1, rounds + 1):
        round_start = time.perf_counter()

        local_states, local_weights = [], []
        local_losses, local_times = [], []
        update_norms, clip_scales = [], []

        for client in clients:
            result = train_one_client_v2(
                global_state=global_state,
                client=client,
                algorithm=algorithm,
                local_epochs=local_epochs,
                lr=LOCAL_LR,
                fedprox_mu=FEDPROX_MU,
                clip_norm=UPDATE_CLIP_NORM,
                noise_multiplier=noise_multiplier
            )

            local_states.append(result["state"])
            local_weights.append(result["samples"])
            local_losses.append(result["loss"])
            local_times.append(result["elapsed"])
            update_norms.append(result["raw_update_norm"])
            clip_scales.append(result["clip_scale"])

            if record_client_log:
                client_log.append({
                    "Seed": seed,
                    "Algorithm": algorithm,
                    "Partition": partition_profile,
                    "Round": round_id,
                    "Client": client["client_id"],
                    "Samples": client["num_samples"],
                    "Benign_Count": client["class_0"],
                    "Attack_Count": client["class_1"],
                    "Local_Loss": result["loss"],
                    "Local_Accuracy": result["accuracy"],
                    "Local_F1": result["f1"],
                    "Update_Norm": result["raw_update_norm"],
                    "Clip_Scale": result["clip_scale"],
                    "Local_Time_s": result["elapsed"]
                })

        global_state = aggregate_client_states(
            local_states, local_weights, global_state
        )
        model.load_state_dict(global_state)

        val = evaluate_model_metrics(
            model, graph_data, graph_data.val_mask
        )

        round_time = time.perf_counter() - round_start

        upload_mb = (model_bytes * len(clients)) / (1024**2)
        download_mb = (model_bytes * len(clients)) / (1024**2)
        comm_mb = upload_mb + download_mb
        total_communication_mb += comm_mb

        history.append({
            "Seed": seed,
            "Algorithm": algorithm,
            "Partition": partition_profile,
            "Noise_Multiplier": noise_multiplier,
            "Round": round_id,
            "Selected_Clients": len(clients),
            "Validation_Loss": val["Validation Loss"],
            "Validation_Accuracy": val["Accuracy"],
            "Validation_Precision": val["Precision"],
            "Validation_Recall": val["Recall"],
            "Validation_F1": val["F1"],
            "Mean_Local_Loss": float(np.mean(local_losses)),
            "Mean_Update_Norm": float(np.mean(update_norms)),
            "Mean_Clip_Scale": float(np.mean(clip_scales)),
            "Round_Runtime_s": round_time,
            "Avg_Local_Runtime_s": float(np.mean(local_times)),
            "Model_Size_MB": model_bytes / (1024**2),
            "Communication_MB_per_Round": comm_mb,
            "Cumulative_Communication_MB": total_communication_mb,
            "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        })

    test = evaluate_model_metrics(
        model, graph_data, graph_data.test_mask
    )

    best_round_idx = int(
        np.argmax([r["Validation_F1"] for r in history])
    )
    best_round = history[best_round_idx]["Round"]

    final_row = {
        "Seed": seed,
        "Clients": num_clients,
        "Algorithm": algorithm,
        "Partition": partition_profile,
        "Rounds": rounds,
        "Local_Epochs": local_epochs,
        "Noise_Multiplier": noise_multiplier,
        "Best_Validation_Round": best_round,
        "Accuracy": test["Accuracy"],
        "Precision": test["Precision"],
        "Recall": test["Recall"],
        "F1": test["F1"],
        "ROC-AUC": test["ROC-AUC"],
        "PR-AUC": test["PR-AUC"],
        "Specificity": test["Specificity"],
        "FAR_FPR": test["FAR_FPR"],
        "FRR_FNR": test["FRR_FNR"],
        "Balanced_Accuracy": test["Balanced Accuracy"],
        "MCC": test["MCC"],
        "TN": test["TN"], "FP": test["FP"],
        "FN": test["FN"], "TP": test["TP"],
        "Average_Round_Runtime_s": float(np.mean([r["Round_Runtime_s"] for r in history])),
        "Communication_MB_per_Round": float(np.mean([r["Communication_MB_per_Round"] for r in history])),
        "Total_Communication_MB": float(history[-1]["Cumulative_Communication_MB"]),
        "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        "Convergence_Round_ValF1": best_round,
    }

    return {
        "model": model,
        "clients": clients,
        "history": pd.DataFrame(history),
        "client_log": pd.DataFrame(client_log),
        "final": final_row,
        "test": test,
    }


FINAL_RUN = run_federated_experiment(
    num_clients=EXPERIMENT_CONFIG["final_num_clients"],
    algorithm="FedProx",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED,
    noise_multiplier=EXPERIMENT_CONFIG["primary_noise_multiplier"],
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"],
)

FINAL_FEDERATED_RESULTS = pd.DataFrame([FINAL_RUN["final"]])
save_csv(FINAL_FEDERATED_RESULTS, "Federated_Final_Test_Results.csv")
save_csv(FINAL_RUN["history"], "Federated_Round_History.csv")
save_csv(FINAL_RUN["client_log"], "Federated_Client_Log.csv")

# Exact confusion matrix CSV
final_cm = np.array([
    [FINAL_RUN["final"]["TN"], FINAL_RUN["final"]["FP"]],
    [FINAL_RUN["final"]["FN"], FINAL_RUN["final"]["TP"]]
], dtype=int)
cm_df = pd.DataFrame(
    final_cm,
    index=["Actual_Benign", "Actual_Attack"],
    columns=["Predicted_Benign", "Predicted_Attack"]
)
cm_df.to_csv(os.path.join(FL_RESULT_DIR, "Federated_Confusion_Matrix.csv"))

torch.save(
    {
        "model_state_dict": FINAL_RUN["model"].state_dict(),
        "config": EXPERIMENT_CONFIG,
        "final_metrics": FINAL_RUN["final"]
    },
    os.path.join(FL_RESULT_DIR, "Federated_Final_FedProx_Model.pt")
)

print("\nFINAL 5-CLIENT FEDPROX RESULT")
display(FINAL_FEDERATED_RESULTS)

# ------------------------------------------------------------
# RECOMMENDATION 9 — PUBLICATION-QUALITY FINAL FIGURES
# ------------------------------------------------------------
def plot_and_save(x, y, xlabel, ylabel, title, filename, marker="o"):
    plt.figure(figsize=(9, 6))
    plt.plot(x, y, marker=marker, linewidth=2)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    path = os.path.join(FL_GRAPH_DIR, filename)
    plt.savefig(path, dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
    return path

h = FINAL_RUN["history"]
plot_and_save(h["Round"], h["Validation_Loss"],
              "Federated Round", "Validation Loss",
              "FedProx Validation Loss",
              "Final_FedProx_Validation_Loss.png")
plot_and_save(h["Round"], h["Validation_Accuracy"] * 100,
              "Federated Round", "Validation Accuracy (%)",
              "FedProx Validation Accuracy",
              "Final_FedProx_Validation_Accuracy.png")
plot_and_save(h["Round"], h["Validation_F1"] * 100,
              "Federated Round", "Validation F1 (%)",
              "FedProx Validation F1",
              "Final_FedProx_Validation_F1.png")
plot_and_save(h["Round"], h["Cumulative_Communication_MB"],
              "Federated Round", "Cumulative Communication (MB)",
              "FedProx Cumulative Communication",
              "Final_FedProx_Cumulative_Communication.png")
plot_and_save(h["Round"], h["Round_Runtime_s"],
              "Federated Round", "Round Runtime (s)",
              "FedProx Round Runtime",
              "Final_FedProx_Round_Runtime.png")

# Final confusion matrix
plt.figure(figsize=(7, 6))
plt.imshow(final_cm, interpolation="nearest")
plt.colorbar()
plt.xticks([0, 1], ["Benign", "Attack"])
plt.yticks([0, 1], ["Benign", "Attack"])
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("FedProx Final Confusion Matrix")
for r in range(2):
    for c in range(2):
        plt.text(c, r, f"{final_cm[r,c]:,}",
                 ha="center", va="center", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_Confusion_Matrix.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ROC/PR curves
y_true = FINAL_RUN["test"]["y_true"]
y_pred = FINAL_RUN["test"]["y_pred"]
y_prob = FINAL_RUN["test"]["y_prob"]

try:
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(fpr, tpr, linewidth=2,
             label=f"FedProx ROC-AUC={FINAL_RUN['final']['ROC-AUC']:.4f}")
    plt.plot([0,1], [0,1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("FedProx ROC Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_ROC.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("ROC figure skipped:", e)

try:
    pr_precision, pr_recall, _ = precision_recall_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(pr_recall, pr_precision, linewidth=2,
             label=f"FedProx PR-AUC={FINAL_RUN['final']['PR-AUC']:.4f}")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("FedProx Precision-Recall Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_PR.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("PR figure skipped:", e)

# ------------------------------------------------------------
# RECOMMENDATION 10 — FEDAVG VS FEDPROX, SAME EVERYTHING ELSE
# ------------------------------------------------------------
# Same split, clients, model, rounds, local epochs and seed.
# Only the server/client objective differs: FedAvg mu=0 vs FedProx mu=0.01.
fedavg_run = run_federated_experiment(
    num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedAvg",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED, noise_multiplier=0.0,
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
)
fedprox_run = FINAL_RUN

fedavg_history = fedavg_run["history"].copy()
fedprox_history = fedprox_run["history"].copy()

comparison_rows = []
for _, a in fedavg_history.iterrows():
    p = fedprox_history[fedprox_history["Round"] == a["Round"]].iloc[0]
    comparison_rows.append({
        "Seed": SEED,
        "Round": int(a["Round"]),
        "FedAvg_Validation_Accuracy": a["Validation_Accuracy"],
        "FedProx_Validation_Accuracy": p["Validation_Accuracy"],
        "FedAvg_Validation_F1": a["Validation_F1"],
        "FedProx_Validation_F1": p["Validation_F1"],
        "FedAvg_Validation_Loss": a["Validation_Loss"],
        "FedProx_Validation_Loss": p["Validation_Loss"],
        "FedAvg_Update_Norm": a["Mean_Update_Norm"],
        "FedProx_Update_Norm": p["Mean_Update_Norm"],
        "FedAvg_Round_Runtime_s": a["Round_Runtime_s"],
        "FedProx_Round_Runtime_s": p["Round_Runtime_s"],
    })

comparison_rows.append({
    "Seed": SEED, "Round": "FINAL",
    "FedAvg_Accuracy": fedavg_run["final"]["Accuracy"],
    "FedProx_Accuracy": fedprox_run["final"]["Accuracy"],
    "FedAvg_Precision": fedavg_run["final"]["Precision"],
    "FedProx_Precision": fedprox_run["final"]["Precision"],
    "FedAvg_Recall": fedavg_run["final"]["Recall"],
    "FedProx_Recall": fedprox_run["final"]["Recall"],
    "FedAvg_F1": fedavg_run["final"]["F1"],
    "FedProx_F1": fedprox_run["final"]["F1"],
    "FedAvg_ROC-AUC": fedavg_run["final"]["ROC-AUC"],
    "FedProx_ROC-AUC": fedprox_run["final"]["ROC-AUC"],
    "FedAvg_Best_Validation_Round": fedavg_run["final"]["Best_Validation_Round"],
    "FedProx_Best_Validation_Round": fedprox_run["final"]["Best_Validation_Round"],
})

FEDAVG_VS_FEDPROX = pd.DataFrame(comparison_rows)
save_csv(FEDAVG_VS_FEDPROX, "FedAvg_vs_FedProx.csv")

plot_and_save(
    fedavg_history["Round"],
    fedavg_history["Validation_F1"] * 100,
    "Round", "Validation F1 (%)",
    "FedAvg vs FedProx — FedAvg",
    "FedAvg_vs_FedProx_FedAvg_F1.png"
)
plt.figure(figsize=(9,6))
plt.plot(fedavg_history["Round"], fedavg_history["Validation_F1"]*100,
         marker="o", linewidth=2, label="FedAvg")
plt.plot(fedprox_history["Round"], fedprox_history["Validation_F1"]*100,
         marker="s", linewidth=2, label="FedProx")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("FedAvg vs FedProx — Convergence")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "FedAvg_vs_FedProx_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ------------------------------------------------------------
# RECOMMENDATION — TGAT-STYLE BASELINE COMPARISON
# ------------------------------------------------------------
# HONESTY NOTE: graph_data (STEP 26) has no per-edge timestamps, so a
# literal TGAT (Xu et al., 2020 — temporal-attention over timestamped
# edges) cannot be trained on it. This substitutes the uploaded model's
# spatial GraphSAGE encoder with a multi-head GATConv encoder (graph
# attention, the mechanism TGAT also builds on) while keeping every other
# component -- input dimension, BiLSTM, temporal attention, fusion head,
# classifier, optimizer, seed, split, client partition and federated
# protocol -- byte-for-byte identical to NovelTemporalClassifier / the
# 5-client FedProx run above, so the comparison isolates the effect of
# attention-based neighborhood aggregation vs. GraphSAGE's mean
# aggregation. Results are reported as "TGAT-style (GAT)", never as a
# reproduction of the published TGAT architecture.
if EXPERIMENT_CONFIG["run_tgat_comparison"]:
    print("\n=== TGAT-STYLE (GAT) VS GRAPHSAGE COMPARISON ===")

    class TGATStyleClassifier(nn.Module):
        def __init__(self, input_dim, hidden_dim=128, embedding_dim=64,
                     num_layers=2, num_classes=2, dropout=0.30, heads=4):
            super().__init__()
            self.gat1 = GATConv(
                input_dim, hidden_dim // heads, heads=heads, dropout=dropout
            )
            self.gat2 = GATConv(
                hidden_dim, embedding_dim, heads=1, concat=False, dropout=dropout
            )
            self.norm1 = nn.LayerNorm(hidden_dim)
            self.norm2 = nn.LayerNorm(embedding_dim)
            self.attention = nn.Sequential(
                nn.Linear(embedding_dim, embedding_dim // 2), nn.Tanh(),
                nn.Linear(embedding_dim // 2, 1)
            )
            self.bilstm = nn.LSTM(
                input_size=embedding_dim, hidden_size=hidden_dim,
                num_layers=num_layers, batch_first=True, bidirectional=True,
                dropout=dropout if num_layers > 1 else 0
            )
            self.temporal_attention = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.Tanh(),
                nn.Linear(hidden_dim, 1)
            )
            self.gnn_projection = nn.Linear(embedding_dim, hidden_dim * 2)
            self.fusion = nn.Sequential(
                nn.Linear(hidden_dim * 4, hidden_dim * 2),
                nn.ReLU(), nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.ReLU(),
                nn.Dropout(dropout), nn.Linear(hidden_dim, num_classes)
            )
            self.dropout = nn.Dropout(dropout)

        def forward(self, x, edge_index=None):
            x1 = self.gat1(x, edge_index)
            x1 = F.relu(self.norm1(x1))
            x1 = self.dropout(x1)

            x2 = self.gat2(x1, edge_index)
            x2 = F.relu(self.norm2(x2))

            attn_scores = self.attention(x2)
            attn_weights = torch.softmax(attn_scores, dim=0)
            attended = x2 * attn_weights

            sequence = attended.unsqueeze(1)
            lstm_output, _ = self.bilstm(sequence)

            t_scores = self.temporal_attention(lstm_output)
            t_weights = torch.softmax(t_scores, dim=1)
            temporal_context = (lstm_output * t_weights).sum(dim=1)

            gnn_residual = self.gnn_projection(x2)
            fused = self.fusion(torch.cat([temporal_context, gnn_residual], dim=1))
            return self.classifier(fused)

    def make_tgat_model():
        return TGATStyleClassifier(
            input_dim=graph_data.num_features,
            hidden_dim=128, embedding_dim=64,
            num_layers=2, num_classes=2, dropout=0.30
        ).to(DEVICE)

    def train_one_client_generic(model_factory, global_state, client,
                                  algorithm, local_epochs, lr, fedprox_mu):
        # Mirrors train_one_client_v2 exactly, parametrized by model_factory
        # so the GraphSAGE and TGAT-style runs are trained identically in
        # every other respect (loss, class weighting, optimizer, clipping).
        local_model = model_factory()
        local_model.load_state_dict(global_state)
        local_model.train()

        x_local, y_local, edge_local = materialize_client(graph_data, client)

        counts = torch.bincount(y_local, minlength=2).float()
        safe = torch.clamp(counts, min=1.0)
        class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
        criterion = nn.CrossEntropyLoss(weight=class_weights)

        optimizer = torch.optim.AdamW(
            local_model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY
        )
        global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
        start = time.perf_counter()

        for _ in range(local_epochs):
            optimizer.zero_grad(set_to_none=True)
            logits = local_model(x_local, edge_local)
            if isinstance(logits, (tuple, list)):
                logits = logits[0]
            loss = criterion(logits, y_local)

            if algorithm.lower() == "fedprox" and fedprox_mu > 0:
                prox = torch.zeros((), device=DEVICE)
                for name, param in local_model.named_parameters():
                    prox += torch.sum((param - global_device[name]) ** 2)
                loss = loss + 0.5 * fedprox_mu * prox

            loss.backward()
            torch.nn.utils.clip_grad_norm_(local_model.parameters(), max_norm=5.0)
            optimizer.step()

        elapsed = time.perf_counter() - start
        local_state = clone_state_dict(local_model)

        with torch.no_grad():
            local_model.eval()
            pred = torch.argmax(local_model(x_local, edge_local), dim=1)

        return {
            "state": local_state,
            "samples": client["num_samples"],
            "loss": float(loss.detach().cpu()),
            "elapsed": elapsed,
            "accuracy": accuracy_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy()
            ),
            "f1": f1_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy(),
                zero_division=0
            ),
        }

    def run_federated_experiment_generic(model_factory, num_clients, algorithm,
                                          partition_profile, rounds, local_epochs,
                                          seed):
        set_all_seeds(seed)
        clients = partition_train_nodes(
            graph_data, num_clients=num_clients, profile=partition_profile,
            seed=seed, dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        model = model_factory()
        global_state = clone_state_dict(model)
        history = []

        for round_id in range(1, rounds + 1):
            states, weights = [], []
            for client in clients:
                result = train_one_client_generic(
                    model_factory, global_state, client,
                    algorithm=algorithm, local_epochs=local_epochs,
                    lr=LOCAL_LR, fedprox_mu=FEDPROX_MU
                )
                states.append(result["state"])
                weights.append(result["samples"])

            global_state = aggregate_client_states(states, weights, global_state)
            model.load_state_dict(global_state)

            val = evaluate_model_metrics(model, graph_data, graph_data.val_mask)
            history.append({
                "Round": round_id,
                "Validation_Accuracy": val["Accuracy"],
                "Validation_F1": val["F1"],
            })

        test = evaluate_model_metrics(model, graph_data, graph_data.test_mask)
        return {"model": model, "history": pd.DataFrame(history), "test": test}

    # Same 5-client, same rounds/local-epochs/seed/partition as the primary
    # FedProx run above, so the GraphSAGE row is a like-for-like comparison
    # rather than a re-use of the 50-client FINAL_RUN numbers.
    graphsage_5client_run = run_federated_experiment(
        num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED, noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )

    tgat_run = run_federated_experiment_generic(
        make_tgat_model, num_clients=EXPERIMENT_CONFIG["final_num_clients"], algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED
    )

    TGAT_COMPARISON = pd.DataFrame([
        {
            "Model": "GraphSAGE (uploaded final classifier)",
            "Clients": 5,
            "Accuracy": graphsage_5client_run["test"]["Accuracy"],
            "F1": graphsage_5client_run["test"]["F1"],
            "ROC-AUC": graphsage_5client_run["test"]["ROC-AUC"],
            "MCC": graphsage_5client_run["test"]["MCC"],
        },
        {
            "Model": "TGAT-style (GATConv, no edge timestamps)",
            "Clients": 5,
            "Accuracy": tgat_run["test"]["Accuracy"],
            "F1": tgat_run["test"]["F1"],
            "ROC-AUC": tgat_run["test"]["ROC-AUC"],
            "MCC": tgat_run["test"]["MCC"],
        },
    ])
    save_csv(TGAT_COMPARISON, "TGAT_Comparison.csv")
    display(TGAT_COMPARISON)

    plt.figure(figsize=(9, 6))
    plt.plot(graphsage_5client_run["history"]["Round"],
              graphsage_5client_run["history"]["Validation_F1"] * 100,
              marker="o", linewidth=2, label="GraphSAGE")
    plt.plot(tgat_run["history"]["Round"],
              tgat_run["history"]["Validation_F1"] * 100,
              marker="s", linewidth=2, label="TGAT-style (GAT)")
    plt.xlabel("Federated Round")
    plt.ylabel("Validation F1 (%)")
    plt.title("GraphSAGE vs TGAT-style Encoder — Convergence (5 clients)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "TGAT_vs_GraphSAGE_Convergence.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "TGAT comparison not executed"
    }]), "TGAT_Comparison.csv")

# ------------------------------------------------------------
# RECOMMENDATION 11 — SCALABILITY: ACTUALLY RUN ONLY EVALUATED COUNTS
# ------------------------------------------------------------
scalability_counts = list(EXPERIMENT_CONFIG["scalability_client_counts"])

if EXPERIMENT_CONFIG["run_optional_large_scalability"]:
    scalability_counts += EXPERIMENT_CONFIG["optional_large_client_counts"]

SCALABILITY_ROWS = []
SCALABILITY_HISTORIES = []

for n_clients in scalability_counts:
    print(f"\n=== SCALABILITY: {n_clients} CLIENTS ===")

    run = run_federated_experiment(
        num_clients=n_clients,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    SCALABILITY_HISTORIES.append(run["history"])

    f = run["final"]
    SCALABILITY_ROWS.append({
        "Clients": n_clients,
        "Accuracy": f["Accuracy"],
        "F1": f["F1"],
        "Average_Round_Runtime_s": f["Average_Round_Runtime_s"],
        "Communication_MB_per_Round": f["Communication_MB_per_Round"],
        "Total_Communication_MB": f["Total_Communication_MB"],
        "Samples_Per_Client_Mean": f["Samples_Per_Client_Mean"],
        "Convergence_Round": f["Convergence_Round_ValF1"],
        "Rounds_Evaluated": EXPERIMENT_CONFIG["rounds"],
        "Status": "COMPLETED"
    })

SCALABILITY_RESULTS = pd.DataFrame(SCALABILITY_ROWS)
save_csv(SCALABILITY_RESULTS, "Scalability_Experiment.csv")

# Scalability figures
for metric, ylabel, fname in [
    ("Average_Round_Runtime_s", "Average Round Runtime (s)", "Scalability_Runtime.png"),
    ("Communication_MB_per_Round", "Communication (MB/round)", "Scalability_Communication.png"),
    ("Accuracy", "Final Test Accuracy", "Scalability_Accuracy.png"),
    ("F1", "Final Test F1", "Scalability_F1.png"),
]:
    plt.figure(figsize=(9,6))
    plt.plot(SCALABILITY_RESULTS["Clients"], SCALABILITY_RESULTS[metric],
             marker="o", linewidth=2)
    plt.xlabel("Number of Clients")
    plt.ylabel(ylabel)
    plt.title(f"Scalability — {ylabel}")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, fname), dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()


plt.figure(figsize=(10,6))
for hist in SCALABILITY_HISTORIES:
    n = int(hist["Selected_Clients"].iloc[0])
    plt.plot(hist["Round"], hist["Validation_F1"]*100,
             marker="o", linewidth=2, label=f"{n} clients")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("Scalability — Convergence Behavior")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Scalability_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


privacy_rows = []
for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
    print(f"\n=== GAUSSIAN UPDATE PERTURBATION: noise={noise} ===")

    run = run_federated_experiment(
        num_clients=5,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=noise,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    f = run["final"]
    privacy_rows.append({
        "Noise_Multiplier": noise,
        "Accuracy": f["Accuracy"],
        "Precision": f["Precision"],
        "Recall": f["Recall"],
        "F1": f["F1"],
        "Status": "COMPLETED",
        "Interpretation": "Privacy-aware Gaussian update perturbation; NOT formal DP"
    })

PRIVACY_RESULTS = pd.DataFrame(privacy_rows)
save_csv(PRIVACY_RESULTS, "Privacy_Sensitivity_Experiment.csv")

plt.figure(figsize=(10,6))
for col, label in [
    ("Accuracy", "Accuracy"),
    ("Precision", "Precision"),
    ("Recall", "Recall"),
    ("F1", "F1"),
]:
    plt.plot(PRIVACY_RESULTS["Noise_Multiplier"],
             PRIVACY_RESULTS[col]*100,
             marker="o", linewidth=2, label=label)
plt.xlabel("Gaussian Update-Noise Multiplier")
plt.ylabel("Test Score (%)")
plt.title("Privacy-Aware Gaussian Update Perturbation Sensitivity")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Privacy_Sensitivity.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


if EXPERIMENT_CONFIG["run_ablation"]:
    print("Ablation execution requires explicit variant-specific model factories.")
    print("No ablation values are generated by default.")
else:
    ablation_status = pd.DataFrame([
        ["Full Model", "NOT_RUN", "Enable run_ablation and variant factories"],
        ["Without BiLSTM / reconstruction", "NOT_RUN", "Requires controlled model variant"],
        ["Without Correlation Filtering", "NOT_RUN", "Current final input pipeline does not apply filtering"],
        ["Without Attention", "NOT_RUN", "Requires controlled model variant"],
        ["FedAvg instead of FedProx", "COMPLETED_SEPARATELY", "See FedAvg_vs_FedProx.csv"],
        ["GraphSAGE only", "NOT_RUN", "Requires controlled model variant"],
        ["GraphSAGE + BiLSTM", "NOT_RUN", "Requires controlled model variant"],
    ], columns=["Variant", "Status", "Reason"])
    save_csv(ablation_status, "Ablation_Study.csv")
    display(ablation_status)


if EXPERIMENT_CONFIG["run_imputer_reconstruction"]:
    raise NotImplementedError(
        "Enable only after adding a complete train/validation/test reconstruction "
        "protocol for BiLSTMImputer. Do not report the untrained imputer as completed."
    )
else:
    reconstruction_status = pd.DataFrame([{
        "Missing_Rate": globals().get("MISSING_RATE", 0.05),
        "Sequence_Length": globals().get("SEQUENCE_LENGTH", 10),
        "RMSE": np.nan,
        "MAE": np.nan,
        "Status": "NOT_COMPLETED",
        "Reason": "BiLSTMImputer is defined but no completed training/evaluation was found"
    }])
    save_csv(reconstruction_status, "BiLSTM_Reconstruction_Status.csv")


graphfedai_available = any(
    name.lower() == "graphfedai" or "graphfedai" in name.lower()
    for name in globals().keys()
)

if graphfedai_available:
    print("GraphFedAI symbol found in runtime; implement the exact baseline adapter before running.")
else:
    baseline_status = pd.DataFrame([{
        "Baseline": "GraphFedAI",
        "Status": "NOT_COMPLETED",
        "Reason": "No GraphFedAI implementation is present in the uploaded notebook",
        "Required_Settings": json.dumps({
            "dataset_split": "same as EFG-IDS",
            "clients": 5,
            "rounds": EXPERIMENT_CONFIG["rounds"],
            "local_epochs": EXPERIMENT_CONFIG["local_epochs"],
            "seed": SEED
        })
    }])
    save_csv(baseline_status, "GraphFedAI_Baseline_Comparison.csv")
    display(baseline_status)


if EXPERIMENT_CONFIG["run_repeated_seeds"]:
    repeated_rows = []
    for seed in EXPERIMENT_CONFIG["seeds"]:
        # Final FedProx
        rp = run_federated_experiment(
            EXPERIMENT_CONFIG["final_num_clients"], "FedProx", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        # FedAvg
        ra = run_federated_experiment(
            EXPERIMENT_CONFIG["final_num_clients"], "FedAvg", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        repeated_rows.extend([
            {"Seed": seed, "Experiment": "FedProx", **rp["final"]},
            {"Seed": seed, "Experiment": "FedAvg", **ra["final"]},
        ])

    repeated_raw = pd.DataFrame(repeated_rows)
    save_csv(repeated_raw, "Repeated_Run_Raw.csv")

    summary = (
        repeated_raw
        .groupby("Experiment")
        .agg(
            Accuracy_mean=("Accuracy", "mean"),
            Accuracy_std=("Accuracy", "std"),
            Precision_mean=("Precision", "mean"),
            Precision_std=("Precision", "std"),
            Recall_mean=("Recall", "mean"),
            Recall_std=("Recall", "std"),
            F1_mean=("F1", "mean"),
            F1_std=("F1", "std"),
            ROCAUC_mean=("ROC-AUC", "mean"),
            ROCAUC_std=("ROC-AUC", "std"),
        )
        .reset_index()
    )
    save_csv(summary, "Repeated_Run_Summary.csv")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Seeds": "42,43,44,45,46",
        "Reason": "Repeated runs disabled by default because of computational cost"
    }]), "Repeated_Run_Summary.csv")

if EXPERIMENT_CONFIG["run_external_bot_iot"]:
    print("\n=== EXTERNAL VALIDATION: BoT-IoT / IDS dataset ===")

    bot_iot_files = sorted(IDS_PATH.glob("*.csv"))
    if not bot_iot_files:
        raise FileNotFoundError(
            f"No CSV files found under {IDS_PATH}; cannot run BoT-IoT "
            "external validation."
        )

    bot_frames = []
    for f in bot_iot_files:
        try:
            bot_frames.append(pd.read_csv(f, low_memory=False))
        except Exception as e:
            print(f"  Skipping {f.name}: {e}")

    BOT_RAW = pd.concat(bot_frames, ignore_index=True)

    bot_label_candidates = [
        c for c in BOT_RAW.columns
        if c.lower() in ("label", "attack", "category", "class")
    ]
    if not bot_label_candidates:
        raise KeyError("Could not locate a label column in the BoT-IoT/IDS files.")
    bot_label_col = bot_label_candidates[0]

    # Reuse the exact binary-label convention from CELL 12/13
    # (create_binary_labels) so "benign" is mapped to 0 identically to the
    # CIC-IoT-2023 training labels.
    BOT_RAW["binary_label"] = create_binary_labels(BOT_RAW[bot_label_col])

    # Feature alignment onto the trained CIC-IoT-2023 feature space:
    # - Columns the scaler was fit on and that BoT-IoT also has are used
    #   directly (matched_features).
    # - Columns the scaler was fit on but BoT-IoT does NOT have are filled
    #   with the TRAINING median (train_median from STEP 22) -- never a
    #   BoT-IoT statistic -- so no target-domain information leaks into a
    #   feature the model was never actually shown from this domain.
    aligned = pd.DataFrame(index=BOT_RAW.index)
    matched_features, filled_features = [], []
    for col in FEATURE_COLUMNS:
        if col in BOT_RAW.columns:
            aligned[col] = pd.to_numeric(BOT_RAW[col], errors="coerce")
            matched_features.append(col)
        else:
            aligned[col] = train_median[col]
            filled_features.append(col)

    aligned = aligned.replace([np.inf, -np.inf], np.nan)
    aligned = aligned.fillna(train_median)

    BOT_X = scaler.transform(aligned[FEATURE_COLUMNS])
    BOT_y = BOT_RAW["binary_label"].astype(int).to_numpy()

    print(f"BoT-IoT rows: {len(BOT_X)}")
    print(f"Matched CIC-IoT-2023 features: {len(matched_features)} / {len(FEATURE_COLUMNS)}")
    print(f"Features filled with training median (not present in BoT-IoT): {len(filled_features)}")
    if filled_features:
        print("  ", filled_features)

    # Build a self-contained kNN graph over the BoT-IoT rows only (same k
    # and metric as STEP 26), so the trained GraphSAGE layers receive the
    # same kind of local-neighborhood structure they were trained on. This
    # graph is independent of graph_data -- no BoT-IoT node is ever
    # connected to a CIC-IoT-2023 node.
    bot_knn = NearestNeighbors(n_neighbors=K_NEIGHBORS + 1, metric="euclidean", n_jobs=-1)
    bot_knn.fit(BOT_X)
    _, bot_neighbors = bot_knn.kneighbors(BOT_X)

    bot_src, bot_dst = [], []
    for i in range(len(BOT_X)):
        for j in bot_neighbors[i, 1:]:
            bot_src.append(i)
            bot_dst.append(j)
    bot_edge_index = torch.tensor([bot_src, bot_dst], dtype=torch.long)

    bot_graph = Data(
        x=torch.tensor(BOT_X, dtype=torch.float32),
        y=torch.tensor(BOT_y, dtype=torch.long),
        edge_index=bot_edge_index
    ).to(DEVICE)

    external_model = FINAL_RUN["model"]
    external_model.eval()
    with torch.no_grad():
        out = external_model(bot_graph.x, bot_graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        bot_probs = torch.softmax(logits, dim=1)[:, 1]
        bot_preds = torch.argmax(logits, dim=1)

    bot_metrics = compute_binary_metrics(
        bot_graph.y.detach().cpu().numpy(),
        bot_preds.detach().cpu().numpy(),
        bot_probs.detach().cpu().numpy()
    )

    EXTERNAL_BOTIOT_RESULTS = pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "COMPLETED",
        "Rows_Evaluated": len(BOT_X),
        "Matched_Features": len(matched_features),
        "Filled_Features": len(filled_features),
        "K_Neighbors": K_NEIGHBORS,
        "Model_Source": "FINAL_RUN (100-client FedProx, trained on CIC-IoT-2023 only)",
        "Accuracy": bot_metrics["Accuracy"],
        "Precision": bot_metrics["Precision"],
        "Recall": bot_metrics["Recall"],
        "F1": bot_metrics["F1"],
        "ROC-AUC": bot_metrics["ROC-AUC"],
        "PR-AUC": bot_metrics["PR-AUC"],
        "Specificity": bot_metrics["Specificity"],
        "Balanced_Accuracy": bot_metrics["Balanced Accuracy"],
        "MCC": bot_metrics["MCC"],
        "TN": bot_metrics["TN"], "FP": bot_metrics["FP"],
        "FN": bot_metrics["FN"], "TP": bot_metrics["TP"],
    }])
    save_csv(EXTERNAL_BOTIOT_RESULTS, "External_Validation_BoTIoT.csv")
    display(EXTERNAL_BOTIOT_RESULTS)

    bot_cm = np.array([
        [bot_metrics["TN"], bot_metrics["FP"]],
        [bot_metrics["FN"], bot_metrics["TP"]]
    ])
    plt.figure(figsize=(7, 6))
    plt.imshow(bot_cm, interpolation="nearest")
    plt.colorbar()
    plt.xticks([0, 1], ["Benign", "Attack"])
    plt.yticks([0, 1], ["Benign", "Attack"])
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.title("BoT-IoT External Validation — Confusion Matrix")
    for r in range(2):
        for c in range(2):
            plt.text(c, r, f"{bot_cm[r, c]:,}", ha="center", va="center", fontsize=14)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "External_BoTIoT_Confusion_Matrix.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "NOT_COMPLETED",
        "Reason": "External feature mapping/evaluation not executed"
    }]), "External_Validation_BoTIoT.csv")


if EXPERIMENT_CONFIG["run_gnnexplainer"]:
    print("\n=== GNNEXPLAINER: TP / TN / FP / FN CASE EXPLANATIONS ===")
    from torch_geometric.explain import Explainer, GNNExplainer

    explain_model = FINAL_RUN["model"]
    explain_model.eval()

    test_idx = torch.where(graph_data.test_mask)[0]
    y_true_test = graph_data.y[test_idx].detach().cpu().numpy()
    y_pred_test = FINAL_RUN["test"]["y_pred"]  # same order as test_idx (mask indexing)

    tp_idx = test_idx[(y_true_test == 1) & (y_pred_test == 1)]
    tn_idx = test_idx[(y_true_test == 0) & (y_pred_test == 0)]
    fp_idx = test_idx[(y_true_test == 0) & (y_pred_test == 1)]
    fn_idx = test_idx[(y_true_test == 1) & (y_pred_test == 0)]

    explain_rng = np.random.default_rng(SEED)

    def sample_case_nodes(pool, n):
        pool = pool.detach().cpu().numpy()
        if len(pool) == 0:
            return np.array([], dtype=np.int64)
        return explain_rng.choice(pool, size=min(n, len(pool)), replace=False)

    n_per_bucket = EXPERIMENT_CONFIG["gnnexplainer_cases_per_bucket"]
    case_nodes = {
        "TP": sample_case_nodes(tp_idx, n_per_bucket),
        "TN": sample_case_nodes(tn_idx, n_per_bucket),
        "FP": sample_case_nodes(fp_idx, n_per_bucket),
        "FN": sample_case_nodes(fn_idx, n_per_bucket),
    }
    for case, nodes in case_nodes.items():
        print(f"  {case}: {len(nodes)} case(s) sampled (pool size available)")

    # NovelTemporalClassifier.forward(x, edge_index) returns raw per-node
    # logits directly (not a tuple), matching Explainer's expected model
    # output for node-level multiclass classification with return_type="raw".
    explainer = Explainer(
        model=explain_model,
        algorithm=GNNExplainer(epochs=EXPERIMENT_CONFIG["gnnexplainer_epochs"]),
        explanation_type="model",
        node_mask_type="attributes",
        edge_mask_type="object",
        model_config=dict(
            mode="multiclass_classification",
            task_level="node",
            return_type="raw",
        ),
    )

    explain_x = graph_data.x.to(DEVICE)
    explain_edge_index = graph_data.edge_index.to(DEVICE)

    explain_rows = []
    for case, nodes in case_nodes.items():
        for node_id in nodes:
            node_id = int(node_id)
            explanation = explainer(
                x=explain_x, edge_index=explain_edge_index, index=node_id
            )
            feat_importance = explanation.node_mask[node_id].detach().cpu().numpy()
            top_k = np.argsort(-feat_importance)[:10]
            for rank, feat_idx in enumerate(top_k, start=1):
                feat_name = (
                    FEATURE_COLUMNS[feat_idx]
                    if feat_idx < len(FEATURE_COLUMNS)
                    else f"feature_{feat_idx}"
                )
                explain_rows.append({
                    "Case": case,
                    "Node": node_id,
                    "Rank": rank,
                    "Feature": feat_name,
                    "Importance": float(feat_importance[feat_idx]),
                })

    GNNEXPLAINER_RESULTS = pd.DataFrame(explain_rows)
    save_csv(GNNEXPLAINER_RESULTS, "GNNExplainer_Results.csv")
    display(GNNEXPLAINER_RESULTS.head(20))

    if len(GNNEXPLAINER_RESULTS):
        top_features = (
            GNNEXPLAINER_RESULTS.groupby("Feature")["Importance"]
            .mean().sort_values(ascending=False).head(15)
        )
        plt.figure(figsize=(10, 6))
        top_features.plot(kind="barh")
        plt.gca().invert_yaxis()
        plt.xlabel("Mean GNNExplainer Feature Importance")
        plt.title("Top Features — GNNExplainer (TP/TN/FP/FN cases)")
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "GNNExplainer_Top_Features.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
    else:
        print("No test-set cases were available in one or more of TP/TN/FP/FN; "
              "GNNExplainer_Results.csv was written empty.")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Cases": "TP,TN,FP,FN",
        "Reason": "GNNExplainer not executed"
    }]), "GNNExplainer_Results.csv")

# ------------------------------------------------------------
# RECOMMENDATION 19 — FORMAL DIFFERENTIAL PRIVACY
# ------------------------------------------------------------
if EXPERIMENT_CONFIG["run_formal_dp"]:
    print("\n=== FORMAL DIFFERENTIAL PRIVACY ACCOUNTING ===")
    print("Mechanism being accounted: clip_and_noise_update (FL-3) clips each")
    print("client's per-round update to L2 norm <= update_clip_norm and adds")
    print("N(0, (noise_multiplier * update_clip_norm)^2) noise coordinatewise")
    print("before it leaves the client. This is a Gaussian mechanism with L2")
    print("sensitivity = update_clip_norm and noise multiplier = noise_multiplier,")
    print("applied once per client per federated round under full participation")
    print("(client sampling probability q = 1, as configured in FL-1/FL-2).")
    print("The epsilon below is therefore a CLIENT-LEVEL DP guarantee on the")
    print("transmitted update sequence, accounted analytically via Renyi DP")
    print("(RDP) composition -- it is not a sample-level guarantee over the")
    print("raw per-record training data, since no per-record subsampling or")
    print("per-record gradient clipping is performed anywhere in this pipeline.")

    def gaussian_mechanism_rdp(alpha, noise_multiplier):
        # Renyi divergence of order alpha between N(0, sigma^2) and
        # N(1, sigma^2) for a mechanism with L2 sensitivity 1 and Gaussian
        # noise of std = noise_multiplier (Mironov, 2017, Prop. 7):
        #   RDP(alpha) = alpha / (2 * sigma^2)
        return alpha / (2.0 * (noise_multiplier ** 2))

    def rdp_to_eps(rdp_orders, rdp_values, delta):
        # Standard RDP -> (epsilon, delta)-DP conversion (Balle et al. 2020 /
        # Mironov 2017), minimized over the candidate Renyi orders.
        candidates = [
            rdp + math.log(1.0 / delta) / (alpha - 1.0)
            for alpha, rdp in zip(rdp_orders, rdp_values)
            if alpha > 1.0
        ]
        return min(candidates) if candidates else float("inf")

    RDP_ORDERS = [1 + x / 10.0 for x in range(1, 100)] + list(range(12, 64))
    DP_DELTA = EXPERIMENT_CONFIG["dp_delta"]
    # One Gaussian mechanism application per client per round; composition
    # count uses the primary experiment's round budget.
    NUM_COMPOSITIONS = EXPERIMENT_CONFIG["rounds"]

    dp_rows = []
    for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
        if noise <= 0:
            dp_rows.append({
                "Noise_Multiplier": noise,
                "Rounds_Composed": NUM_COMPOSITIONS,
                "Delta": DP_DELTA,
                "Epsilon": float("inf"),
                "Note": "No noise added: the mechanism provides no DP guarantee "
                        "(epsilon is undefined/infinite).",
            })
            continue

        rdp_values = [
            NUM_COMPOSITIONS * gaussian_mechanism_rdp(alpha, noise)
            for alpha in RDP_ORDERS
        ]
        epsilon = rdp_to_eps(RDP_ORDERS, rdp_values, DP_DELTA)

        dp_rows.append({
            "Noise_Multiplier": noise,
            "Rounds_Composed": NUM_COMPOSITIONS,
            "Delta": DP_DELTA,
            "Epsilon": epsilon,
            "Note": "Client-level (epsilon, delta)-DP for the transmitted update "
                    "via RDP composition of the Gaussian mechanism (analytic "
                    "moments accountant; sensitivity = update_clip_norm; q = 1).",
        })

    FORMAL_DP_RESULTS = pd.DataFrame(dp_rows)
    save_csv(FORMAL_DP_RESULTS, "Formal_Differential_Privacy.csv")
    display(FORMAL_DP_RESULTS)

    finite_dp = FORMAL_DP_RESULTS[np.isfinite(FORMAL_DP_RESULTS["Epsilon"])]
    if len(finite_dp):
        plt.figure(figsize=(9, 6))
        plt.plot(finite_dp["Noise_Multiplier"], finite_dp["Epsilon"],
                  marker="o", linewidth=2)
        plt.xlabel("Gaussian Noise Multiplier")
        plt.ylabel(f"Epsilon (delta={DP_DELTA:g})")
        plt.title(f"Formal Client-Level DP: Epsilon vs Noise Multiplier "
                  f"over {NUM_COMPOSITIONS} Rounds")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "Formal_DP_Epsilon.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "Formal DP accounting not executed"
    }]), "Formal_Differential_Privacy.csv")

# ------------------------------------------------------------
# RECOMMENDATION 20 — EXPERIMENT MANIFEST / LOG EVIDENCE
# ------------------------------------------------------------
completed_in_this_cell = [
    "400-client FedProx final run (single seed 42 unless repeated-seed experiment is enabled)",
    "IID/mild-non-IID/strong-non-IID client distribution verification",
    "FedAvg vs FedProx controlled comparison",
    "100/200/400-client scalability",
    "0/0.005/0.01/0.02 Gaussian update perturbation sensitivity",
]
not_claimed_as_completed = [
    "GraphFedAI rerun (implementation absent)",
    "BiLSTM reconstruction RMSE/MAE",
    "Full controlled ablation unless enabled",
    "20/100-client scalability unless explicitly enabled and executed",
]

# Each of these five now has a real, executable implementation above (not a
# NotImplementedError stub); whether it actually ran -- and therefore which
# list it belongs in -- is read from the same EXPERIMENT_CONFIG flags that
# gated its code block, so the manifest can never claim a result that the
# flags say was skipped.
conditionally_completed = {
    "run_repeated_seeds": "Repeated seeds (5-seed FedProx/FedAvg mean +/- std)",
    "run_external_bot_iot": "BoT-IoT external validation",
    "run_gnnexplainer": "GNNExplainer (TP/TN/FP/FN feature attributions)",
    "run_formal_dp": "Formal client-level Differential Privacy (RDP accountant)",
    "run_tgat_comparison": "TGAT-style (GATConv) vs GraphSAGE comparison",
}
for flag, label in conditionally_completed.items():
    if EXPERIMENT_CONFIG.get(flag, False):
        completed_in_this_cell.append(label)
    else:
        not_claimed_as_completed.append(
            label + " (disabled: EXPERIMENT_CONFIG['" + flag + "'] is False)"
        )

manifest = {
    "completed_in_this_cell": completed_in_this_cell,
    "not_claimed_as_completed": not_claimed_as_completed,
    "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
}
with open(os.path.join(FL_RESULT_DIR, "Experiment_Manifest.json"), "w") as f:
    json.dump(manifest, f, indent=2)

print("\n" + "="*100)
print("MASTER EXPERIMENT SUITE FINISHED")
print("="*100)
print("Results:", FL_RESULT_DIR)
print("Graphs :", FL_GRAPH_DIR)
print("No unexecuted result has been represented as a completed measurement.")
```

```python



import os, time, math, json, random, copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, confusion_matrix,
    balanced_accuracy_score, matthews_corrcoef, roc_curve,
    precision_recall_curve
)
from sklearn.neighbors import NearestNeighbors
from torch_geometric.data import Data
from torch_geometric.nn import GATConv


BACKUP_DIR = os.path.join(FL_RESULT_DIR, "Backups")
os.makedirs(BACKUP_DIR, exist_ok=True)


EXPERIMENT_CONFIG = {
    "dataset": "CIC-IoT-2023",
    "dataset": "/content/drive/MyDrive/IDS dataset",
    "final_num_clients": 100,
    "final_algorithm": "FedProx",
    "final_partition": "mild_non_iid",
    "rounds": 10,
    "local_epochs": 20,
    "local_lr": LOCAL_LR,
    "fedprox_mu": FEDPROX_MU,
    "update_clip_norm": UPDATE_CLIP_NORM,
    "primary_noise_multiplier": 0.0,
    "seeds": [42, 43, 44, 45, 46],
    "scalability_client_counts": [5, 30, 100],
    "optional_large_client_counts": [20, 100],
    "privacy_noise_multipliers": [0.0, 0.005, 0.01, 0.02],
    "mild_noniid_dirichlet_alpha": 0.5,
    "strong_noniid_dirichlet_alpha": 0.1,
    "correlation_threshold": 0.90,
    "run_optional_large_scalability": False,
    "run_ablation": False,
    "run_repeated_seeds": True,
    "run_imputer_reconstruction": False,
    "run_external_bot_iot": True,
    "run_gnnexplainer": True,
    "run_formal_dp": True,
    "run_tgat_comparison": True,
    "dp_delta": 1e-5,
    "gnnexplainer_cases_per_bucket": 5,
    "gnnexplainer_epochs": 100,
}


print(json.dumps(EXPERIMENT_CONFIG, indent=2))


def save_csv(df, filename):
    path = os.path.join(FL_RESULT_DIR, filename)
    df.to_csv(path, index=False)
    print(f"Saved: {path}")
    return path

with open(os.path.join(FL_RESULT_DIR, "Experiment_Config.json"), "w") as f:
    json.dump(EXPERIMENT_CONFIG, f, indent=2)


def compute_binary_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()

    specificity = tn / (tn + fp) if (tn + fp) else 0.0
    far_fpr = fp / (fp + tn) if (fp + tn) else 0.0
    frr_fnr = fn / (fn + tp) if (fn + tp) else 0.0


    try:
        roc_auc = roc_auc_score(y_true, y_prob)
    except ValueError:
        roc_auc = np.nan
    try:
        pr_auc = average_precision_score(y_true, y_prob)
    except ValueError:
        pr_auc = np.nan

    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1": f1_score(y_true, y_pred, zero_division=0),
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Specificity": specificity,
        "FAR_FPR": far_fpr,
        "FRR_FNR": frr_fnr,
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "MCC": matthews_corrcoef(y_true, y_pred),
        "TN": int(tn), "FP": int(fp), "FN": int(fn), "TP": int(tp),
    }

def evaluate_model_metrics(model, graph, mask):
    model.eval()
    with torch.no_grad():
        out = model(graph.x, graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        logits = logits[mask]
        labels = graph.y[mask]
        probs = torch.softmax(logits, dim=1)[:, 1]
        preds = torch.argmax(logits, dim=1)

    y_true = labels.detach().cpu().numpy()
    y_pred = preds.detach().cpu().numpy()
    y_prob = probs.detach().cpu().numpy()

    metrics = compute_binary_metrics(y_true, y_pred, y_prob)
    metrics["Validation Loss"] = float(nn.CrossEntropyLoss()(logits, labels).detach().cpu())
    metrics["y_true"] = y_true
    metrics["y_pred"] = y_pred
    metrics["y_prob"] = y_prob
    return metrics


ARCHITECTURE_VERIFICATION = pd.DataFrame([
    ["GraphSAGE layer 1", "NovelTemporalClassifier.gnn1", True, "Active in final classifier"],
    ["GraphSAGE layer 2", "NovelTemporalClassifier.gnn2", True, "Active in final classifier"],
    ["TGAT", "No TGAT class/call found", False, "Do not claim TGAT in final architecture"],
    ["Feature attention", "NovelTemporalClassifier.attention", True, "Active before BiLSTM"],
    ["BiLSTM in classifier", "NovelTemporalClassifier.bilstm", True, "Executes with sequence length 1"],
    ["Temporal attention", "NovelTemporalClassifier.temporal_attention", True, "Executes over length-1 sequence"],
    ["BiLSTM imputer", "BiLSTMImputer", False, "Defined, but no completed training/evaluation found"],
], columns=["Component", "Code Path", "Active", "Verification"])
save_csv(ARCHITECTURE_VERIFICATION, "Architecture_Verification.csv")
display(ARCHITECTURE_VERIFICATION)


def partition_train_nodes(graph, num_clients, profile="iid", seed=42,
                          dirichlet_alpha=0.5, min_samples=1):
    """
    IID:
        class-balanced random allocation, so every client receives
        approximately the same class proportions.

    Mild/strong non-IID:
        class-conditional Dirichlet allocation. Lower alpha => stronger
        heterogeneity. The split is deterministic for a fixed seed.
    """
    rng = np.random.default_rng(seed)
    train_nodes = torch.where(graph.train_mask.detach().cpu())[0].numpy()
    y = graph.y.detach().cpu().numpy()

    if profile == "iid":

        client_lists = [[] for _ in range(num_clients)]
        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            for j, node in enumerate(cls_nodes):
                client_lists[j % num_clients].append(int(node))
    else:
        alpha = dirichlet_alpha
        client_lists = [[] for _ in range(num_clients)]

        for cls in [0, 1]:
            cls_nodes = train_nodes[y[train_nodes] == cls].copy()
            rng.shuffle(cls_nodes)
            proportions = rng.dirichlet(np.full(num_clients, alpha))
            counts = rng.multinomial(len(cls_nodes), proportions)

            start = 0
            for cid, count in enumerate(counts):
                client_lists[cid].extend(
                    cls_nodes[start:start+count].tolist()
                )
                start += count


        if any(len(v) < min_samples for v in client_lists):
            return partition_train_nodes(
                graph, num_clients, profile, seed + 1,
                dirichlet_alpha, min_samples
            )

    clients = []
    for cid, nodes_list in enumerate(client_lists):
        nodes_np = np.asarray(nodes_list, dtype=np.int64)
        rng.shuffle(nodes_np)
        nodes = torch.tensor(nodes_np, dtype=torch.long)

        local_edge_index, _ = subgraph(
            nodes,
            graph.edge_index.detach().cpu(),
            relabel_nodes=True,
            num_nodes=graph.num_nodes
        )

        counts = np.bincount(y[nodes_np], minlength=2)

        clients.append({
            "client_id": cid,
            "global_nodes": nodes,
            "edge_index": local_edge_index,
            "num_samples": len(nodes_np),
            "class_0": int(counts[0]),
            "class_1": int(counts[1]),
        })

    return clients

def clients_to_distribution_df(clients, profile, seed):
    rows = []
    for c in clients:
        total = max(c["num_samples"], 1)
        rows.append({
            "Seed": seed,
            "Partition": profile,
            "Client": c["client_id"],
            "Samples": c["num_samples"],
            "Benign_Count": c["class_0"],
            "Attack_Count": c["class_1"],
            "Benign_Fraction": c["class_0"] / total,
            "Attack_Fraction": c["class_1"] / total,
        })
    return pd.DataFrame(rows)

distribution_frames = []
for profile, alpha in [
    ("iid", None),
    ("mild_non_iid", EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]),
    ("strong_non_iid", EXPERIMENT_CONFIG["strong_noniid_dirichlet_alpha"]),
]:
    cset = partition_train_nodes(
        graph_data, 5, profile=profile, seed=SEED,
        dirichlet_alpha=alpha if alpha is not None else 0.5
    )
    distribution_frames.append(
        clients_to_distribution_df(cset, profile, SEED)
    )

CLIENT_DISTRIBUTION_DF = pd.concat(distribution_frames, ignore_index=True)


CURRENT_PARTITION_DIAGNOSIS = pd.DataFrame([{
    "Current_Uploaded_Partition": "random shuffled train nodes + np.array_split",
    "Class_Balance_Enforced": False,
    "Controlled_NonIID_Enforced": False,
    "Revised_Setups": "IID, mild_non_iid, strong_non_iid",
    "Conclusion": "Current setup should not be described as proper non-IID."
}])
save_csv(CURRENT_PARTITION_DIAGNOSIS, "Client_Distribution_Diagnosis.csv")
save_csv(CLIENT_DISTRIBUTION_DF, "Client_Distribution.csv")

print("\nClient-distribution verification:")
display(CLIENT_DISTRIBUTION_DF)

# ------------------------------------------------------------
# RECOMMENDATION 6 — MODEL FACTORY
# ------------------------------------------------------------
def set_all_seeds(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

def make_experiment_model():
    return NovelTemporalClassifier(
        input_dim=graph_data.num_features,
        hidden_dim=128,
        embedding_dim=64,
        num_layers=2,
        num_classes=2,
        dropout=0.30
    ).to(DEVICE)

def count_model_parameters(model):
    return sum(p.numel() for p in model.parameters())

# ------------------------------------------------------------
# RECOMMENDATION 7 — FEDERATED TRAINING ENGINE
# ------------------------------------------------------------
def train_one_client_v2(global_state, client, algorithm="FedProx",
                        local_epochs=1, lr=0.001, fedprox_mu=0.01,
                        clip_norm=1.0, noise_multiplier=0.0):
    local_model = make_experiment_model()
    local_model.load_state_dict(global_state)
    local_model.train()

    x_local, y_local, edge_local = materialize_client(
        graph_data, client
    )

    counts = torch.bincount(y_local, minlength=2).float()
    safe = torch.clamp(counts, min=1.0)
    class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    optimizer = torch.optim.AdamW(
        local_model.parameters(),
        lr=lr,
        weight_decay=WEIGHT_DECAY
    )

    global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
    start = time.perf_counter()

    for _ in range(local_epochs):
        optimizer.zero_grad(set_to_none=True)
        logits = local_model(x_local, edge_local)
        if isinstance(logits, (tuple, list)):
            logits = logits[0]

        loss = criterion(logits, y_local)

        if algorithm.lower() == "fedprox" and fedprox_mu > 0:
            prox = torch.zeros((), device=DEVICE)
            for name, param in local_model.named_parameters():
                prox += torch.sum(
                    (param - global_device[name]) ** 2
                )
            loss = loss + 0.5 * fedprox_mu * prox

        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            local_model.parameters(), max_norm=5.0
        )
        optimizer.step()

    elapsed = time.perf_counter() - start
    local_state = clone_state_dict(local_model)


    if noise_multiplier > 0:
        transmitted, raw_norm, clip_scale = clip_and_noise_update(
            local_state, global_state,
            clip_norm=clip_norm,
            noise_multiplier=noise_multiplier
        )
    else:
        transmitted = local_state
        raw_norm = state_dict_l2_distance(local_state, global_state)
        clip_scale = 1.0

    with torch.no_grad():
        local_model.eval()
        pred_logits = local_model(x_local, edge_local)
        if isinstance(pred_logits, (tuple, list)):
            pred_logits = pred_logits[0]
        pred = torch.argmax(pred_logits, dim=1)

    return {
        "state": transmitted,
        "samples": client["num_samples"],
        "loss": float(loss.detach().cpu()),
        "accuracy": accuracy_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy()
        ),
        "f1": f1_score(
            y_local.detach().cpu().numpy(),
            pred.detach().cpu().numpy(),
            zero_division=0
        ),
        "elapsed": elapsed,
        "raw_update_norm": raw_norm,
        "clip_scale": clip_scale,
    }

def run_federated_experiment(
    num_clients=5,
    algorithm="FedProx",
    partition_profile="mild_non_iid",
    rounds=10,
    local_epochs=1,
    seed=42,
    noise_multiplier=0.0,
    dirichlet_alpha=0.5,
    record_client_log=True
):
    set_all_seeds(seed)

    clients = partition_train_nodes(
        graph_data,
        num_clients=num_clients,
        profile=partition_profile,
        seed=seed,
        dirichlet_alpha=dirichlet_alpha
    )

    model = make_experiment_model()
    global_state = clone_state_dict(model)

    history = []
    client_log = []

    model_bytes = sum(
        t.numel() * t.element_size()
        for t in global_state.values()
    )

    total_communication_mb = 0.0

    for round_id in range(1, rounds + 1):
        round_start = time.perf_counter()

        local_states, local_weights = [], []
        local_losses, local_times = [], []
        update_norms, clip_scales = [], []

        for client in clients:
            result = train_one_client_v2(
                global_state=global_state,
                client=client,
                algorithm=algorithm,
                local_epochs=local_epochs,
                lr=LOCAL_LR,
                fedprox_mu=FEDPROX_MU,
                clip_norm=UPDATE_CLIP_NORM,
                noise_multiplier=noise_multiplier
            )

            local_states.append(result["state"])
            local_weights.append(result["samples"])
            local_losses.append(result["loss"])
            local_times.append(result["elapsed"])
            update_norms.append(result["raw_update_norm"])
            clip_scales.append(result["clip_scale"])

            if record_client_log:
                client_log.append({
                    "Seed": seed,
                    "Algorithm": algorithm,
                    "Partition": partition_profile,
                    "Round": round_id,
                    "Client": client["client_id"],
                    "Samples": client["num_samples"],
                    "Benign_Count": client["class_0"],
                    "Attack_Count": client["class_1"],
                    "Local_Loss": result["loss"],
                    "Local_Accuracy": result["accuracy"],
                    "Local_F1": result["f1"],
                    "Update_Norm": result["raw_update_norm"],
                    "Clip_Scale": result["clip_scale"],
                    "Local_Time_s": result["elapsed"]
                })

        global_state = aggregate_client_states(
            local_states, local_weights, global_state
        )
        model.load_state_dict(global_state)

        val = evaluate_model_metrics(
            model, graph_data, graph_data.val_mask
        )

        round_time = time.perf_counter() - round_start

        upload_mb = (model_bytes * len(clients)) / (1024**2)
        download_mb = (model_bytes * len(clients)) / (1024**2)
        comm_mb = upload_mb + download_mb
        total_communication_mb += comm_mb

        history.append({
            "Seed": seed,
            "Algorithm": algorithm,
            "Partition": partition_profile,
            "Noise_Multiplier": noise_multiplier,
            "Round": round_id,
            "Selected_Clients": len(clients),
            "Validation_Loss": val["Validation Loss"],
            "Validation_Accuracy": val["Accuracy"],
            "Validation_Precision": val["Precision"],
            "Validation_Recall": val["Recall"],
            "Validation_F1": val["F1"],
            "Mean_Local_Loss": float(np.mean(local_losses)),
            "Mean_Update_Norm": float(np.mean(update_norms)),
            "Mean_Clip_Scale": float(np.mean(clip_scales)),
            "Round_Runtime_s": round_time,
            "Avg_Local_Runtime_s": float(np.mean(local_times)),
            "Model_Size_MB": model_bytes / (1024**2),
            "Communication_MB_per_Round": comm_mb,
            "Cumulative_Communication_MB": total_communication_mb,
            "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        })

    test = evaluate_model_metrics(
        model, graph_data, graph_data.test_mask
    )

    best_round_idx = int(
        np.argmax([r["Validation_F1"] for r in history])
    )
    best_round = history[best_round_idx]["Round"]

    final_row = {
        "Seed": seed,
        "Clients": num_clients,
        "Algorithm": algorithm,
        "Partition": partition_profile,
        "Rounds": rounds,
        "Local_Epochs": local_epochs,
        "Noise_Multiplier": noise_multiplier,
        "Best_Validation_Round": best_round,
        "Accuracy": test["Accuracy"],
        "Precision": test["Precision"],
        "Recall": test["Recall"],
        "F1": test["F1"],
        "ROC-AUC": test["ROC-AUC"],
        "PR-AUC": test["PR-AUC"],
        "Specificity": test["Specificity"],
        "FAR_FPR": test["FAR_FPR"],
        "FRR_FNR": test["FRR_FNR"],
        "Balanced_Accuracy": test["Balanced Accuracy"],
        "MCC": test["MCC"],
        "TN": test["TN"], "FP": test["FP"],
        "FN": test["FN"], "TP": test["TP"],
        "Average_Round_Runtime_s": float(np.mean([r["Round_Runtime_s"] for r in history])),
        "Communication_MB_per_Round": float(np.mean([r["Communication_MB_per_Round"] for r in history])),
        "Total_Communication_MB": float(history[-1]["Cumulative_Communication_MB"]),
        "Samples_Per_Client_Mean": float(np.mean([c["num_samples"] for c in clients])),
        "Convergence_Round_ValF1": best_round,
    }

    return {
        "model": model,
        "clients": clients,
        "history": pd.DataFrame(history),
        "client_log": pd.DataFrame(client_log),
        "final": final_row,
        "test": test,
    }


FINAL_RUN = run_federated_experiment(
    num_clients=EXPERIMENT_CONFIG["final_num_clients"],
    algorithm="FedProx",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED,
    noise_multiplier=EXPERIMENT_CONFIG["primary_noise_multiplier"],
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"],
)

FINAL_FEDERATED_RESULTS = pd.DataFrame([FINAL_RUN["final"]])
save_csv(FINAL_FEDERATED_RESULTS, "Federated_Final_Test_Results.csv")
save_csv(FINAL_RUN["history"], "Federated_Round_History.csv")
save_csv(FINAL_RUN["client_log"], "Federated_Client_Log.csv")

# Exact confusion matrix CSV
final_cm = np.array([
    [FINAL_RUN["final"]["TN"], FINAL_RUN["final"]["FP"]],
    [FINAL_RUN["final"]["FN"], FINAL_RUN["final"]["TP"]]
], dtype=int)
cm_df = pd.DataFrame(
    final_cm,
    index=["Actual_Benign", "Actual_Attack"],
    columns=["Predicted_Benign", "Predicted_Attack"]
)
cm_df.to_csv(os.path.join(FL_RESULT_DIR, "Federated_Confusion_Matrix.csv"))

torch.save(
    {
        "model_state_dict": FINAL_RUN["model"].state_dict(),
        "config": EXPERIMENT_CONFIG,
        "final_metrics": FINAL_RUN["final"]
    },
    os.path.join(FL_RESULT_DIR, "Federated_Final_FedProx_Model.pt")
)

print("\nFINAL 5-CLIENT FEDPROX RESULT")
display(FINAL_FEDERATED_RESULTS)

# ------------------------------------------------------------
# RECOMMENDATION 9 — PUBLICATION-QUALITY FINAL FIGURES
# ------------------------------------------------------------
def plot_and_save(x, y, xlabel, ylabel, title, filename, marker="o"):
    plt.figure(figsize=(9, 6))
    plt.plot(x, y, marker=marker, linewidth=2)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    path = os.path.join(FL_GRAPH_DIR, filename)
    plt.savefig(path, dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
    return path

h = FINAL_RUN["history"]
plot_and_save(h["Round"], h["Validation_Loss"],
              "Federated Round", "Validation Loss",
              "FedProx Validation Loss",
              "Final_FedProx_Validation_Loss.png")
plot_and_save(h["Round"], h["Validation_Accuracy"] * 100,
              "Federated Round", "Validation Accuracy (%)",
              "FedProx Validation Accuracy",
              "Final_FedProx_Validation_Accuracy.png")
plot_and_save(h["Round"], h["Validation_F1"] * 100,
              "Federated Round", "Validation F1 (%)",
              "FedProx Validation F1",
              "Final_FedProx_Validation_F1.png")
plot_and_save(h["Round"], h["Cumulative_Communication_MB"],
              "Federated Round", "Cumulative Communication (MB)",
              "FedProx Cumulative Communication",
              "Final_FedProx_Cumulative_Communication.png")
plot_and_save(h["Round"], h["Round_Runtime_s"],
              "Federated Round", "Round Runtime (s)",
              "FedProx Round Runtime",
              "Final_FedProx_Round_Runtime.png")

# Final confusion matrix
plt.figure(figsize=(7, 6))
plt.imshow(final_cm, interpolation="nearest")
plt.colorbar()
plt.xticks([0, 1], ["Benign", "Attack"])
plt.yticks([0, 1], ["Benign", "Attack"])
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("FedProx Final Confusion Matrix")
for r in range(2):
    for c in range(2):
        plt.text(c, r, f"{final_cm[r,c]:,}",
                 ha="center", va="center", fontsize=14)
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_Confusion_Matrix.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ROC/PR curves
y_true = FINAL_RUN["test"]["y_true"]
y_pred = FINAL_RUN["test"]["y_pred"]
y_prob = FINAL_RUN["test"]["y_prob"]

try:
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(fpr, tpr, linewidth=2,
             label=f"FedProx ROC-AUC={FINAL_RUN['final']['ROC-AUC']:.4f}")
    plt.plot([0,1], [0,1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("FedProx ROC Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_ROC.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("ROC figure skipped:", e)

try:
    pr_precision, pr_recall, _ = precision_recall_curve(y_true, y_prob)
    plt.figure(figsize=(9, 6))
    plt.plot(pr_recall, pr_precision, linewidth=2,
             label=f"FedProx PR-AUC={FINAL_RUN['final']['PR-AUC']:.4f}")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("FedProx Precision-Recall Curve")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "Final_FedProx_PR.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
except Exception as e:
    print("PR figure skipped:", e)

# ------------------------------------------------------------
# RECOMMENDATION 10 — FEDAVG VS FEDPROX, SAME EVERYTHING ELSE
# ------------------------------------------------------------
# Same split, clients, model, rounds, local epochs and seed.
# Only the server/client objective differs: FedAvg mu=0 vs FedProx mu=0.01.
fedavg_run = run_federated_experiment(
    num_clients=5, algorithm="FedAvg",
    partition_profile=EXPERIMENT_CONFIG["final_partition"],
    rounds=EXPERIMENT_CONFIG["rounds"],
    local_epochs=EXPERIMENT_CONFIG["local_epochs"],
    seed=SEED, noise_multiplier=0.0,
    dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
)
fedprox_run = FINAL_RUN

fedavg_history = fedavg_run["history"].copy()
fedprox_history = fedprox_run["history"].copy()

comparison_rows = []
for _, a in fedavg_history.iterrows():
    p = fedprox_history[fedprox_history["Round"] == a["Round"]].iloc[0]
    comparison_rows.append({
        "Seed": SEED,
        "Round": int(a["Round"]),
        "FedAvg_Validation_Accuracy": a["Validation_Accuracy"],
        "FedProx_Validation_Accuracy": p["Validation_Accuracy"],
        "FedAvg_Validation_F1": a["Validation_F1"],
        "FedProx_Validation_F1": p["Validation_F1"],
        "FedAvg_Validation_Loss": a["Validation_Loss"],
        "FedProx_Validation_Loss": p["Validation_Loss"],
        "FedAvg_Update_Norm": a["Mean_Update_Norm"],
        "FedProx_Update_Norm": p["Mean_Update_Norm"],
        "FedAvg_Round_Runtime_s": a["Round_Runtime_s"],
        "FedProx_Round_Runtime_s": p["Round_Runtime_s"],
    })

comparison_rows.append({
    "Seed": SEED, "Round": "FINAL",
    "FedAvg_Accuracy": fedavg_run["final"]["Accuracy"],
    "FedProx_Accuracy": fedprox_run["final"]["Accuracy"],
    "FedAvg_Precision": fedavg_run["final"]["Precision"],
    "FedProx_Precision": fedprox_run["final"]["Precision"],
    "FedAvg_Recall": fedavg_run["final"]["Recall"],
    "FedProx_Recall": fedprox_run["final"]["Recall"],
    "FedAvg_F1": fedavg_run["final"]["F1"],
    "FedProx_F1": fedprox_run["final"]["F1"],
    "FedAvg_ROC-AUC": fedavg_run["final"]["ROC-AUC"],
    "FedProx_ROC-AUC": fedprox_run["final"]["ROC-AUC"],
    "FedAvg_Best_Validation_Round": fedavg_run["final"]["Best_Validation_Round"],
    "FedProx_Best_Validation_Round": fedprox_run["final"]["Best_Validation_Round"],
})

FEDAVG_VS_FEDPROX = pd.DataFrame(comparison_rows)
save_csv(FEDAVG_VS_FEDPROX, "FedAvg_vs_FedProx.csv")

plot_and_save(
    fedavg_history["Round"],
    fedavg_history["Validation_F1"] * 100,
    "Round", "Validation F1 (%)",
    "FedAvg vs FedProx — FedAvg",
    "FedAvg_vs_FedProx_FedAvg_F1.png"
)
plt.figure(figsize=(9,6))
plt.plot(fedavg_history["Round"], fedavg_history["Validation_F1"]*100,
         marker="o", linewidth=2, label="FedAvg")
plt.plot(fedprox_history["Round"], fedprox_history["Validation_F1"]*100,
         marker="s", linewidth=2, label="FedProx")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("FedAvg vs FedProx — Convergence")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "FedAvg_vs_FedProx_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()

# ------------------------------------------------------------
# RECOMMENDATION — TGAT-STYLE BASELINE COMPARISON
# ------------------------------------------------------------
# HONESTY NOTE: graph_data (STEP 26) has no per-edge timestamps, so a
# literal TGAT (Xu et al., 2020 — temporal-attention over timestamped
# edges) cannot be trained on it. This substitutes the uploaded model's
# spatial GraphSAGE encoder with a multi-head GATConv encoder (graph
# attention, the mechanism TGAT also builds on) while keeping every other
# component -- input dimension, BiLSTM, temporal attention, fusion head,
# classifier, optimizer, seed, split, client partition and federated
# protocol -- byte-for-byte identical to NovelTemporalClassifier / the
# 5-client FedProx run above, so the comparison isolates the effect of
# attention-based neighborhood aggregation vs. GraphSAGE's mean
# aggregation. Results are reported as "TGAT-style (GAT)", never as a
# reproduction of the published TGAT architecture.
if EXPERIMENT_CONFIG["run_tgat_comparison"]:
    print("\n=== TGAT-STYLE (GAT) VS GRAPHSAGE COMPARISON ===")

    class TGATStyleClassifier(nn.Module):
        def __init__(self, input_dim, hidden_dim=128, embedding_dim=64,
                     num_layers=2, num_classes=2, dropout=0.30, heads=4):
            super().__init__()
            self.gat1 = GATConv(
                input_dim, hidden_dim // heads, heads=heads, dropout=dropout
            )
            self.gat2 = GATConv(
                hidden_dim, embedding_dim, heads=1, concat=False, dropout=dropout
            )
            self.norm1 = nn.LayerNorm(hidden_dim)
            self.norm2 = nn.LayerNorm(embedding_dim)
            self.attention = nn.Sequential(
                nn.Linear(embedding_dim, embedding_dim // 2), nn.Tanh(),
                nn.Linear(embedding_dim // 2, 1)
            )
            self.bilstm = nn.LSTM(
                input_size=embedding_dim, hidden_size=hidden_dim,
                num_layers=num_layers, batch_first=True, bidirectional=True,
                dropout=dropout if num_layers > 1 else 0
            )
            self.temporal_attention = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.Tanh(),
                nn.Linear(hidden_dim, 1)
            )
            self.gnn_projection = nn.Linear(embedding_dim, hidden_dim * 2)
            self.fusion = nn.Sequential(
                nn.Linear(hidden_dim * 4, hidden_dim * 2),
                nn.ReLU(), nn.Dropout(dropout)
            )
            self.classifier = nn.Sequential(
                nn.Linear(hidden_dim * 2, hidden_dim), nn.ReLU(),
                nn.Dropout(dropout), nn.Linear(hidden_dim, num_classes)
            )
            self.dropout = nn.Dropout(dropout)

        def forward(self, x, edge_index=None):
            x1 = self.gat1(x, edge_index)
            x1 = F.relu(self.norm1(x1))
            x1 = self.dropout(x1)

            x2 = self.gat2(x1, edge_index)
            x2 = F.relu(self.norm2(x2))

            attn_scores = self.attention(x2)
            attn_weights = torch.softmax(attn_scores, dim=0)
            attended = x2 * attn_weights

            sequence = attended.unsqueeze(1)
            lstm_output, _ = self.bilstm(sequence)

            t_scores = self.temporal_attention(lstm_output)
            t_weights = torch.softmax(t_scores, dim=1)
            temporal_context = (lstm_output * t_weights).sum(dim=1)

            gnn_residual = self.gnn_projection(x2)
            fused = self.fusion(torch.cat([temporal_context, gnn_residual], dim=1))
            return self.classifier(fused)

    def make_tgat_model():
        return TGATStyleClassifier(
            input_dim=graph_data.num_features,
            hidden_dim=128, embedding_dim=64,
            num_layers=2, num_classes=2, dropout=0.30
        ).to(DEVICE)

    def train_one_client_generic(model_factory, global_state, client,
                                  algorithm, local_epochs, lr, fedprox_mu):
        # Mirrors train_one_client_v2 exactly, parametrized by model_factory
        # so the GraphSAGE and TGAT-style runs are trained identically in
        # every other respect (loss, class weighting, optimizer, clipping).
        local_model = model_factory()
        local_model.load_state_dict(global_state)
        local_model.train()

        x_local, y_local, edge_local = materialize_client(graph_data, client)

        counts = torch.bincount(y_local, minlength=2).float()
        safe = torch.clamp(counts, min=1.0)
        class_weights = (counts.sum() / (2.0 * safe)).to(DEVICE)
        criterion = nn.CrossEntropyLoss(weight=class_weights)

        optimizer = torch.optim.AdamW(
            local_model.parameters(), lr=lr, weight_decay=WEIGHT_DECAY
        )
        global_device = {k: v.to(DEVICE) for k, v in global_state.items()}
        start = time.perf_counter()

        for _ in range(local_epochs):
            optimizer.zero_grad(set_to_none=True)
            logits = local_model(x_local, edge_local)
            if isinstance(logits, (tuple, list)):
                logits = logits[0]
            loss = criterion(logits, y_local)

            if algorithm.lower() == "fedprox" and fedprox_mu > 0:
                prox = torch.zeros((), device=DEVICE)
                for name, param in local_model.named_parameters():
                    prox += torch.sum((param - global_device[name]) ** 2)
                loss = loss + 0.5 * fedprox_mu * prox

            loss.backward()
            torch.nn.utils.clip_grad_norm_(local_model.parameters(), max_norm=5.0)
            optimizer.step()

        elapsed = time.perf_counter() - start
        local_state = clone_state_dict(local_model)

        with torch.no_grad():
            local_model.eval()
            pred = torch.argmax(local_model(x_local, edge_local), dim=1)

        return {
            "state": local_state,
            "samples": client["num_samples"],
            "loss": float(loss.detach().cpu()),
            "elapsed": elapsed,
            "accuracy": accuracy_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy()
            ),
            "f1": f1_score(
                y_local.detach().cpu().numpy(), pred.detach().cpu().numpy(),
                zero_division=0
            ),
        }

    def run_federated_experiment_generic(model_factory, num_clients, algorithm,
                                          partition_profile, rounds, local_epochs,
                                          seed):
        set_all_seeds(seed)
        clients = partition_train_nodes(
            graph_data, num_clients=num_clients, profile=partition_profile,
            seed=seed, dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        model = model_factory()
        global_state = clone_state_dict(model)
        history = []

        for round_id in range(1, rounds + 1):
            states, weights = [], []
            for client in clients:
                result = train_one_client_generic(
                    model_factory, global_state, client,
                    algorithm=algorithm, local_epochs=local_epochs,
                    lr=LOCAL_LR, fedprox_mu=FEDPROX_MU
                )
                states.append(result["state"])
                weights.append(result["samples"])

            global_state = aggregate_client_states(states, weights, global_state)
            model.load_state_dict(global_state)

            val = evaluate_model_metrics(model, graph_data, graph_data.val_mask)
            history.append({
                "Round": round_id,
                "Validation_Accuracy": val["Accuracy"],
                "Validation_F1": val["F1"],
            })

        test = evaluate_model_metrics(model, graph_data, graph_data.test_mask)
        return {"model": model, "history": pd.DataFrame(history), "test": test}

    # Same 5-client, same rounds/local-epochs/seed/partition as the primary
    # FedProx run above, so the GraphSAGE row is a like-for-like comparison
    # rather than a re-use of the 50-client FINAL_RUN numbers.
    graphsage_5client_run = run_federated_experiment(
        num_clients=5, algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED, noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )

    tgat_run = run_federated_experiment_generic(
        make_tgat_model, num_clients=5, algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED
    )

    TGAT_COMPARISON = pd.DataFrame([
        {
            "Model": "GraphSAGE (uploaded final classifier)",
            "Clients": 5,
            "Accuracy": graphsage_5client_run["test"]["Accuracy"],
            "F1": graphsage_5client_run["test"]["F1"],
            "ROC-AUC": graphsage_5client_run["test"]["ROC-AUC"],
            "MCC": graphsage_5client_run["test"]["MCC"],
        },
        {
            "Model": "TGAT-style (GATConv, no edge timestamps)",
            "Clients": 5,
            "Accuracy": tgat_run["test"]["Accuracy"],
            "F1": tgat_run["test"]["F1"],
            "ROC-AUC": tgat_run["test"]["ROC-AUC"],
            "MCC": tgat_run["test"]["MCC"],
        },
    ])
    save_csv(TGAT_COMPARISON, "TGAT_Comparison.csv")
    display(TGAT_COMPARISON)

    plt.figure(figsize=(9, 6))
    plt.plot(graphsage_5client_run["history"]["Round"],
              graphsage_5client_run["history"]["Validation_F1"] * 100,
              marker="o", linewidth=2, label="GraphSAGE")
    plt.plot(tgat_run["history"]["Round"],
              tgat_run["history"]["Validation_F1"] * 100,
              marker="s", linewidth=2, label="TGAT-style (GAT)")
    plt.xlabel("Federated Round")
    plt.ylabel("Validation F1 (%)")
    plt.title("GraphSAGE vs TGAT-style Encoder — Convergence (5 clients)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "TGAT_vs_GraphSAGE_Convergence.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "TGAT comparison not executed"
    }]), "TGAT_Comparison.csv")

# ------------------------------------------------------------
# RECOMMENDATION 11 — SCALABILITY: ACTUALLY RUN ONLY EVALUATED COUNTS
# ------------------------------------------------------------
scalability_counts = list(EXPERIMENT_CONFIG["scalability_client_counts"])

if EXPERIMENT_CONFIG["run_optional_large_scalability"]:
    scalability_counts += EXPERIMENT_CONFIG["optional_large_client_counts"]

SCALABILITY_ROWS = []
SCALABILITY_HISTORIES = []

for n_clients in scalability_counts:
    print(f"\n=== SCALABILITY: {n_clients} CLIENTS ===")

    run = run_federated_experiment(
        num_clients=n_clients,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=0.0,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    SCALABILITY_HISTORIES.append(run["history"])

    f = run["final"]
    SCALABILITY_ROWS.append({
        "Clients": n_clients,
        "Accuracy": f["Accuracy"],
        "F1": f["F1"],
        "Average_Round_Runtime_s": f["Average_Round_Runtime_s"],
        "Communication_MB_per_Round": f["Communication_MB_per_Round"],
        "Total_Communication_MB": f["Total_Communication_MB"],
        "Samples_Per_Client_Mean": f["Samples_Per_Client_Mean"],
        "Convergence_Round": f["Convergence_Round_ValF1"],
        "Rounds_Evaluated": EXPERIMENT_CONFIG["rounds"],
        "Status": "COMPLETED"
    })

SCALABILITY_RESULTS = pd.DataFrame(SCALABILITY_ROWS)
save_csv(SCALABILITY_RESULTS, "Scalability_Experiment.csv")

# Scalability figures
for metric, ylabel, fname in [
    ("Average_Round_Runtime_s", "Average Round Runtime (s)", "Scalability_Runtime.png"),
    ("Communication_MB_per_Round", "Communication (MB/round)", "Scalability_Communication.png"),
    ("Accuracy", "Final Test Accuracy", "Scalability_Accuracy.png"),
    ("F1", "Final Test F1", "Scalability_F1.png"),
]:
    plt.figure(figsize=(9,6))
    plt.plot(SCALABILITY_RESULTS["Clients"], SCALABILITY_RESULTS[metric],
             marker="o", linewidth=2)
    plt.xlabel("Number of Clients")
    plt.ylabel(ylabel)
    plt.title(f"Scalability — {ylabel}")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, fname), dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()


plt.figure(figsize=(10,6))
for hist in SCALABILITY_HISTORIES:
    n = int(hist["Selected_Clients"].iloc[0])
    plt.plot(hist["Round"], hist["Validation_F1"]*100,
             marker="o", linewidth=2, label=f"{n} clients")
plt.xlabel("Federated Round")
plt.ylabel("Validation F1 (%)")
plt.title("Scalability — Convergence Behavior")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Scalability_Convergence.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


privacy_rows = []
for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
    print(f"\n=== GAUSSIAN UPDATE PERTURBATION: noise={noise} ===")

    run = run_federated_experiment(
        num_clients=5,
        algorithm="FedProx",
        partition_profile=EXPERIMENT_CONFIG["final_partition"],
        rounds=EXPERIMENT_CONFIG["rounds"],
        local_epochs=EXPERIMENT_CONFIG["local_epochs"],
        seed=SEED,
        noise_multiplier=noise,
        dirichlet_alpha=EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
    )
    f = run["final"]
    privacy_rows.append({
        "Noise_Multiplier": noise,
        "Accuracy": f["Accuracy"],
        "Precision": f["Precision"],
        "Recall": f["Recall"],
        "F1": f["F1"],
        "Status": "COMPLETED",
        "Interpretation": "Privacy-aware Gaussian update perturbation; NOT formal DP"
    })

PRIVACY_RESULTS = pd.DataFrame(privacy_rows)
save_csv(PRIVACY_RESULTS, "Privacy_Sensitivity_Experiment.csv")

plt.figure(figsize=(10,6))
for col, label in [
    ("Accuracy", "Accuracy"),
    ("Precision", "Precision"),
    ("Recall", "Recall"),
    ("F1", "F1"),
]:
    plt.plot(PRIVACY_RESULTS["Noise_Multiplier"],
             PRIVACY_RESULTS[col]*100,
             marker="o", linewidth=2, label=label)
plt.xlabel("Gaussian Update-Noise Multiplier")
plt.ylabel("Test Score (%)")
plt.title("Privacy-Aware Gaussian Update Perturbation Sensitivity")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(FL_GRAPH_DIR, "Privacy_Sensitivity.png"),
            dpi=400, bbox_inches="tight")
plt.show()
plt.close()


if EXPERIMENT_CONFIG["run_ablation"]:
    print("Ablation execution requires explicit variant-specific model factories.")
    print("No ablation values are generated by default.")
else:
    ablation_status = pd.DataFrame([
        ["Full Model", "NOT_RUN", "Enable run_ablation and variant factories"],
        ["Without BiLSTM / reconstruction", "NOT_RUN", "Requires controlled model variant"],
        ["Without Correlation Filtering", "NOT_RUN", "Current final input pipeline does not apply filtering"],
        ["Without Attention", "NOT_RUN", "Requires controlled model variant"],
        ["FedAvg instead of FedProx", "COMPLETED_SEPARATELY", "See FedAvg_vs_FedProx.csv"],
        ["GraphSAGE only", "NOT_RUN", "Requires controlled model variant"],
        ["GraphSAGE + BiLSTM", "NOT_RUN", "Requires controlled model variant"],
    ], columns=["Variant", "Status", "Reason"])
    save_csv(ablation_status, "Ablation_Study.csv")
    display(ablation_status)


if EXPERIMENT_CONFIG["run_imputer_reconstruction"]:
    raise NotImplementedError(
        "Enable only after adding a complete train/validation/test reconstruction "
        "protocol for BiLSTMImputer. Do not report the untrained imputer as completed."
    )
else:
    reconstruction_status = pd.DataFrame([{
        "Missing_Rate": globals().get("MISSING_RATE", 0.05),
        "Sequence_Length": globals().get("SEQUENCE_LENGTH", 10),
        "RMSE": np.nan,
        "MAE": np.nan,
        "Status": "NOT_COMPLETED",
        "Reason": "BiLSTMImputer is defined but no completed training/evaluation was found"
    }])
    save_csv(reconstruction_status, "BiLSTM_Reconstruction_Status.csv")


graphfedai_available = any(
    name.lower() == "graphfedai" or "graphfedai" in name.lower()
    for name in globals().keys()
)

if graphfedai_available:
    print("GraphFedAI symbol found in runtime; implement the exact baseline adapter before running.")
else:
    baseline_status = pd.DataFrame([{
        "Baseline": "GraphFedAI",
        "Status": "NOT_COMPLETED",
        "Reason": "No GraphFedAI implementation is present in the uploaded notebook",
        "Required_Settings": json.dumps({
            "dataset_split": "same as EFG-IDS",
            "clients": 5,
            "rounds": EXPERIMENT_CONFIG["rounds"],
            "local_epochs": EXPERIMENT_CONFIG["local_epochs"],
            "seed": SEED
        })
    }])
    save_csv(baseline_status, "GraphFedAI_Baseline_Comparison.csv")
    display(baseline_status)


if EXPERIMENT_CONFIG["run_repeated_seeds"]:
    repeated_rows = []
    for seed in EXPERIMENT_CONFIG["seeds"]:
        # Final FedProx
        rp = run_federated_experiment(
            5, "FedProx", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        # FedAvg
        ra = run_federated_experiment(
            5, "FedAvg", EXPERIMENT_CONFIG["final_partition"],
            EXPERIMENT_CONFIG["rounds"], EXPERIMENT_CONFIG["local_epochs"],
            seed, 0.0, EXPERIMENT_CONFIG["mild_noniid_dirichlet_alpha"]
        )
        repeated_rows.extend([
            {"Seed": seed, "Experiment": "FedProx", **rp["final"]},
            {"Seed": seed, "Experiment": "FedAvg", **ra["final"]},
        ])

    repeated_raw = pd.DataFrame(repeated_rows)
    save_csv(repeated_raw, "Repeated_Run_Raw.csv")

    summary = (
        repeated_raw
        .groupby("Experiment")
        .agg(
            Accuracy_mean=("Accuracy", "mean"),
            Accuracy_std=("Accuracy", "std"),
            Precision_mean=("Precision", "mean"),
            Precision_std=("Precision", "std"),
            Recall_mean=("Recall", "mean"),
            Recall_std=("Recall", "std"),
            F1_mean=("F1", "mean"),
            F1_std=("F1", "std"),
            ROCAUC_mean=("ROC-AUC", "mean"),
            ROCAUC_std=("ROC-AUC", "std"),
        )
        .reset_index()
    )
    save_csv(summary, "Repeated_Run_Summary.csv")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Seeds": "42,43,44,45,46",
        "Reason": "Repeated runs disabled by default because of computational cost"
    }]), "Repeated_Run_Summary.csv")

if EXPERIMENT_CONFIG["run_external_bot_iot"]:
    print("\n=== EXTERNAL VALIDATION: BoT-IoT / IDS dataset ===")

    bot_iot_files = sorted(IDS_PATH.glob("*.csv"))
    if not bot_iot_files:
        raise FileNotFoundError(
            f"No CSV files found under {IDS_PATH}; cannot run BoT-IoT "
            "external validation."
        )

    bot_frames = []
    for f in bot_iot_files:
        try:
            bot_frames.append(pd.read_csv(f, low_memory=False))
        except Exception as e:
            print(f"  Skipping {f.name}: {e}")

    BOT_RAW = pd.concat(bot_frames, ignore_index=True)

    bot_label_candidates = [
        c for c in BOT_RAW.columns
        if c.lower() in ("label", "attack", "category", "class")
    ]
    if not bot_label_candidates:
        raise KeyError("Could not locate a label column in the BoT-IoT/IDS files.")
    bot_label_col = bot_label_candidates[0]

    # Reuse the exact binary-label convention from CELL 12/13
    # (create_binary_labels) so "benign" is mapped to 0 identically to the
    # CIC-IoT-2023 training labels.
    BOT_RAW["binary_label"] = create_binary_labels(BOT_RAW[bot_label_col])

    # Feature alignment onto the trained CIC-IoT-2023 feature space:
    # - Columns the scaler was fit on and that BoT-IoT also has are used
    #   directly (matched_features).
    # - Columns the scaler was fit on but BoT-IoT does NOT have are filled
    #   with the TRAINING median (train_median from STEP 22) -- never a
    #   BoT-IoT statistic -- so no target-domain information leaks into a
    #   feature the model was never actually shown from this domain.
    aligned = pd.DataFrame(index=BOT_RAW.index)
    matched_features, filled_features = [], []
    for col in FEATURE_COLUMNS:
        if col in BOT_RAW.columns:
            aligned[col] = pd.to_numeric(BOT_RAW[col], errors="coerce")
            matched_features.append(col)
        else:
            aligned[col] = train_median[col]
            filled_features.append(col)

    aligned = aligned.replace([np.inf, -np.inf], np.nan)
    aligned = aligned.fillna(train_median)

    BOT_X = scaler.transform(aligned[FEATURE_COLUMNS])
    BOT_y = BOT_RAW["binary_label"].astype(int).to_numpy()

    print(f"BoT-IoT rows: {len(BOT_X)}")
    print(f"Matched CIC-IoT-2023 features: {len(matched_features)} / {len(FEATURE_COLUMNS)}")
    print(f"Features filled with training median (not present in BoT-IoT): {len(filled_features)}")
    if filled_features:
        print("  ", filled_features)

    # Build a self-contained kNN graph over the BoT-IoT rows only (same k
    # and metric as STEP 26), so the trained GraphSAGE layers receive the
    # same kind of local-neighborhood structure they were trained on. This
    # graph is independent of graph_data -- no BoT-IoT node is ever
    # connected to a CIC-IoT-2023 node.
    bot_knn = NearestNeighbors(n_neighbors=K_NEIGHBORS + 1, metric="euclidean", n_jobs=-1)
    bot_knn.fit(BOT_X)
    _, bot_neighbors = bot_knn.kneighbors(BOT_X)

    bot_src, bot_dst = [], []
    for i in range(len(BOT_X)):
        for j in bot_neighbors[i, 1:]:
            bot_src.append(i)
            bot_dst.append(j)
    bot_edge_index = torch.tensor([bot_src, bot_dst], dtype=torch.long)

    bot_graph = Data(
        x=torch.tensor(BOT_X, dtype=torch.float32),
        y=torch.tensor(BOT_y, dtype=torch.long),
        edge_index=bot_edge_index
    ).to(DEVICE)

    external_model = FINAL_RUN["model"]
    external_model.eval()
    with torch.no_grad():
        out = external_model(bot_graph.x, bot_graph.edge_index)
        logits = out[0] if isinstance(out, (tuple, list)) else out
        bot_probs = torch.softmax(logits, dim=1)[:, 1]
        bot_preds = torch.argmax(logits, dim=1)

    bot_metrics = compute_binary_metrics(
        bot_graph.y.detach().cpu().numpy(),
        bot_preds.detach().cpu().numpy(),
        bot_probs.detach().cpu().numpy()
    )

    EXTERNAL_BOTIOT_RESULTS = pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "COMPLETED",
        "Rows_Evaluated": len(BOT_X),
        "Matched_Features": len(matched_features),
        "Filled_Features": len(filled_features),
        "K_Neighbors": K_NEIGHBORS,
        "Model_Source": "FINAL_RUN (100-client FedProx, trained on CIC-IoT-2023 only)",
        "Accuracy": bot_metrics["Accuracy"],
        "Precision": bot_metrics["Precision"],
        "Recall": bot_metrics["Recall"],
        "F1": bot_metrics["F1"],
        "ROC-AUC": bot_metrics["ROC-AUC"],
        "PR-AUC": bot_metrics["PR-AUC"],
        "Specificity": bot_metrics["Specificity"],
        "Balanced_Accuracy": bot_metrics["Balanced Accuracy"],
        "MCC": bot_metrics["MCC"],
        "TN": bot_metrics["TN"], "FP": bot_metrics["FP"],
        "FN": bot_metrics["FN"], "TP": bot_metrics["TP"],
    }])
    save_csv(EXTERNAL_BOTIOT_RESULTS, "External_Validation_BoTIoT.csv")
    display(EXTERNAL_BOTIOT_RESULTS)

    bot_cm = np.array([
        [bot_metrics["TN"], bot_metrics["FP"]],
        [bot_metrics["FN"], bot_metrics["TP"]]
    ])
    plt.figure(figsize=(7, 6))
    plt.imshow(bot_cm, interpolation="nearest")
    plt.colorbar()
    plt.xticks([0, 1], ["Benign", "Attack"])
    plt.yticks([0, 1], ["Benign", "Attack"])
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.title("BoT-IoT External Validation — Confusion Matrix")
    for r in range(2):
        for c in range(2):
            plt.text(c, r, f"{bot_cm[r, c]:,}", ha="center", va="center", fontsize=14)
    plt.tight_layout()
    plt.savefig(os.path.join(FL_GRAPH_DIR, "External_BoTIoT_Confusion_Matrix.png"),
                dpi=400, bbox_inches="tight")
    plt.show()
    plt.close()
else:
    save_csv(pd.DataFrame([{
        "Dataset": "BoT-IoT",
        "Status": "NOT_COMPLETED",
        "Reason": "External feature mapping/evaluation not executed"
    }]), "External_Validation_BoTIoT.csv")


if EXPERIMENT_CONFIG["run_gnnexplainer"]:
    print("\n=== GNNEXPLAINER: TP / TN / FP / FN CASE EXPLANATIONS ===")
    from torch_geometric.explain import Explainer, GNNExplainer

    explain_model = FINAL_RUN["model"]
    explain_model.eval()

    test_idx = torch.where(graph_data.test_mask)[0]
    y_true_test = graph_data.y[test_idx].detach().cpu().numpy()
    y_pred_test = FINAL_RUN["test"]["y_pred"]  # same order as test_idx (mask indexing)

    tp_idx = test_idx[(y_true_test == 1) & (y_pred_test == 1)]
    tn_idx = test_idx[(y_true_test == 0) & (y_pred_test == 0)]
    fp_idx = test_idx[(y_true_test == 0) & (y_pred_test == 1)]
    fn_idx = test_idx[(y_true_test == 1) & (y_pred_test == 0)]

    explain_rng = np.random.default_rng(SEED)

    def sample_case_nodes(pool, n):
        pool = pool.detach().cpu().numpy()
        if len(pool) == 0:
            return np.array([], dtype=np.int64)
        return explain_rng.choice(pool, size=min(n, len(pool)), replace=False)

    n_per_bucket = EXPERIMENT_CONFIG["gnnexplainer_cases_per_bucket"]
    case_nodes = {
        "TP": sample_case_nodes(tp_idx, n_per_bucket),
        "TN": sample_case_nodes(tn_idx, n_per_bucket),
        "FP": sample_case_nodes(fp_idx, n_per_bucket),
        "FN": sample_case_nodes(fn_idx, n_per_bucket),
    }
    for case, nodes in case_nodes.items():
        print(f"  {case}: {len(nodes)} case(s) sampled (pool size available)")

    # NovelTemporalClassifier.forward(x, edge_index) returns raw per-node
    # logits directly (not a tuple), matching Explainer's expected model
    # output for node-level multiclass classification with return_type="raw".
    explainer = Explainer(
        model=explain_model,
        algorithm=GNNExplainer(epochs=EXPERIMENT_CONFIG["gnnexplainer_epochs"]),
        explanation_type="model",
        node_mask_type="attributes",
        edge_mask_type="object",
        model_config=dict(
            mode="multiclass_classification",
            task_level="node",
            return_type="raw",
        ),
    )

    explain_x = graph_data.x.to(DEVICE)
    explain_edge_index = graph_data.edge_index.to(DEVICE)

    explain_rows = []
    for case, nodes in case_nodes.items():
        for node_id in nodes:
            node_id = int(node_id)
            explanation = explainer(
                x=explain_x, edge_index=explain_edge_index, index=node_id
            )
            feat_importance = explanation.node_mask[node_id].detach().cpu().numpy()
            top_k = np.argsort(-feat_importance)[:10]
            for rank, feat_idx in enumerate(top_k, start=1):
                feat_name = (
                    FEATURE_COLUMNS[feat_idx]
                    if feat_idx < len(FEATURE_COLUMNS)
                    else f"feature_{feat_idx}"
                )
                explain_rows.append({
                    "Case": case,
                    "Node": node_id,
                    "Rank": rank,
                    "Feature": feat_name,
                    "Importance": float(feat_importance[feat_idx]),
                })

    GNNEXPLAINER_RESULTS = pd.DataFrame(explain_rows)
    save_csv(GNNEXPLAINER_RESULTS, "GNNExplainer_Results.csv")
    display(GNNEXPLAINER_RESULTS.head(20))

    if len(GNNEXPLAINER_RESULTS):
        top_features = (
            GNNEXPLAINER_RESULTS.groupby("Feature")["Importance"]
            .mean().sort_values(ascending=False).head(15)
        )
        plt.figure(figsize=(10, 6))
        top_features.plot(kind="barh")
        plt.gca().invert_yaxis()
        plt.xlabel("Mean GNNExplainer Feature Importance")
        plt.title("Top Features — GNNExplainer (TP/TN/FP/FN cases)")
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "GNNExplainer_Top_Features.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
    else:
        print("No test-set cases were available in one or more of TP/TN/FP/FN; "
              "GNNExplainer_Results.csv was written empty.")
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Cases": "TP,TN,FP,FN",
        "Reason": "GNNExplainer not executed"
    }]), "GNNExplainer_Results.csv")

# ------------------------------------------------------------
# RECOMMENDATION 19 — FORMAL DIFFERENTIAL PRIVACY
# ------------------------------------------------------------
if EXPERIMENT_CONFIG["run_formal_dp"]:
    print("\n=== FORMAL DIFFERENTIAL PRIVACY ACCOUNTING ===")
    print("Mechanism being accounted: clip_and_noise_update (FL-3) clips each")
    print("client's per-round update to L2 norm <= update_clip_norm and adds")
    print("N(0, (noise_multiplier * update_clip_norm)^2) noise coordinatewise")
    print("before it leaves the client. This is a Gaussian mechanism with L2")
    print("sensitivity = update_clip_norm and noise multiplier = noise_multiplier,")
    print("applied once per client per federated round under full participation")
    print("(client sampling probability q = 1, as configured in FL-1/FL-2).")
    print("The epsilon below is therefore a CLIENT-LEVEL DP guarantee on the")
    print("transmitted update sequence, accounted analytically via Renyi DP")
    print("(RDP) composition -- it is not a sample-level guarantee over the")
    print("raw per-record training data, since no per-record subsampling or")
    print("per-record gradient clipping is performed anywhere in this pipeline.")

    def gaussian_mechanism_rdp(alpha, noise_multiplier):
        # Renyi divergence of order alpha between N(0, sigma^2) and
        # N(1, sigma^2) for a mechanism with L2 sensitivity 1 and Gaussian
        # noise of std = noise_multiplier (Mironov, 2017, Prop. 7):
        #   RDP(alpha) = alpha / (2 * sigma^2)
        return alpha / (2.0 * (noise_multiplier ** 2))

    def rdp_to_eps(rdp_orders, rdp_values, delta):
        # Standard RDP -> (epsilon, delta)-DP conversion (Balle et al. 2020 /
        # Mironov 2017), minimized over the candidate Renyi orders.
        candidates = [
            rdp + math.log(1.0 / delta) / (alpha - 1.0)
            for alpha, rdp in zip(rdp_orders, rdp_values)
            if alpha > 1.0
        ]
        return min(candidates) if candidates else float("inf")

    RDP_ORDERS = [1 + x / 10.0 for x in range(1, 100)] + list(range(12, 64))
    DP_DELTA = EXPERIMENT_CONFIG["dp_delta"]
    # One Gaussian mechanism application per client per round; composition
    # count uses the primary experiment's round budget.
    NUM_COMPOSITIONS = EXPERIMENT_CONFIG["rounds"]

    dp_rows = []
    for noise in EXPERIMENT_CONFIG["privacy_noise_multipliers"]:
        if noise <= 0:
            dp_rows.append({
                "Noise_Multiplier": noise,
                "Rounds_Composed": NUM_COMPOSITIONS,
                "Delta": DP_DELTA,
                "Epsilon": float("inf"),
                "Note": "No noise added: the mechanism provides no DP guarantee "
                        "(epsilon is undefined/infinite).",
            })
            continue

        rdp_values = [
            NUM_COMPOSITIONS * gaussian_mechanism_rdp(alpha, noise)
            for alpha in RDP_ORDERS
        ]
        epsilon = rdp_to_eps(RDP_ORDERS, rdp_values, DP_DELTA)

        dp_rows.append({
            "Noise_Multiplier": noise,
            "Rounds_Composed": NUM_COMPOSITIONS,
            "Delta": DP_DELTA,
            "Epsilon": epsilon,
            "Note": "Client-level (epsilon, delta)-DP for the transmitted update "
                    "via RDP composition of the Gaussian mechanism (analytic "
                    "moments accountant; sensitivity = update_clip_norm; q = 1).",
        })

    FORMAL_DP_RESULTS = pd.DataFrame(dp_rows)
    save_csv(FORMAL_DP_RESULTS, "Formal_Differential_Privacy.csv")
    display(FORMAL_DP_RESULTS)

    finite_dp = FORMAL_DP_RESULTS[np.isfinite(FORMAL_DP_RESULTS["Epsilon"])]
    if len(finite_dp):
        plt.figure(figsize=(9, 6))
        plt.plot(finite_dp["Noise_Multiplier"], finite_dp["Epsilon"],
                  marker="o", linewidth=2)
        plt.xlabel("Gaussian Noise Multiplier")
        plt.ylabel(f"Epsilon (delta={DP_DELTA:g})")
        plt.title(f"Formal Client-Level DP: Epsilon vs Noise Multiplier "
                  f"over {NUM_COMPOSITIONS} Rounds")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig(os.path.join(FL_GRAPH_DIR, "Formal_DP_Epsilon.png"),
                    dpi=400, bbox_inches="tight")
        plt.show()
        plt.close()
else:
    save_csv(pd.DataFrame([{
        "Status": "NOT_COMPLETED",
        "Reason": "Formal DP accounting not executed"
    }]), "Formal_Differential_Privacy.csv")

# ------------------------------------------------------------
# RECOMMENDATION 20 — EXPERIMENT MANIFEST / LOG EVIDENCE
# ------------------------------------------------------------
completed_in_this_cell = [
    "5-client FedProx final run",
    "IID/mild-non-IID/strong-non-IID client distribution verification",
    "FedAvg vs FedProx controlled comparison",
    "2/5/10-client scalability",
    "0/0.005/0.01/0.02 Gaussian update perturbation sensitivity",
]
not_claimed_as_completed = [
    "GraphFedAI rerun (implementation absent)",
    "BiLSTM reconstruction RMSE/MAE",
    "Full controlled ablation unless enabled",
    "20/100-client scalability unless explicitly enabled and executed",
]

# Each of these five now has a real, executable implementation above (not a
# NotImplementedError stub); whether it actually ran -- and therefore which
# list it belongs in -- is read from the same EXPERIMENT_CONFIG flags that
# gated its code block, so the manifest can never claim a result that the
# flags say was skipped.
conditionally_completed = {
    "run_repeated_seeds": "Repeated seeds (5-seed FedProx/FedAvg mean +/- std)",
    "run_external_bot_iot": "BoT-IoT external validation",
    "run_gnnexplainer": "GNNExplainer (TP/TN/FP/FN feature attributions)",
    "run_formal_dp": "Formal client-level Differential Privacy (RDP accountant)",
    "run_tgat_comparison": "TGAT-style (GATConv) vs GraphSAGE comparison",
}
for flag, label in conditionally_completed.items():
    if EXPERIMENT_CONFIG.get(flag, False):
        completed_in_this_cell.append(label)
    else:
        not_claimed_as_completed.append(
            label + " (disabled: EXPERIMENT_CONFIG['" + flag + "'] is False)"
        )

manifest = {
    "completed_in_this_cell": completed_in_this_cell,
    "not_claimed_as_completed": not_claimed_as_completed,
    "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
}
with open(os.path.join(FL_RESULT_DIR, "Experiment_Manifest.json"), "w") as f:
    json.dump(manifest, f, indent=2)

print("\n" + "="*100)
print("MASTER EXPERIMENT SUITE FINISHED")
print("="*100)
print("Results:", FL_RESULT_DIR)
print("Graphs :", FL_GRAPH_DIR)
print("No unexecuted result has been represented as a completed measurement.")
```
